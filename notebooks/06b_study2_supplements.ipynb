## 1. Settings

- **Question:** Which run, which gate, which test settings does this notebook
  work with?
- **Expectation:** the same values as notebook 06, so that a difference in a
  result can only come from the embeddings.
- **Status:** added on 2026-10-18 after the Arm 1 results, same specification as Arm 1.
- **Unit and test:** none, settings only.
- **Output:** none.

```python
# Settings of the supplements. Every value is the one of notebook 06.
from pathlib import Path as _Path
import os as _os
if _Path.cwd().name == "notebooks":
    _os.chdir("..")
_os.environ["HF_HUB_OFFLINE"] = "1"
_os.environ["TRANSFORMERS_OFFLINE"] = "1"
del _Path, _os

RUN_TAG        = "run5"
PRIMARY_GATE   = 0.90       # content-preservation gate of the main analysis
N_PERM, SEED   = 5000, 42   # permutations and seed of every test
JD_REPR        = "mean"     # representation of a job description
LEX_MAIN       = "A"        # dictionary of the main analysis
ALPHA_PRE      = 0.025      # the two tests fixed in advance, 0.05 split

KEPT_13 = [
    "nurse", "software_engineer", "accountant", "architect", "psychologist",
    "attorney", "teacher", "surgeon", "physician", "professor", "dentist",
    "paralegal", "dietitian",
]
ARMS = ("communion", "agency")

OUT_BASE = f"outputs/study2_{RUN_TAG}"
ARM2_DIR = f"{OUT_BASE}/arm2"

# The gate diagnosis of section 9 reads the borderline cases from the protocol.
GATE_LOW, GATE_HIGH = 0.85, 0.95
N_BORDERLINE = 20           # per arm, drawn with the seed above
CONTEXT_TOKENS = 8          # tokens before and after a changed place

print(f"gate {PRIMARY_GATE}  permutations {N_PERM}  seed {SEED}  "
      f"dictionary {LEX_MAIN}")
print(f"occupations {len(KEPT_13)}  arms {ARMS}")
print(f"outputs go to {OUT_BASE}/")
```

## 2. Inputs

- **Question:** Are the embeddings of Arm 2 and the tables of notebook 06 all
  present, and do they describe the same sample?
- **Expectation:** every biography and every advertisement of the pool appears
  in both, and the embeddings are the ones the metadata describes.
- **Status:** added on 2026-10-18 after the Arm 1 results, same specification as Arm 1.
- **Unit and test:** control, no statistic.
- **Output:** none.

```python
# Libraries, the shared study module, and the saved inputs.
import json
import numpy as np
import pandas as pd
import sys
from pathlib import Path
from scipy.stats import spearmanr
import statsmodels.formula.api as smf

sys.path.insert(0, str(Path.cwd()))
import importlib.util
_spec = importlib.util.spec_from_file_location("study2_core",
                                               Path("src/study2_core.py"))
study2_core = importlib.util.module_from_spec(_spec)
_spec.loader.exec_module(study2_core)

ARM2_BIO = Path(ARM2_DIR) / "arm2_bio_embeddings.parquet"
ARM2_JD = Path(ARM2_DIR) / "arm2_jd_embeddings.parquet"
ARM2_META = Path(ARM2_DIR) / "arm2_metadata.json"
POOL_FILE = Path(OUT_BASE) / "study2_delta_r5_pool.csv"
JD_LEVEL_FILE = Path(OUT_BASE) / "study2_jd_level.csv"
JD_EDA_FILE = Path(OUT_BASE) / "study2_jd_eda.csv"
BIO_SLOPES_FILE = Path(OUT_BASE) / "study2_fit_bio_slopes.csv"
CP_GATE_FILE = Path(OUT_BASE) / "study2_cp_gate.csv"
PROTOCOL_FILE = Path(OUT_BASE) / "substitution_log_v2.csv"
ARM2_TESTS_FILE = Path(OUT_BASE) / "study2_arm2_tests.csv"
ARM2_FIT_FILE = Path(OUT_BASE) / "study2_arm2_fit_tests.csv"

for _path in (ARM2_BIO, ARM2_JD, ARM2_META, POOL_FILE, JD_LEVEL_FILE,
              JD_EDA_FILE, BIO_SLOPES_FILE, CP_GATE_FILE, PROTOCOL_FILE,
              ARM2_TESTS_FILE):
    if not _path.exists():
        raise FileNotFoundError(
            f"{_path} not found. This notebook reads the outputs of the full "
            f"run of notebook 06 and the embeddings of notebook 06a.")

meta_arm2 = json.loads(ARM2_META.read_text(encoding="utf-8"))
print(f"model      : {meta_arm2.get('model')}")
print(f"revision   : {meta_arm2.get('revision')}")
print(f"dimension  : {meta_arm2.get('embedding_dim')}  normalised: "
      f"{meta_arm2.get('normalised')}")
print(f"truncated  : {meta_arm2.get('n_truncated')} (expected 0)")
if meta_arm2.get("n_truncated"):
    raise AssertionError(
        f"{meta_arm2['n_truncated']} texts were truncated in Arm 2; the "
        f"comparison needs whole texts.")

df_pool = pd.read_csv(POOL_FILE)
df_jd_level = pd.read_csv(JD_LEVEL_FILE)
df_jd_eda = pd.read_csv(JD_EDA_FILE)
df_bio = pd.read_csv(BIO_SLOPES_FILE)
print()
print(f"pool records of notebook 06 : {len(df_pool)}")
print(f"advertisements              : {len(df_jd_level)}")
print(f"biographies with a balance  : {len(df_bio)}")

_bio_emb_raw = pd.read_parquet(ARM2_BIO)
_jd_emb_raw = pd.read_parquet(ARM2_JD)
print(f"embeddings of Arm 2         : {len(_bio_emb_raw)} biography versions, "
      f"{len(_jd_emb_raw)} advertisements")

# The embeddings are read into one matrix per occupation, in the order of the
# pool: row i is biography i, column j is advertisement j.
emb_orig, emb_manip, emb_jd = {}, {}, {}
_originals = _bio_emb_raw[_bio_emb_raw["version"] == "original"]
for occ in KEPT_13:
    _rows = _originals[_originals["occ"] == occ].sort_values("bio_idx")
    if list(_rows["bio_idx"]) != list(range(len(_rows))):
        raise AssertionError(
            f"{occ}: the biographies of Arm 2 are not numbered without a gap.")
    emb_orig[occ] = np.vstack([np.asarray(e) for e in _rows["embedding"]])
    _jd_rows = _jd_emb_raw[_jd_emb_raw["occ"] == occ].sort_values("jd_idx")
    if list(_jd_rows["jd_idx"]) != list(range(len(_jd_rows))):
        raise AssertionError(
            f"{occ}: the advertisements of Arm 2 are not numbered without a "
            f"gap.")
    emb_jd[occ] = np.vstack([np.asarray(e) for e in _jd_rows["embedding"]])
    emb_manip[occ] = {}
    for arm in ARMS:
        _manip_rows = (_bio_emb_raw[(_bio_emb_raw["occ"] == occ)
                                    & (_bio_emb_raw["version"] == arm)]
                       .set_index("bio_idx"))
        # A biography the manipulation did not change carries the text of the
        # original, so Arm 2 embedded it once; its row is the original row.
        _matrix = emb_orig[occ].copy()
        for _bio_idx, _row in _manip_rows.iterrows():
            _matrix[int(_bio_idx)] = np.asarray(_row["embedding"])
        emb_manip[occ][arm] = _matrix

print()
print(f"{'occ':<20} {'bios':>6} {'ads':>6} {'changed communion':>18} "
      f"{'changed agency':>15}")
for occ in KEPT_13:
    _changed = {arm: int(((_bio_emb_raw["occ"] == occ)
                          & (_bio_emb_raw["version"] == arm)).sum())
                for arm in ARMS}
    print(f"{occ:<20} {emb_orig[occ].shape[0]:>6} {emb_jd[occ].shape[0]:>6} "
          f"{_changed['communion']:>18} {_changed['agency']:>15}")

# The two samples have to cover the same biographies and advertisements.
_mismatch = []
for occ in KEPT_13:
    _n_bio_06 = int(((df_pool["occ"] == occ)
                     & (df_pool["arm"] == "communion")).sum())
    _n_jd_06 = int((df_jd_level["occ"] == occ).sum())
    if emb_orig[occ].shape[0] != _n_bio_06 or emb_jd[occ].shape[0] != _n_jd_06:
        _mismatch.append((occ, emb_orig[occ].shape[0], _n_bio_06,
                          emb_jd[occ].shape[0], _n_jd_06))
print()
print(f"control occupations whose sample differs from notebook 06: "
      f"{len(_mismatch)} (expected 0)")
if _mismatch:
    raise AssertionError(
        f"The embeddings of Arm 2 do not cover the pool of notebook 06: "
        f"{_mismatch[:3]}")
```

## 3. The change per advertisement

- **Question:** How far does the manipulation move a biography against each
  single advertisement, measured with the embeddings of Arm 2?
- **Expectation:** none; this section produces the quantity that the tests
  below read.
- **Status:** added on 2026-10-18 after the Arm 1 results, same specification as Arm 1.
- **Unit and test:** biography and advertisement; no test, the change in
  percentile rank per advertisement, averaged over the changed biographies of
  the gate.
- **Output:** `study2_arm2_jd_level.csv`.

```python
# The change per advertisement, formed exactly as in notebook 06: the rank of
# a biography among the others for that advertisement, before and after the
# manipulation, with only that one biography changed.
_gate_rows = df_pool[df_pool["sim_cp"] >= PRIMARY_GATE]
_changed_rows = _gate_rows[_gate_rows["n_subs"] >= 1]
print(f"pool records at the gate of {PRIMARY_GATE}: {len(_gate_rows)}")
print(f"of those changed by the manipulation      : {len(_changed_rows)}")
print(f"  per arm: {dict(_changed_rows['arm'].value_counts())}")

delta_j_arm2 = {}          # (occ, arm) -> change per advertisement
delta_bio_arm2 = []        # one row per biography and arm
for occ in KEPT_13:
    _so = emb_orig[occ] @ emb_jd[occ].T
    _ppo = study2_core.pool_pct_per_jd(_so)
    for arm in ARMS:
        _sm = emb_manip[occ][arm] @ emb_jd[occ].T
        _ppm = study2_core.indiv_cf_pct_per_jd(_so, _sm)
        _per_bio = _ppm - _ppo
        _sub = _changed_rows[(_changed_rows["occ"] == occ)
                             & (_changed_rows["arm"] == arm)]
        _idx = _sub["bio_idx"].to_numpy(int)
        if len(_idx) < 5:
            raise AssertionError(
                f"{occ}/{arm}: {len(_idx)} changed biographies at the gate, "
                f"at least 5 are needed.")
        delta_j_arm2[(occ, arm)] = _per_bio[_idx].mean(axis=0)
        for _i in _idx:
            delta_bio_arm2.append({"occ": occ, "bio_idx": int(_i), "arm": arm,
                                   "delta": float(_per_bio[_i].mean())})
df_arm2_delta_bio = pd.DataFrame(delta_bio_arm2)
print()
print(f"biographies with a change per advertisement: {len(df_arm2_delta_bio)}")
print(f"advertisement vectors formed: {len(delta_j_arm2)} "
      f"(13 occupations x 2 arms)")

# The table per advertisement, with the columns of study2_jd_level.csv.
_rows = []
for occ in KEPT_13:
    _level = df_jd_level[df_jd_level["occ"] == occ].sort_values("jd_idx")
    for _position, (_, _row) in enumerate(_level.iterrows()):
        _entry = {"occ": occ, "jd_idx": int(_row["jd_idx"]),
                  "comm_dens": float(_row["comm_dens"]),
                  "ag_dens": float(_row["ag_dens"]),
                  "balance": float(_row["balance"]),
                  "comm_dens_cw": float(_row["comm_dens_cw"]),
                  "ag_dens_cw": float(_row["ag_dens_cw"]),
                  "balance_cw": float(_row["balance_cw"]),
                  "balance_fw": float(_row["balance_fw"]),
                  "balance_AB": float(_row["balance_AB"]),
                  "n_words": float(_row["n_words"]),
                  "n_tokens": float(_row["n_tokens"])}
        for arm in ARMS:
            _entry[f"dj_pool_{arm}_rank"] = float(delta_j_arm2[(occ, arm)][_position])
        _rows.append(_entry)
df_arm2_jd_level = pd.DataFrame(_rows)
print(f"rows of the advertisement table: {len(df_arm2_jd_level)}")
print(df_arm2_jd_level.groupby("occ")[["dj_pool_communion_rank",
                                       "dj_pool_agency_rank"]].mean().round(3)
      .to_string())
```

## 4. Control against notebook 06

- **Question:** Does this notebook reproduce the numbers that notebook 06 and
  the script of 2026-10-13 already report for Arm 2?
- **Expectation:** the overall effect of section 15 and the two tests fixed in
  advance come out identical; a difference means the computation here is not
  the one of notebook 06.
- **Status:** added on 2026-10-18 after the Arm 1 results, same specification as Arm 1.
- **Unit and test:** control; the run stops on any deviation.
- **Output:** none.

```python
# The overall effect of Arm 2, as notebook 06 section 15 reports it. The change
# per biography is the mean over the advertisements, which is what that section
# tests.
_arm2_tests = pd.read_csv(ARM2_TESTS_FILE)
print(f"{'arm':<10} {'n here':>7} {'n there':>8} {'median here':>12} "
      f"{'median there':>13} {'up here':>8} {'up there':>9} {'p here':>11} "
      f"{'p there':>11}")
print("-" * 96)
_worst = 0.0
for arm in ARMS:
    _rows = []
    for occ in KEPT_13:
        _so = emb_orig[occ] @ emb_jd[occ].T
        _sm = emb_manip[occ][arm] @ emb_jd[occ].T
        _delta = (study2_core.individual_cf_ranks(_so, _sm)
                  - study2_core.percentile_ranks(_so))
        _sub = df_pool[(df_pool["occ"] == occ) & (df_pool["arm"] == arm)
                       & (df_pool["sim_cp"] >= PRIMARY_GATE)]
        _rows.extend(_delta[_sub["bio_idx"].to_numpy(int)])
    _here = np.asarray(_rows, dtype=float)
    _there = _arm2_tests[(_arm2_tests["arm"] == arm)
                         & (_arm2_tests["subset"] == "gate")].iloc[0]
    _p_here = study2_core._safe_wilcoxon(_here)
    _values = {"n": (len(_here), int(_there["n"])),
               "median": (float(np.median(_here)), float(_there["median_delta"])),
               "share_up": (float((_here > 0).mean()), float(_there["share_up"])),
               "p": (float(_p_here), float(_there["p_wilcoxon"]))}
    print(f"{arm:<10} {_values['n'][0]:>7} {_values['n'][1]:>8} "
          f"{_values['median'][0]:>12.6f} {_values['median'][1]:>13.6f} "
          f"{_values['share_up'][0]:>8.4f} {_values['share_up'][1]:>9.4f} "
          f"{_values['p'][0]:>11.4e} {_values['p'][1]:>11.4e}")
    for _name, (_here_value, _there_value) in _values.items():
        _worst = max(_worst, abs(float(_here_value) - float(_there_value)))
print()
print(f"control largest deviation from section 15: {_worst:.3g} "
      f"(expected below 1e-09)")
if _worst > 1e-9:
    raise AssertionError(
        f"The overall effect of Arm 2 computed here deviates from notebook 06 "
        f"section 15 by {_worst:.3g}. The two do not use the same embeddings "
        f"or the same selection of biographies.")
print("the overall effect of Arm 2 is the one of notebook 06")
```

## 5. The treatment table of Arm 2

- **Question:** How far does the manipulation move a biography under the
  search encoder, over the biographies it changed and that pass the gate?
- **Expectation:** the quantity of notebook 06 section 14.1, formed on the
  embeddings of Arm 2; no direction is fixed in advance for Arm 2.
- **Status:** added on 2026-11-04, the recipe of notebook 06 section 14.1
  without a change.
- **Unit and test:** biography; the median change in percentile points and the
  Wilcoxon signed rank test over the changed biographies of the gate, per arm
  and per occupation against the Bonferroni threshold 0.05/26.
- **Output:** none; the table is printed and its pooled rows are controlled
  against `study2_arm2_treatment.csv`.


```python
# The treatment table of Arm 2, formed as notebook 06 section 14.1 forms it
# for Arm 1: the change in percentile rank per biography, averaged over the
# advertisements, summarised over the biographies the manipulation changed.
BONF_26 = 0.05 / 26
ARM2_TREATMENT_FILE = Path(OUT_BASE) / "study2_arm2_treatment.csv"
if not ARM2_TREATMENT_FILE.exists():
    raise FileNotFoundError(
        f"{ARM2_TREATMENT_FILE} not found. It carries the pooled rows this "
        f"section is controlled against.")
print(f"Bonferroni threshold over the 26 occupation rows: p < {BONF_26:.5f}")

_delta_rows = []
for occ in KEPT_13:
    _so = emb_orig[occ] @ emb_jd[occ].T
    _base = study2_core.percentile_ranks(_so)
    for arm in ARMS:
        _sm = emb_manip[occ][arm] @ emb_jd[occ].T
        _delta = study2_core.individual_cf_ranks(_so, _sm) - _base
        for _bio_idx in range(len(_delta)):
            _delta_rows.append({"occ": occ, "bio_idx": _bio_idx, "arm": arm,
                                "delta": float(_delta[_bio_idx])})

# The number of substitutions and the content-preservation similarity belong
# to the text and not to the encoder, so both come from notebook 06.
df_arm2_pool = pd.DataFrame(_delta_rows).merge(
    df_pool[["occ", "bio_idx", "arm", "n_subs", "sim_cp"]],
    on=["occ", "bio_idx", "arm"], how="left")
_absent = int(df_arm2_pool["n_subs"].isna().sum())
print(f"records of Arm 2: {len(df_arm2_pool)}, without a match in notebook 06:"
      f" {_absent} (expected 0)")
if _absent:
    raise AssertionError(
        f"{_absent} biographies of Arm 2 do not appear in the pool of "
        f"notebook 06, so it is unknown whether the manipulation changed them.")

_gated_a2 = df_arm2_pool[df_arm2_pool["sim_cp"] >= PRIMARY_GATE]
_t_occ = study2_core.treatment_table(_gated_a2, ["occ", "arm"])
_t_all = study2_core.treatment_table(_gated_a2, ["arm"])
_t_all.insert(0, "occ", "POOLED")
df_arm2_treatment = pd.concat([_t_occ, _t_all], ignore_index=True)

# The shares that rise and that fall are taken over the changed biographies,
# so they and the share that stays put add up to one.
_share_rows = []
for _row in df_arm2_treatment.itertuples():
    _sub = _gated_a2[(_gated_a2["arm"] == _row.arm) & (_gated_a2["n_subs"] >= 1)]
    if _row.occ != "POOLED":
        _sub = _sub[_sub["occ"] == _row.occ]
    _values = _sub["delta"].to_numpy(float)
    _share_rows.append({"n_moved": int((np.abs(_values) > 0).sum()),
                        "share_up_of_changed": float((_values > 0).mean()),
                        "share_down_of_changed": float((_values < 0).mean())})
df_arm2_treatment = pd.concat([df_arm2_treatment, pd.DataFrame(_share_rows)],
                              axis=1)
df_arm2_treatment["below_bonferroni"] = (
    df_arm2_treatment["p_wilcoxon_treated"] < BONF_26)

_cols = ["occ", "arm", "n", "n_treated", "n_moved", "median_treated",
         "share_up_of_changed", "share_down_of_changed", "p_wilcoxon_treated",
         "below_bonferroni"]
print()
print(df_arm2_treatment[_cols].to_string(index=False))
_occ_rows = df_arm2_treatment[df_arm2_treatment["occ"] != "POOLED"]
print()
print(f"occupation rows below the Bonferroni threshold: "
      f"{int(_occ_rows['below_bonferroni'].sum())} of {len(_occ_rows)}")

# Control: the pooled rows are the ones of study2_arm2_treatment.csv. Both
# ways are printed next to each other before the run stops on a deviation.
_saved_a2 = pd.read_csv(ARM2_TREATMENT_FILE).set_index("arm")
print()
print(f"  {'arm':<10} {'quantity':<22} {'here':>14} {'saved':>14} "
      f"{'deviation':>11}")
print("  " + "-" * 76)
_worst_a2 = 0.0
for arm in ARMS:
    _sub = _gated_a2[(_gated_a2["arm"] == arm) & (_gated_a2["n_subs"] >= 1)]
    _values = _sub["delta"].to_numpy(float)
    _moved = _values[np.abs(_values) > 0]
    _here = {"n_gate": float((_gated_a2["arm"] == arm).sum()),
             "n_changed": float(len(_values)), "n_moved": float(len(_moved)),
             "median_delta_changed": float(np.median(_values)),
             "mean_delta_changed": float(_values.mean()),
             "share_up_of_moved": float((_moved > 0).mean()),
             "share_down_of_moved": float((_moved < 0).mean()),
             "p_wilcoxon_changed": float(study2_core._safe_wilcoxon(_values))}
    for _key, _value in _here.items():
        _there = float(_saved_a2.loc[arm, _key])
        _deviation = abs(_value - _there)
        _worst_a2 = max(_worst_a2, _deviation)
        print(f"  {arm:<10} {_key:<22} {_value:>14.8f} {_there:>14.8f} "
              f"{_deviation:>11.3g}")
print(f"  largest deviation: {_worst_a2:.3g} (expected below 1e-06)")
if _worst_a2 > 1e-6:
    raise AssertionError(
        f"The treatment table of Arm 2 deviates from "
        f"{ARM2_TREATMENT_FILE.name} by {_worst_a2:.3g}. Both ways are "
        f"printed above; the two computations do not agree.")
```

## 6. The change by the wording of the advertisement

- **Question:** Is the change larger against advertisements that carry more of
  the manipulated dimension?
- **Expectation:** the matched pair, a communal manipulation against a communal
  advertisement, shows the stronger relation; the sign follows the one of
  notebook 06 section 14.4.
- **Status:** added on 2026-10-18 after the Arm 1 results, same specification as Arm 1.
- **Unit and test:** advertisement; Spearman within the occupation with a
  pooled permutation test, 5,000 permutations, seed 42, two-sided.
- **Output:** `study2_arm2_jd_language_test.csv`,
  `study2_arm2_jd_language_test_per_occ.csv`.

```python
# The same test as notebook 06 section 14.4, on the change of Arm 2: both arms
# against the communion and the agency density, each also without the D1
# function words.
_densities = {}
for occ in KEPT_13:
    _level = df_jd_level[df_jd_level["occ"] == occ].sort_values("jd_idx")
    _densities[occ] = {"communion": _level["comm_dens"].to_numpy(float),
                       "agency": _level["ag_dens"].to_numpy(float),
                       "communion_cw": _level["comm_dens_cw"].to_numpy(float),
                       "agency_cw": _level["ag_dens_cw"].to_numpy(float)}

_pool_rows, _occ_rows = [], []
for arm in ARMS:
    for _dimension in ("communion", "agency", "communion_cw", "agency_cw"):
        _pairing = ("matched" if arm == _dimension.replace("_cw", "")
                    else "crossed")
        _delta_by_occ = {occ: delta_j_arm2[(occ, arm)] for occ in KEPT_13}
        _dens_by_occ = {occ: _densities[occ][_dimension] for occ in KEPT_13}
        _tab, _pool = study2_core.jd_language_test(
            _delta_by_occ, _dens_by_occ, n_perm=N_PERM, seed=SEED)
        print(f"{arm:<10} against {_dimension:<13} ({_pairing:<7}) "
              f"rho {_pool.get('rho', float('nan')):>+6.3f}  "
              f"p {_pool.get('p_rho', float('nan')):>7.4f}  "
              f"diff hi-lo {_pool.get('diff_hi_lo', float('nan')):>+7.3f}  "
              f"p {_pool.get('p_diff', float('nan')):>7.4f}  "
              f"advertisements {int(_pool.get('n_jds', 0))}")
        _pool_rows.append({"encoder": "arm2", "sample": "pool",
                           "delta": "delta_rank", "arm": arm,
                           "jd_dimension": _dimension, "pairing": _pairing,
                           "n_occupations": int(_pool.get("n_occ", 0)),
                           "n_jds": int(_pool.get("n_jds", 0)),
                           "rho": float(_pool.get("rho", float("nan"))),
                           "p_rho": float(_pool.get("p_rho", float("nan"))),
                           "diff_hi_lo": float(_pool.get("diff_hi_lo", float("nan"))),
                           "p_diff": float(_pool.get("p_diff", float("nan")))})
        _tab = _tab.copy()
        _tab.insert(0, "jd_dimension", _dimension)
        _tab.insert(0, "arm", arm)
        _tab.insert(0, "encoder", "arm2")
        _occ_rows.append(_tab)
df_arm2_jd_language = pd.DataFrame(_pool_rows)
df_arm2_jd_language_per_occ = pd.concat(_occ_rows, ignore_index=True)
print()
print(f"pooled rows {len(df_arm2_jd_language)}, rows per occupation "
      f"{len(df_arm2_jd_language_per_occ)}")
```

## 7. Joint model of the two densities

- **Question:** Does one of the two densities carry the relation once the other
  one and the length of the advertisement are held constant?
- **Expectation:** none fixed in advance; the model separates what the single
  correlations leave together.
- **Status:** added on 2026-10-18 after the Arm 1 results, same specification as Arm 1.
- **Unit and test:** advertisement; ordinary least squares with occupation
  effects and HC1 standard errors, as in notebook 06 section 14.6.
- **Output:** `study2_arm2_jd_joint_model.csv`.

```python
# The joint model of notebook 06 section 14.6, on the change of Arm 2. The
# densities and the length are standardised within the occupation, so the
# coefficients are comparable across occupations.
_jdj = df_arm2_jd_level.copy()
for _v in ("comm_dens", "ag_dens", "comm_dens_cw", "ag_dens_cw"):
    _jdj[f"{_v}_z"] = _jdj.groupby("occ")[_v].transform(
        lambda x: (x - x.mean()) / x.std(ddof=1) if x.std(ddof=1) > 0 else 0.0)
_jdj["logtok_raw"] = np.log(_jdj["n_tokens"].clip(lower=1))
_jdj["logtok_z"] = _jdj.groupby("occ")["logtok_raw"].transform(
    lambda x: (x - x.mean()) / x.std(ddof=1) if x.std(ddof=1) > 0 else 0.0)

_outcomes = [f"dj_pool_{arm}_rank" for arm in ARMS]
_density_sets = [("primary", "comm_dens_z", "ag_dens_z"),
                 ("robustness", "comm_dens_cw_z", "ag_dens_cw_z")]
_jm_rows = []
print(f"occupations: {_jdj['occ'].nunique()}")
print(f"  {'outcome':<28} {'status':<11} {'n':>5} {'comm_z':>8} {'p':>7} "
      f"{'ag_z':>8} {'p':>7} {'logtok_z':>9} {'p':>7} {'R2':>6}")
for _status, _comm, _ag in _density_sets:
    for _y in _outcomes:
        _sub = _jdj[["occ", _y, _comm, _ag, "logtok_z"]].dropna()
        if len(_sub) < 10 or _sub["occ"].nunique() < 2:
            print(f"  {_y:<28} {_status:<11} left out: {len(_sub)} rows, "
                  f"{_sub['occ'].nunique()} occupations")
            continue
        _res = smf.ols(f"{_y} ~ {_comm} + {_ag} + logtok_z + C(occ)",
                       data=_sub).fit(cov_type="HC1")
        print(f"  {_y:<28} {_status:<11} {int(_res.nobs):>5} "
              f"{float(_res.params[_comm]):>+8.3f} "
              f"{float(_res.pvalues[_comm]):>7.4f} "
              f"{float(_res.params[_ag]):>+8.3f} "
              f"{float(_res.pvalues[_ag]):>7.4f} "
              f"{float(_res.params['logtok_z']):>+9.3f} "
              f"{float(_res.pvalues['logtok_z']):>7.4f} "
              f"{float(_res.rsquared):>6.3f}")
        _jm_rows.append({"encoder": "arm2", "outcome": _y, "status": _status,
                         "x_communion": _comm, "x_agency": _ag,
                         "n": int(_res.nobs),
                         "b_comm_z": float(_res.params[_comm]),
                         "p_comm_z": float(_res.pvalues[_comm]),
                         "b_ag_z": float(_res.params[_ag]),
                         "p_ag_z": float(_res.pvalues[_ag]),
                         "b_logtok_z": float(_res.params["logtok_z"]),
                         "p_logtok_z": float(_res.pvalues["logtok_z"]),
                         "r2": float(_res.rsquared)})
df_arm2_jd_joint_model = pd.DataFrame(_jm_rows)
print(f"  {len(df_arm2_jd_joint_model)} models fitted")
```

## 8. T1 and T2 for Arm 2

- **Question:** T1, does the gap between women and men follow the wording of
  the advertisement? T2, does a biography that the model ranks higher against
  communal advertisements carry the more communal wording?
- **Expectation:** both correlations positive, the sign fixed in advance for
  Arm 1 and kept here.
- **Status:** added on 2026-10-18 after the Arm 1 results, same specification as Arm 1, with the variants A u B, without the D1
  function words, and on that part alone.
- **Unit and test:** T1 on the advertisement, T2 on the biography; Spearman
  within the occupation with a pooled permutation test, 5,000 permutations,
  seed 42, two-sided, alpha 0.025.
- **Output:** `study2_arm2_fit_tests_full.csv`.

```python
# T1 and T2 of notebook 06 sections 14.13 and 14.14, on the ranks of Arm 2,
# with the variants of the dictionary and of the function words.
_ranks_arm2, _gap_arm2, _slope_arm2 = {}, {}, {}
for occ in KEPT_13:
    _ranks_arm2[occ] = study2_core.pool_pct_per_jd(emb_orig[occ] @ emb_jd[occ].T)

_female = {occ: df_bio[df_bio["occ"] == occ].sort_values("bio_idx")
           ["female"].to_numpy(int) for occ in KEPT_13}
for occ in KEPT_13:
    _matrix, _flags = _ranks_arm2[occ], _female[occ]
    if len(_flags) != _matrix.shape[0]:
        raise AssertionError(
            f"{occ}: {len(_flags)} biographies in the table and "
            f"{_matrix.shape[0]} in the ranks.")
    if (_flags == 1).sum() < 2 or (_flags == 0).sum() < 2:
        raise AssertionError(
            f"{occ}: women {int((_flags == 1).sum())}, men "
            f"{int((_flags == 0).sum())}; at least two of each are needed.")
    _gap_arm2[occ] = (_matrix[_flags == 1].mean(axis=0)
                      - _matrix[_flags == 0].mean(axis=0))

_jd_variants = {"balance": "balance", "balance_AB": "balance_AB",
                "balance_cw": "balance_cw", "balance_fw": "balance_fw"}
_bio_variants = {"balance": "bio_balance", "balance_AB": "bio_balance_AB",
                 "balance_cw": "bio_balance_cw", "balance_fw": "bio_balance_fw"}
_fit_rows = []
print(f"  {'test':<4} {'variant':<12} {'status':<11} {'occ':>4} {'rho':>8} "
      f"{'p':>9} {'diff hi-lo':>11} {'p':>9}")
print("  " + "-" * 74)
for _variant, _jd_column in _jd_variants.items():
    _status = "primary" if _variant == "balance" else "robustness"
    _dens = {occ: df_jd_level[df_jd_level["occ"] == occ].sort_values("jd_idx")
             [_jd_column].to_numpy(float) for occ in KEPT_13}
    _, _pool = study2_core.jd_language_test(_gap_arm2, _dens, n_perm=N_PERM,
                                            seed=SEED)
    print(f"  {'T1':<4} {_variant:<12} {_status:<11} "
          f"{int(_pool.get('n_occ', 0)):>4} {float(_pool.get('rho', float('nan'))):>+8.4f} "
          f"{float(_pool.get('p_rho', float('nan'))):>9.4f} "
          f"{float(_pool.get('diff_hi_lo', float('nan'))):>+11.4f} "
          f"{float(_pool.get('p_diff', float('nan'))):>9.4f}")
    _fit_rows.append({"encoder": "arm2", "test": "T1", "variant": _variant,
                      "status": _status, "unit": "job description",
                      "y": "gap_f_minus_m", "x": _jd_column,
                      "expected_sign": "+", "alpha": ALPHA_PRE,
                      "n_occupations": int(_pool.get("n_occ", 0)),
                      "n_points": int(_pool.get("n_jds", 0)),
                      "rho": float(_pool.get("rho", float("nan"))),
                      "p_rho": float(_pool.get("p_rho", float("nan"))),
                      "diff_hi_lo": float(_pool.get("diff_hi_lo", float("nan"))),
                      "p_diff": float(_pool.get("p_diff", float("nan")))})

for _variant, _bio_column in _bio_variants.items():
    _status = "primary" if _variant == "balance" else "robustness"
    _jd_column = _jd_variants[_variant]
    _slopes = {}
    for occ in KEPT_13:
        _values = (df_jd_level[df_jd_level["occ"] == occ].sort_values("jd_idx")
                   [_jd_column].to_numpy(float))
        _centred = _values - _values.mean()
        _denominator = float(_centred @ _centred)
        if _denominator < 1e-12:
            raise AssertionError(
                f"{occ}: the variant {_variant} of the advertisements does not "
                f"vary, so no slope can be formed on it.")
        _matrix = _ranks_arm2[occ]
        _slope = ((_matrix - _matrix.mean(axis=1, keepdims=True)) @ _centred
                  / _denominator)
        if abs(_slope.mean()) > 1e-7:
            raise AssertionError(
                f"control mean slope {occ}/{_variant}: {_slope.mean():.2e} "
                f"(expected below 1e-7)")
        _slopes[occ] = _slope
    if _variant == "balance":
        _slope_arm2 = dict(_slopes)
    _bio_values = {occ: df_bio[df_bio["occ"] == occ].sort_values("bio_idx")
                   [_bio_column].to_numpy(float) for occ in KEPT_13}
    _, _pool = study2_core.jd_language_test(_slopes, _bio_values,
                                            n_perm=N_PERM, seed=SEED)
    print(f"  {'T2':<4} {_variant:<12} {_status:<11} "
          f"{int(_pool.get('n_occ', 0)):>4} {float(_pool.get('rho', float('nan'))):>+8.4f} "
          f"{float(_pool.get('p_rho', float('nan'))):>9.4f} "
          f"{float(_pool.get('diff_hi_lo', float('nan'))):>+11.4f} "
          f"{float(_pool.get('p_diff', float('nan'))):>9.4f}")
    _fit_rows.append({"encoder": "arm2", "test": "T2", "variant": _variant,
                      "status": _status, "unit": "biography",
                      "y": "slope_over_jds", "x": _bio_column,
                      "expected_sign": "+", "alpha": ALPHA_PRE,
                      "n_occupations": int(_pool.get("n_occ", 0)),
                      "n_points": int(_pool.get("n_jds", 0)),
                      "rho": float(_pool.get("rho", float("nan"))),
                      "p_rho": float(_pool.get("p_rho", float("nan"))),
                      "diff_hi_lo": float(_pool.get("diff_hi_lo", float("nan"))),
                      "p_diff": float(_pool.get("p_diff", float("nan")))})
df_arm2_fit_tests = pd.DataFrame(_fit_rows)

# Control: the two main rows have to be the ones of the script of 2026-10-13.
if ARM2_FIT_FILE.exists():
    _saved = pd.read_csv(ARM2_FIT_FILE)
    _saved = _saved[_saved["arm"] == "arm2"].set_index("test")
    print()
    print(f"  {'test':<4} {'rho here':>10} {'rho 13.10.':>11} "
          f"{'p here':>10} {'p 13.10.':>10} {'deviation':>11}")
    _worst_fit = 0.0
    for _test in ("T1", "T2"):
        _here = df_arm2_fit_tests[(df_arm2_fit_tests["test"] == _test)
                                  & (df_arm2_fit_tests["variant"] == "balance")].iloc[0]
        _there = _saved.loc[_test]
        _deviation = abs(float(_here["rho"]) - float(_there["rho"]))
        _worst_fit = max(_worst_fit, _deviation)
        print(f"  {_test:<4} {float(_here['rho']):>+10.6f} "
              f"{float(_there['rho']):>+11.6f} {float(_here['p_rho']):>10.5f} "
              f"{float(_there['p_rho']):>10.5f} {_deviation:>11.3g}")
    print(f"  largest deviation: {_worst_fit:.3g} (expected below 1e-06)")
    if _worst_fit > 1e-6:
        raise AssertionError(
            f"T1 and T2 of Arm 2 deviate from the script of 2026-10-13 by "
            f"{_worst_fit:.3g}. Both ways are printed above; the two "
            f"computations do not agree.")
else:
    print()
    print(f"{ARM2_FIT_FILE} is not present, so the two main rows are not "
          f"compared with the script of 2026-10-13.")
```

## 9. The fit by gender

- **Question:** Do the biographies of women and of men differ in how their rank
  follows the wording of the advertisement, and does the manipulation change
  that?
- **Expectation:** none fixed in advance; the section is read as a description.
- **Status:** added on 2026-10-18 after the Arm 1 results, same specification as Arm 1.
- **Unit and test:** biography; ordinary least squares with occupation effects
  and HC1 standard errors, and the signed rank test on the change of the slope,
  as in notebook 06 section 14.15.
- **Output:** `study2_arm2_fit_gender.csv`.

```python
# The fit by gender, as notebook 06 section 14.15, on the ranks of Arm 2. The
# slope of a biography over the advertisements is the quantity; model 3 repeats
# it on the manipulated version of the changed biographies at the gate.
from scipy.stats import wilcoxon

_slope_rows = []
for occ in KEPT_13:
    _values = df_bio[df_bio["occ"] == occ].sort_values("bio_idx")
    for _i, _row in enumerate(_values.itertuples()):
        _slope_rows.append({"occ": occ, "bio_idx": int(_row.bio_idx),
                            "female": int(_row.female),
                            "word_count": float(_row.word_count),
                            "communion_rate": float(_row.communion_rate),
                            "agency_rate": float(_row.agency_rate),
                            "bio_balance": float(_row.bio_balance),
                            "s_bal": float(_slope_arm2[occ][_i])})
_fs = pd.DataFrame(_slope_rows)
_fs["log_wc"] = np.log(_fs["word_count"].clip(lower=1))
print(f"biographies with a slope: {len(_fs)}  women "
      f"{int((_fs['female'] == 1).sum())}  men {int((_fs['female'] == 0).sum())}")

_gender_rows = []
_m1 = smf.ols("s_bal ~ female + C(occ)", data=_fs).fit(cov_type="HC1")
_b1 = float(_m1.params["female"])
_ci1 = _m1.conf_int().loc["female"]
print(f"  model 1 (s_bal ~ female + C(occ)): b={_b1:>+.4f}  "
      f"SE={float(_m1.bse['female']):.4f}  "
      f"95%CI=[{float(_ci1[0]):>+.4f}, {float(_ci1[1]):>+.4f}]  "
      f"p={float(_m1.pvalues['female']):.4f}  n={int(_m1.nobs)}")
_gender_rows.append({"encoder": "arm2", "model": "m1", "y": "s_bal",
                     "predictors": "female+C(occ)", "b_female": _b1,
                     "se": float(_m1.bse["female"]), "ci_lo": float(_ci1[0]),
                     "ci_hi": float(_ci1[1]),
                     "p": float(_m1.pvalues["female"]), "n": int(_m1.nobs)})

_m2 = smf.ols("s_bal ~ female + communion_rate + agency_rate + log_wc + C(occ)",
              data=_fs).fit(cov_type="HC1")
_b2 = float(_m2.params["female"])
_ci2 = _m2.conf_int().loc["female"]
if _b1 == 0.0:
    raise AssertionError(
        "The coefficient of model 1 is exactly zero, so the share by which "
        "model 2 reduces it is not defined.")
_reduction = 1.0 - _b2 / _b1
print(f"  model 2 (+ rates + log_wc)      : b={_b2:>+.4f}  "
      f"SE={float(_m2.bse['female']):.4f}  p={float(_m2.pvalues['female']):.4f}")
print(f"  1 - b_female(M2)/b_female(M1)   : {_reduction:>.3f}")
_gender_rows.append({"encoder": "arm2", "model": "m2", "y": "s_bal",
                     "predictors": "female+comm_rate+ag_rate+log_wc+C(occ)",
                     "b_female": _b2, "se": float(_m2.bse["female"]),
                     "ci_lo": float(_ci2[0]), "ci_hi": float(_ci2[1]),
                     "p": float(_m2.pvalues["female"]), "n": int(_m2.nobs),
                     "b_communion_rate": float(_m2.params["communion_rate"]),
                     "b_agency_rate": float(_m2.params["agency_rate"]),
                     "reduction_b_female": _reduction})

# Model 3: the same slope after the manipulation, on the changed biographies of
# the gate, and the change of the slope per gender.
print()
print(f"  after the manipulation, gate {PRIMARY_GATE}:")
for arm in ARMS:
    _rows = []
    for occ in KEPT_13:
        _so = emb_orig[occ] @ emb_jd[occ].T
        _sm = emb_manip[occ][arm] @ emb_jd[occ].T
        _pcf = study2_core.indiv_cf_pct_per_jd(_so, _sm)
        _values = (df_jd_level[df_jd_level["occ"] == occ].sort_values("jd_idx")
                   ["balance"].to_numpy(float))
        _centred = _values - _values.mean()
        _s_cf = ((_pcf - _pcf.mean(axis=1, keepdims=True)) @ _centred
                 / float(_centred @ _centred))
        _sub = df_pool[(df_pool["occ"] == occ) & (df_pool["arm"] == arm)
                       & (df_pool["sim_cp"] >= PRIMARY_GATE)
                       & (df_pool["n_subs"] >= 1)]
        for _bio_idx in _sub["bio_idx"].to_numpy(int):
            _rows.append({"occ": occ, "bio_idx": int(_bio_idx),
                          "s_bal_cf": float(_s_cf[_bio_idx])})
    _cf = pd.DataFrame(_rows).merge(_fs[["occ", "bio_idx", "female", "s_bal"]],
                                    on=["occ", "bio_idx"], how="inner")
    _cf["delta_s"] = _cf["s_bal_cf"] - _cf["s_bal"]
    _m3 = smf.ols("s_bal_cf ~ female + C(occ)", data=_cf).fit(cov_type="HC1")
    print(f"    {arm:<10} b_female={float(_m3.params['female']):>+.4f}  "
          f"p={float(_m3.pvalues['female']):.4f}  n={int(_m3.nobs)}")
    _gender_rows.append({"encoder": "arm2", "model": f"m3_{arm}",
                         "y": "s_bal_cf", "predictors": "female+C(occ)",
                         "b_female": float(_m3.params["female"]),
                         "se": float(_m3.bse["female"]),
                         "p": float(_m3.pvalues["female"]), "n": int(_m3.nobs)})
    for _value, _label in ((1, "women"), (0, "men")):
        _side = _cf[_cf["female"] == _value]["delta_s"].dropna()
        if len(_side) < 5 or not (_side != 0).any():
            print(f"    {arm}/{_label}: {len(_side)} changes, test left out")
            continue
        _, _p = wilcoxon(_side)
        print(f"    {arm}/{_label}: median change of the slope "
              f"{float(_side.median()):>+.4f}  p={float(_p):.4f}  "
              f"n={len(_side)}")
        _gender_rows.append({"encoder": "arm2", "model": f"delta_slope_{arm}",
                             "y": "s_bal_cf - s_bal", "predictors": _label,
                             "b_female": float("nan"),
                             "median_delta_slope": float(_side.median()),
                             "p": float(_p), "n": int(len(_side))})
df_arm2_fit_gender = pd.DataFrame(_gender_rows)
print()
print(f"rows: {len(df_arm2_fit_gender)}")
```

## 10. The content-preservation gate

- **Question:** How close to the original do the manipulated biographies stay,
  and what does a biography near the threshold look like?
- **Expectation:** the great majority sits far above 0.90; the cases between
  0.85 and 0.95 are the ones a reader should be able to judge.
- **Status:** added on 2026-10-18 after the Arm 1 results, same specification as Arm 1.
- **Unit and test:** biography; quantiles and counts, no test, plus a sample of
  borderline cases drawn with seed 42.
- **Output:** `study2_gate_distribution.csv`,
  `study2_gate_borderline_examples.csv`.

```python
# The distribution of the similarity between a biography and its manipulated
# version, over the changed biographies of the pool, and the borderline cases.
_changed_pool = df_pool[df_pool["n_subs"] >= 1]
print(f"changed biographies of the pool: {len(_changed_pool)}")
print(f"  per arm: {dict(_changed_pool['arm'].value_counts())}")
print()
print(f"  {'arm':<10} {'n':>6} {'q01':>7} {'q05':>7} {'q10':>7} {'q25':>7} "
      f"{'q50':>7} {'<0.85':>7} {'<0.90':>7} {'<0.95':>7}")
print("  " + "-" * 80)
_gate_rows = []
for arm in list(ARMS) + ["both"]:
    _sub = (_changed_pool if arm == "both"
            else _changed_pool[_changed_pool["arm"] == arm])
    _sim = _sub["sim_cp"].to_numpy(float)
    _quantiles = {f"q{int(q * 100):02d}": float(np.quantile(_sim, q))
                  for q in (0.01, 0.05, 0.10, 0.25, 0.50)}
    _counts = {f"n_below_{str(t).replace('.', '')}": int((_sim < t).sum())
               for t in (0.85, 0.90, 0.95)}
    print(f"  {arm:<10} {len(_sim):>6} "
          + " ".join(f"{_quantiles[k]:>7.4f}" for k in sorted(_quantiles))
          + " " + " ".join(f"{_counts[k]:>7}" for k in sorted(_counts)))
    _gate_rows.append({"arm": arm, "n_changed": len(_sim), **_quantiles,
                       **_counts,
                       "share_below_090": float((_sim < 0.90).mean())})
df_gate_distribution = pd.DataFrame(_gate_rows)

# The borderline cases, with the places the manipulation changed. The protocol
# carries the context of every occurrence, eight tokens before and after.
_protocol = pd.read_csv(PROTOCOL_FILE)
_protocol = _protocol[(_protocol["sample"] == "pool")
                      & (_protocol["action_taken"].isin(["replaced", "deleted"]))]
print()
print(f"protocol rows of the pool with a change: {len(_protocol)}")
_rng = np.random.default_rng(SEED)
_example_rows = []
for arm in ARMS:
    _band = _changed_pool[(_changed_pool["arm"] == arm)
                          & (_changed_pool["sim_cp"] >= GATE_LOW)
                          & (_changed_pool["sim_cp"] < GATE_HIGH)]
    _take = min(N_BORDERLINE, len(_band))
    print(f"  {arm:<10} biographies between {GATE_LOW} and {GATE_HIGH}: "
          f"{len(_band)}, drawn: {_take}")
    if _take == 0:
        continue
    _chosen = _band.iloc[_rng.choice(len(_band), size=_take, replace=False)]
    for _row in _chosen.itertuples():
        _places = _protocol[(_protocol["occ"] == _row.occ)
                            & (_protocol["arm"] == arm)
                            & (_protocol["bio_idx"] == _row.bio_idx)]
        for _place in _places.itertuples():
            _example_rows.append({
                "arm": arm, "occ": _row.occ, "bio_idx": int(_row.bio_idx),
                "sim_cp": float(_row.sim_cp), "n_subs": int(_row.n_subs),
                "delta": float(_row.delta), "token": _place.token,
                "rule_id": _place.rule_id, "action_taken": _place.action_taken,
                "target": _place.target,
                "context_original": _place.context_original,
                "context_manipulated": _place.context_manipulated})
df_gate_borderline = pd.DataFrame(_example_rows)
print()
print(f"borderline biographies: "
      f"{df_gate_borderline[['arm', 'occ', 'bio_idx']].drop_duplicates().shape[0]}, "
      f"changed places in them: {len(df_gate_borderline)}")
if not df_gate_borderline.empty:
    print(f"  substitutions per biography: median "
          f"{df_gate_borderline.groupby(['arm', 'occ', 'bio_idx']).size().median():.0f}")
    print()
    print("the first three places, as they stand in the protocol:")
    for _row in df_gate_borderline.head(3).itertuples():
        print(f"  {_row.occ}/{_row.arm}/{_row.bio_idx} sim_cp "
              f"{_row.sim_cp:.4f}  {_row.token} -> {_row.target}")
        print(f"    before: {_row.context_original}")
        print(f"    after : {_row.context_manipulated}")
```

## 11. Save outputs

- **Question:** Does every table of sections 6 to 10 reach the output folder?
- **Expectation:** all eight files, none of them empty, and none of them
  overwriting a table of notebook 06.
- **Status:** added on 2026-10-18 after the Arm 1 results, same specification as Arm 1.
- **Unit and test:** control; the run stops if a table is missing or empty.
- **Output:** the eight files listed in the head of the notebook.

```python
# Write the tables of this notebook. The names all carry arm2 or gate, so none
# of them is a table of notebook 06; a name that already exists stops the run.
TABLES = [
    ("study2_arm2_jd_language_test.csv",         "df_arm2_jd_language"),
    ("study2_arm2_jd_language_test_per_occ.csv", "df_arm2_jd_language_per_occ"),
    ("study2_arm2_jd_joint_model.csv",           "df_arm2_jd_joint_model"),
    ("study2_arm2_fit_tests_full.csv",           "df_arm2_fit_tests"),
    ("study2_arm2_fit_gender.csv",               "df_arm2_fit_gender"),
    ("study2_arm2_jd_level.csv",                 "df_arm2_jd_level"),
    ("study2_gate_distribution.csv",             "df_gate_distribution"),
    ("study2_gate_borderline_examples.csv",      "df_gate_borderline"),
]
_never_built = [name for _, name in TABLES if name not in globals()]
if _never_built:
    raise AssertionError(
        f"These tables were never built: {_never_built}. The sections that "
        f"fill them have to run before this cell.")
_no_rows = [name for _, name in TABLES if globals()[name].empty]
if _no_rows:
    raise AssertionError(
        f"These tables hold no rows: {_no_rows}. An empty table is not a "
        f"result, so the run stops here.")

written = {}
for _file, _name in TABLES:
    _path = Path(OUT_BASE) / _file
    globals()[_name].to_csv(_path, index=False, encoding="utf-8")
    written[_file] = len(globals()[_name])

print(f"  {'file':<46} {'rows':>7}  {'KB':>8}")
print("  " + "-" * 66)
_failed = []
for _file in sorted(written):
    _path = Path(OUT_BASE) / _file
    if not _path.exists() or _path.stat().st_size == 0 or written[_file] == 0:
        _failed.append(_file)
        continue
    print(f"  {_file:<46} {written[_file]:>7} "
          f"{_path.stat().st_size / 1024:>8.1f}")
if _failed:
    raise AssertionError(f"Missing or empty after writing: {_failed}.")
print()
print(f"files written: {len(written)} tables in {OUT_BASE}/")
```

## 12. The values of Figure 5 and Figure 6

- **Question:** How do the three gaps run across the fifths of the
  advertisements, and do the three screeners place them in the same order?
- **Expectation:** none fixed in advance; the section gathers what the
  two figures draw.
- **Status:** added on 2026-11-04, the heading revised on 2026-11-13, when
  Figure 6 took the original rankings over, and revised on 2026-11-14, when
  the gap between the biographies of women and of men under the language
  model moved to the rule of notebook 07. The ranks of Arm 1 come from the
  embeddings notebook 06 cached, those of Arm 2 from section 8, those of
  Arm 3 from the tables of notebook 07; nothing is embedded here.
- **Unit and test:** advertisement; means per fifth, no test. The fifths are
  cut within the occupation on the balance of the advertisement, and the
  occupations enter with equal weight. The gap of the language model is the
  one notebook 07 cell s07_gender forms for the test it fixed in advance: per
  advertisement over the biographies it was rated against, and only where at
  least three of each gender carry a value. Section 13.2 and section 25 read
  it from here, and the section controls it against
  `arm3/main/study2_arm3_gender_gap.csv`.
- **Output:** `study2_fig5_values.csv`.


```python
# The values behind Figure 5: three gaps per screener and fifth of the
# advertisements. The fifths are cut within the occupation on the balance of
# the advertisement under dictionary A, and the occupations enter with equal
# weight. Panels (b) and (c) use the original biographies only. The gap
# between the biographies of women and of men, which T1 reads, is formed for
# the language model exactly as notebook 07 forms it, so Figure 6, section 13.2
# and section 25 all carry the one value of the test fixed in advance.
N_PARTS = 5
ARM1_CACHE = Path("outputs/study2_cache/sentence_transformers_all_mpnet_base_v2")
ARM3_DIR = Path(OUT_BASE) / "arm3" / "main"
MEETING_DIR = Path("outputs/meeting_1310")
ARM3_ORIGINALS = ARM3_DIR / "study2_arm3_originals.csv"
ARM3_PCT_BY_JD = ARM3_DIR / "study2_arm3_pct_by_jd.csv"
ARM3_GENDER_GAP = ARM3_DIR / "study2_arm3_gender_gap.csv"
FIG5_VALUES = Path(OUT_BASE) / "study2_fig5_values.csv"
SCREENERS = (("arm1", "SBERT"), ("arm2", "Qwen3-Embedding"),
             ("arm3", "language model"))
# Notebook 07 leaves an advertisement out of the gap when fewer than three
# biographies of a gender carry a value.
ARM3_MIN_PER_GENDER = 3

_needed = [ARM1_CACHE / f"{occ}_pool_orig.npy" for occ in KEPT_13]
_needed += [ARM1_CACHE / f"{occ}_jds.npy" for occ in KEPT_13]
_needed += [ARM3_ORIGINALS, ARM3_GENDER_GAP]
for _path in _needed:
    if not _path.exists():
        raise FileNotFoundError(
            f"{_path} not found. Panel (b) and panel (c) read the ranks of "
            f"Arm 1 from the cache of notebook 06 and the ranks of Arm 3 from "
            f"the outputs of notebook 07.")


def _fifths(values):
    """The fifths of a quantity, cut within one occupation."""
    return pd.qcut(values.rank(method="first"), N_PARTS, labels=False,
                   duplicates="drop") + 1


# The fifths of the advertisements and of the biographies.
_jd_parts = df_jd_eda[["occ", "jd_idx", "balance"]].copy()
_jd_parts["jd_fifth"] = _jd_parts.groupby("occ")["balance"].transform(_fifths)
_bio_parts = df_bio[["occ", "bio_idx", "female", "bio_balance"]].copy()
_bio_parts["bio_fifth"] = (_bio_parts.groupby("occ")["bio_balance"]
                           .transform(_fifths))
print(f"advertisements {len(_jd_parts)}, biographies {len(_bio_parts)}, "
      f"fifths cut within the occupation on the balance of dictionary "
      f"{LEX_MAIN}")

# The ranks of the original biographies per advertisement. Arm 1 comes from
# the embeddings notebook 06 cached, Arm 2 from the ranks of section 8, Arm 3
# from the table notebook 07 wrote. Nothing is embedded here.
_ranks = {"arm2": _ranks_arm2}
_ranks["arm1"] = {}
_saved_pct = (df_pool[df_pool["arm"] == "communion"]
              .set_index(["occ", "bio_idx"])["pct_orig"])
_worst_rank = 0.0
for occ in KEPT_13:
    _bio_cache = np.load(ARM1_CACHE / f"{occ}_pool_orig.npy",
                         allow_pickle=True).item()["mean"]
    _jd_cache = np.load(ARM1_CACHE / f"{occ}_jds.npy", allow_pickle=True).item()
    _ranks["arm1"][occ] = study2_core.pool_pct_per_jd(
        study2_core.jd_similarity(_bio_cache, _jd_cache, how=JD_REPR))
    _worst_rank = max(_worst_rank, float(np.abs(
        _ranks["arm1"][occ].mean(axis=1)
        - _saved_pct.loc[occ].sort_index().to_numpy()).max()))
print(f"control ranks of Arm 1 against the saved pct_orig: largest deviation "
      f"{_worst_rank:.3g} (expected below 1e-06)")
if _worst_rank > 1e-6:
    raise AssertionError(
        f"The ranks of Arm 1 read from the cache deviate from the saved "
        f"pct_orig by {_worst_rank:.3g}.")

_arm3_ranks = pd.read_csv(ARM3_ORIGINALS)
_arm3_ranks["jd_idx"] = [int(str(_j).rsplit("jd", 1)[1])
                         for _j in _arm3_ranks["jd_id"]]
_arm3_ranks = _arm3_ranks.rename(columns={"pct_orig": "pct"})

# Arm 3 rated twenty advertisements per occupation, so its fifths are cut
# among those twenty and hold four each.
_arm3_parts = _jd_parts.merge(
    _arm3_ranks[["occ", "jd_idx"]].drop_duplicates(), on=["occ", "jd_idx"])
_arm3_parts["jd_fifth"] = (_arm3_parts.groupby("occ")["balance"]
                           .transform(_fifths))
print(f"advertisements of Arm 3: {len(_arm3_parts)}, per occupation and fifth "
      f"{sorted(_arm3_parts.groupby(['occ', 'jd_fifth']).size().unique())}")


def _gap_arm3_per_ad():
    """The gap between the biographies of women and of men per advertisement
    under the language model, formed as notebook 07 cell s07_gender forms it:
    over the biographies that advertisement was rated against, and only where
    at least three of each gender carry a percentile rank."""
    _rows, _left_out = [], []
    for (occ, jd_id), group in _arm3_ranks.groupby(["occ", "jd_id"]):
        _f = group[group["female"] == 1]["pct"]
        _m = group[group["female"] == 0]["pct"]
        if len(_f) < ARM3_MIN_PER_GENDER or len(_m) < ARM3_MIN_PER_GENDER:
            _left_out.append((occ, jd_id, len(_f), len(_m)))
            continue
        _rows.append({"occ": occ,
                      "jd_idx": int(str(jd_id).rsplit("jd", 1)[1]),
                      "c": float(_f.mean() - _m.mean())})
    print(f"gap of the language model: {len(_rows)} advertisements carry one, "
          f"left out because fewer than {ARM3_MIN_PER_GENDER} biographies of "
          f"a gender were rated: {len(_left_out)}")
    if _left_out:
        print(f"  the first of them: {_left_out[:3]}")
    return pd.DataFrame(_rows).set_index(["occ", "jd_idx"])["c"]


_arm3_gap = _gap_arm3_per_ad()

# Control: per advertisement this is the gap notebook 07 wrote for its test.
# The route over a matrix of biographies by advertisements, which panel (b)
# takes, gives the same numbers to 1e-13; the gaps hold exact ties, however,
# so a difference in the last bit moves the rank correlation of section 25 in
# the fifth decimal, and the figure and the table would then carry a second
# value of a test that is fixed in advance.
_saved_gap = pd.read_csv(ARM3_GENDER_GAP)
_saved_gap = _saved_gap[_saved_gap["measure"] == "percentile"].copy()
_saved_gap["jd_idx"] = [int(str(_j).rsplit("jd", 1)[1])
                        for _j in _saved_gap["jd_id"]]
_saved_gap = _saved_gap.set_index(["occ", "jd_idx"])["gap_f_minus_m"]
_missing_gap = sorted(set(_saved_gap.index) ^ set(_arm3_gap.index))
_worst_gap = float((_arm3_gap - _saved_gap.reindex(_arm3_gap.index))
                   .abs().max())
print(f"control the gap of the language model against "
      f"{ARM3_GENDER_GAP.name}: advertisements in one table only "
      f"{len(_missing_gap)} (expected 0), largest deviation {_worst_gap:.3g} "
      f"(expected below 1e-09)")
if _missing_gap or _worst_gap > 1e-9:
    raise AssertionError(
        f"The gap of the language model is not the one notebook 07 wrote: "
        f"{len(_missing_gap)} advertisements differ in the two tables and the "
        f"largest deviation is {_worst_gap:.3g}.")

_changed_gate = df_pool[(df_pool["sim_cp"] >= PRIMARY_GATE)
                        & (df_pool["n_subs"] >= 1)
                        & (df_pool["arm"] == "communion")]


def _gaps_per_ad(screener, jd_parts):
    """Panels (b) and (c) per advertisement: the most communal fifth of the
    biographies minus the most agentic one, and women minus men. For the
    language model the second gap is the one of notebook 07, read from
    _arm3_gap; an advertisement without such a gap carries neither panel."""
    rows = []
    for occ in KEPT_13:
        _bios = _bio_parts[_bio_parts["occ"] == occ].sort_values("bio_idx")
        _ads = jd_parts[jd_parts["occ"] == occ]
        _female = _bios["female"].to_numpy(int)
        _fifth = _bios["bio_fifth"].to_numpy(int)
        if screener == "arm3":
            _wide = (_arm3_ranks[_arm3_ranks["occ"] == occ]
                     .pivot(index="bio_idx", columns="jd_idx", values="pct"))
            _with_gap = set(_arm3_gap.loc[occ].index)
            _columns = [c for c in _ads["jd_idx"]
                        if c in _wide.columns and c in _with_gap]
            _matrix = _wide.reindex(index=_bios["bio_idx"])[_columns].to_numpy(float)
            _index = np.asarray(_columns, dtype=int)
            _keep = ~np.isnan(_matrix).any(axis=1)
            _matrix, _female, _fifth = _matrix[_keep], _female[_keep], _fifth[_keep]
        else:
            _index = _ads["jd_idx"].to_numpy(int)
            _matrix = _ranks[screener][occ][:, _index]
        if (_female == 1).sum() < 2 or (_female == 0).sum() < 2:
            raise AssertionError(
                f"{occ}: women {int((_female == 1).sum())}, men "
                f"{int((_female == 0).sum())}; at least two of each are needed.")
        _gap_b = (_matrix[_fifth == N_PARTS].mean(axis=0)
                  - _matrix[_fifth == 1].mean(axis=0))
        if screener == "arm3":
            _gap_c = _arm3_gap.loc[occ].reindex(_index).to_numpy(float)
        else:
            _gap_c = (_matrix[_female == 1].mean(axis=0)
                      - _matrix[_female == 0].mean(axis=0))
        _fifth_of_ad = _ads.set_index("jd_idx").loc[_index, "jd_fifth"].to_numpy(int)
        for _k in range(len(_index)):
            rows.append({"occ": occ, "jd_idx": int(_index[_k]),
                         "jd_fifth": int(_fifth_of_ad[_k]),
                         "b": float(_gap_b[_k]), "c": float(_gap_c[_k])})
    return pd.DataFrame(rows)


def _edited_per_ad(screener):
    """Panel (a) per advertisement: the mean change of the biographies the
    communal manipulation changed, at the gate."""
    if screener == "arm1":
        return (df_jd_level[["occ", "jd_idx", "dj_pool_communion_rank"]]
                .rename(columns={"dj_pool_communion_rank": "a"}))
    if screener == "arm2":
        return (df_arm2_jd_level[["occ", "jd_idx", "dj_pool_communion_rank"]]
                .rename(columns={"dj_pool_communion_rank": "a"}))
    if not ARM3_PCT_BY_JD.exists():
        print(f"panel (a) for Arm 3: not available, {ARM3_PCT_BY_JD.name} is "
              f"not among the outputs of notebook 07")
        return None
    _per_pair = pd.read_csv(ARM3_PCT_BY_JD)
    _per_pair = _per_pair[_per_pair["arm"] == "communion"].copy()
    _per_pair["jd_idx"] = [int(str(_j).rsplit("jd", 1)[1])
                           for _j in _per_pair["jd_id"]]
    _selected = _per_pair.merge(_changed_gate[["occ", "bio_idx"]],
                                on=["occ", "bio_idx"])
    print(f"panel (a) for Arm 3: biographies "
          f"{_selected[['occ', 'bio_idx']].drop_duplicates().shape[0]} of "
          f"{_per_pair[['occ', 'bio_idx']].drop_duplicates().shape[0]} rated, "
          f"pairs {len(_selected)}")
    return (_selected.groupby(["occ", "jd_idx"])["delta_pct"].mean()
            .reset_index().rename(columns={"delta_pct": "a"}))


def _summarise(frame, column, screener, panel):
    """Per fifth: the mean over the occupations, which the figure draws, and
    the mean over the advertisements, which the controls compare."""
    _by_occ = frame.groupby(["occ", "jd_fifth"])[column].mean().reset_index()
    rows = []
    for _part in range(1, N_PARTS + 1):
        _equal = _by_occ[_by_occ["jd_fifth"] == _part][column]
        _over_ads = frame[frame["jd_fifth"] == _part][column]
        rows.append({"screener": screener, "panel": panel, "jd_fifth": _part,
                     "mean_occupations_equal": float(_equal.mean()),
                     "mean_over_ads": float(_over_ads.mean()),
                     "n_ads": int(len(_over_ads)),
                     "n_occupations": int(_equal.notna().sum())})
    return pd.DataFrame(rows)


_tables = []
for _screener, _label in SCREENERS:
    _parts = _arm3_parts if _screener == "arm3" else _jd_parts
    _gaps = _gaps_per_ad(_screener, _parts)
    _tables.append(_summarise(_gaps, "b", _screener, "b"))
    _tables.append(_summarise(_gaps, "c", _screener, "c"))
    _edited = _edited_per_ad(_screener)
    if _edited is not None:
        _edited = _edited.merge(_parts[["occ", "jd_idx", "jd_fifth"]],
                                on=["occ", "jd_idx"])
        _tables.append(_summarise(_edited, "a", _screener, "a"))
df_fig5 = pd.concat(_tables, ignore_index=True).sort_values(
    ["screener", "panel", "jd_fifth"]).reset_index(drop=True)

# Control: panels (b) and (c) of Arm 1 are the ones of the two saved tables.
_controls = [(MEETING_DIR / "fig_gaps_values.csv", "T2", "b"),
             (MEETING_DIR / "fig_gaps_values.csv", "T1", "c")]
_worst_fig5 = 0.0
for _path, _test, _panel in _controls:
    if not _path.exists():
        raise FileNotFoundError(
            f"{_path} not found. Panels (b) and (c) of Arm 1 are controlled "
            f"against it.")
    _saved = pd.read_csv(_path)
    _saved = _saved[(_saved["screener"] == "SBERT")
                    & (_saved["test"] == _test)].sort_values("jd_fifth")
    _here = df_fig5[(df_fig5["screener"] == "arm1")
                    & (df_fig5["panel"] == _panel)].sort_values("jd_fifth")
    for _column in ("mean_occupations_equal", "mean_over_ads"):
        _worst_fig5 = max(_worst_fig5, float(np.abs(
            _here[_column].to_numpy() - _saved[_column].to_numpy()).max()))
_quintiles = pd.read_csv(MEETING_DIR / "t1_gap_by_balance_quintile.csv")
_quintiles = _quintiles[_quintiles["level"] == "pooled"].sort_values("quintile")
_here = df_fig5[(df_fig5["screener"] == "arm1")
                & (df_fig5["panel"] == "c")].sort_values("jd_fifth")
_worst_fig5 = max(_worst_fig5, float(np.abs(
    _here["mean_over_ads"].to_numpy()
    - _quintiles["mean_gap_f_minus_m"].to_numpy()).max()))
_count_gap = int(np.abs(_here["n_ads"].to_numpy()
                        - _quintiles["n_jds"].to_numpy()).max())
print()
print(f"control panels (b) and (c) of Arm 1 against fig_gaps_values.csv and "
      f"t1_gap_by_balance_quintile.csv: largest deviation {_worst_fig5:.3g} "
      f"(expected below 1e-06), largest deviation of the counts {_count_gap}")
if _worst_fig5 > 1e-6 or _count_gap:
    raise AssertionError(
        f"Panels (b) and (c) of Arm 1 deviate from the saved tables by "
        f"{_worst_fig5:.3g} and {_count_gap} advertisements.")

df_fig5.to_csv(FIG5_VALUES, index=False, encoding="utf-8")
print(f"written: {FIG5_VALUES} ({len(df_fig5)} rows)")
print()
print(df_fig5.round(4).to_string(index=False))
```

### 13.1 Figure 5: the experiment

- **Question:** By how much does the communal edit move a biography, and does
  that depend on how communal the advertisement is?
- **Expectation:** the ones of Table 2; the section draws the two quantities
  that table reports and fixes no direction of its own.
- **Status:** rewritten on 2026-11-12, revised on 2026-11-13 (one scale:
  communal balance) and on 2026-11-14: median in (a), axis labels. It
  replaces panel (a) of the old Figure 5. The median of panel (a) is
  controlled against Table 2, the mean it prints against the mean of the five
  points of panel (b), and panel (b) reproduces the values section 12 wrote
  for its panel (a).
- **Unit and test:** biography in panel (a) and panel (c), advertisement in
  panel (b); the median and the mean, no test. Panel (a) draws the median
  with a bootstrap interval, the quantity Table 2 reports, and prints next to
  it the mean of the same change over all advertisements with the occupations
  weighted equally, which is what panel (b) splits by fifth and what
  section 27 takes apart. Panel (b) cuts the fifths on the communal balance
  of the advertisement, the scale section 25 reports the test of H4a on, and
  carries no interval; its ticks name the fifth and the mean balance of the
  advertisements in it, the value section 19 prints with the occupations
  weighted equally. Both panels of the upper row mark one rank place in the
  ranking of an occupation in both directions.
- **Output:** `outputs/figures/fig6_experiment.png`.

```python
# Figure 5: the experiment, from the values of the section above and from the
# tables the three screeners wrote. Panel (a) holds the median change of the
# biographies the communal manipulation changed, the number Table 2 reports
# for H3; the mean of the same change over all advertisements with the
# occupations weighted equally, which is what panel (b) splits by fifth, is
# printed next to it but not drawn, and section 27 takes the distance between
# the two apart. Panel (b) holds the mean change by fifth of the advertisements
# on their
# communal balance, the scale on which the write-up reports the test of H4a;
# the fifths are the ones of section 12 and the occupations enter with equal
# weight. Panel (c) holds the median change of the fit slope of a biography
# per gender, which is H4b; the slope is the one of notebook 06
# section 14.15, and the three screeners are controlled against the medians
# notebook 06 section 14.15, section 9 and section 24 print. The two panels of
# the upper row share a y axis; panel (c) carries its own. Every interval is a
# percentile interval from 2,000 bootstrap draws with the seed of the run:
# panel (a) and panel (c) draw the biographies, panel (b) carries none. Both
# panels of the upper row mark one rank place in the ranking of an occupation
# as a dashed line, so a change can be read against the step the measure
# itself makes.
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

FIG6_PNG = Path("outputs/figures/fig6_experiment.png")
ARM3_DELTA_MAIN = ARM3_DIR / "study2_arm3_delta.csv"
ARM1_MANIP = "{occ}_communion_manip_pool_run5.npy"
COLOURS = {"arm1": "#2a78d6", "arm2": "#eb6834", "arm3": "#1baf7a"}
LABELS = {"arm1": "SBERT", "arm2": "Qwen3-Embedding", "arm3": "language model"}
# In the figures the language model is named LLM; the printed tables keep the
# long name.
FIG_LABELS = {"arm1": "SBERT", "arm2": "Qwen3-Embedding", "arm3": "LLM"}
TICKS_A = ("SBERT", "Qwen3-\nEmbedding", "LLM")
# Every panel over the fifths of the advertisements carries these ticks
# and this label, in Figure 5, Figure 6 and the figure of Appendix E. The
# panels are between 1.4 and 2.3 inches wide, so a tick that names the end
# of the scale underneath the number runs over the edge of its panel; the
# direction of the scale is therefore part of the label of the axis.
LABEL_FIFTHS = "Ads in fifths of communal balance within occupation"
# The label is longer than the panels it sits under are wide, so it is
# set one step smaller than the other labels of the axes, and the ticks, which
# carry two lines, one step smaller again.
LABEL_FIFTHS_SIZE = 5.4
TICKS_FIFTHS_SIZE = 5.2
INK, GREY, LIGHT = "#1a1a1a", "#52514e", "#b5b4ae"
N_BOOT = 2000
# The medians Table 2 reports for H3, to three decimals, and the medians of the
# change of the slope per gender, women first, as notebook 06 section 14.15,
# section 9 and section 24 print them. Both panels have to meet them.
MEDIANS_TABLE2 = {"arm1": 0.345, "arm2": -0.153, "arm3": -0.152}
MEDIANS_SLOPE = {"arm1": (-0.0102, -0.0030), "arm2": (-0.0201, -0.0117),
                 "arm3": (-0.0085, -0.0254)}
GENDERS = ((1, "women"), (0, "men"))

plt.rcParams["font.family"] = "serif"
plt.rcParams["mathtext.fontset"] = "dejavuserif"   # the Delta of the y label
if not ARM3_DELTA_MAIN.exists():
    raise FileNotFoundError(
        f"{ARM3_DELTA_MAIN} not found. Panel (a) reads the change per "
        f"biography of Arm 3 from the outputs of notebook 07.")

# One rank place in the ranking of an occupation. The percentile runs from 0
# to 100 over the biographies of the pool, so one place is 100 divided by
# their number less one. The two panels of the upper row mark it in both
# directions.
_pool_sizes = sorted(df_pool[df_pool["arm"] == "communion"]
                     .groupby("occ").size().unique())
if len(_pool_sizes) != 1:
    raise AssertionError(
        f"The occupations hold different numbers of biographies "
        f"{_pool_sizes}, so one rank place is not one distance.")
N_BIOS_POOL = int(_pool_sizes[0])
RANK_PLACE = 100.0 / (N_BIOS_POOL - 1)
print(f"one rank place among the {N_BIOS_POOL} biographies of an occupation: "
      f"{RANK_PLACE:.3f} percentile points")

# The ticks of every panel over the fifths name the fifth and, under it, the
# mean balance of the advertisements in it, with the occupations weighted
# equally: the quantity section 19 prints in the first table of its first
# block. Arm 3 cuts its fifths among the twenty advertisements it rated, so
# its means are printed next to the ones of the whole set the ticks carry.
_balance_fifth = (_jd_parts.groupby(["occ", "jd_fifth"])["balance"].mean()
                  .groupby("jd_fifth").mean())
_balance_fifth_arm3 = (_arm3_parts.groupby(["occ", "jd_fifth"])["balance"]
                       .mean().groupby("jd_fifth").mean())
TICKS_FIFTHS = tuple(f"{_k}\n({_balance_fifth.loc[_k]:+.1f})"
                     for _k in range(1, N_PARTS + 1))
print(f"mean balance per fifth with the occupations weighted equally: over "
      f"all {len(_jd_parts)} advertisements "
      f"{[round(float(_v), 3) for _v in _balance_fifth]}, over the "
      f"{len(_arm3_parts)} of Arm 3 "
      f"{[round(float(_v), 3) for _v in _balance_fifth_arm3]}")

# Panel (a): the change per biography of the changed biographies at the gate.
# Arm 1 and Arm 2 carry it in percentile points; of Arm 3, which rates on a
# scale of its own, the column in percentile points is the comparable one, and
# the biographies it could not place carry no percentile.
_a3_delta = pd.read_csv(ARM3_DELTA_MAIN)
_a3_delta = _a3_delta[(_a3_delta["arm"] == "communion")
                      & (_a3_delta["sim_cp"] >= PRIMARY_GATE)]
_bio_delta = {
    "arm1": df_pool[(df_pool["arm"] == "communion")
                    & (df_pool["sim_cp"] >= PRIMARY_GATE)
                    & (df_pool["n_subs"] >= 1)]["delta"].to_numpy(float),
    "arm2": df_arm2_delta_bio[df_arm2_delta_bio["arm"] == "communion"]
            ["delta"].to_numpy(float),
    "arm3": _a3_delta["delta_pct"].dropna().to_numpy(float),
}


def _boot_median(values, rng):
    """The median and its percentile interval over draws of the biographies."""
    _drawn = np.median(values[rng.integers(0, len(values),
                                           (N_BOOT, len(values)))], axis=1)
    return (float(np.median(values)), float(np.percentile(_drawn, 2.5)),
            float(np.percentile(_drawn, 97.5)))


_medians = {}
print(f"Panel (a): the median change of the changed biographies at the gate "
      f"{PRIMARY_GATE}, communal arm")
print(f"  {'screener':<16} {'n_bios':>7} {'median':>8} {'ci_lo':>8} "
      f"{'ci_hi':>8} {'Table 2':>8}")
print("  " + "-" * 60)
for _screener, _label in SCREENERS:
    _values = _bio_delta[_screener]
    _medians[_screener] = _boot_median(_values, np.random.default_rng(SEED))
    _median, _lo, _hi = _medians[_screener]
    print(f"  {_label:<16} {len(_values):>7} {_median:>+8.3f} {_lo:>+8.3f} "
          f"{_hi:>+8.3f} {MEDIANS_TABLE2[_screener]:>+8.3f}")
    if abs(_median - MEDIANS_TABLE2[_screener]) > 5e-4:
        raise AssertionError(
            f"The median of {_label} is {_median:+.4f} and does not meet the "
            f"{MEDIANS_TABLE2[_screener]:+.3f} of Table 2.")

# Panel (b): the fifths of section 12, cut within the occupation on the
# communal balance of the advertisement; Arm 3 cuts among the twenty
# advertisements it rated.
_per_cell = _arm3_parts.groupby(["occ", "jd_fifth"]).size().unique()
print()
print(f"Panel (b): advertisements per fifth of the communal balance, "
      f"{len(_jd_parts)} in all and {len(_arm3_parts)} for Arm 3, per "
      f"occupation and fifth {sorted(int(_n) for _n in _per_cell)}")


def _edited_per_ad_w24(screener):
    """The mean change per advertisement over the biographies the communal
    manipulation changed, at the gate. It is the quantity section 12 holds in
    its panel (a)."""
    if screener == "arm1":
        return (df_jd_level[["occ", "jd_idx", "dj_pool_communion_rank"]]
                .rename(columns={"dj_pool_communion_rank": "a"}))
    if screener == "arm2":
        return (df_arm2_jd_level[["occ", "jd_idx", "dj_pool_communion_rank"]]
                .rename(columns={"dj_pool_communion_rank": "a"}))
    _pairs = pd.read_csv(ARM3_PCT_BY_JD)
    _pairs = _pairs[_pairs["arm"] == "communion"].copy()
    _pairs["jd_idx"] = [int(str(_j).rsplit("jd", 1)[1])
                        for _j in _pairs["jd_id"]]
    _pairs = _pairs.merge(_changed_gate[["occ", "bio_idx"]],
                          on=["occ", "bio_idx"])
    return (_pairs.groupby(["occ", "jd_idx"])["delta_pct"].mean()
            .reset_index().rename(columns={"delta_pct": "a"}))


def _mean_fifths(frame, column):
    """Per fifth the mean over the occupations, which the figure draws, and the
    number of advertisements behind it."""
    _means = np.empty(N_PARTS)
    _counts = np.zeros(N_PARTS, dtype=int)
    for _k, _part in enumerate(range(1, N_PARTS + 1)):
        _per_occ = []
        for occ in KEPT_13:
            _values = frame[(frame["occ"] == occ)
                            & (frame["jd_fifth"] == _part)][column]
            _values = _values.to_numpy(float)
            if not len(_values):
                continue
            _counts[_k] += len(_values)
            _per_occ.append(_values.mean())
        if not _per_occ:
            raise AssertionError(
                f"No occupation has an advertisement in fifth {_part}, so the "
                f"figure cannot be drawn.")
        _means[_k] = float(np.mean(_per_occ))
    return _means, _counts


_fifth_values, _overall_means = {}, {}
for _screener, _label in SCREENERS:
    _parts = _arm3_parts if _screener == "arm3" else _jd_parts
    _frame_b = _edited_per_ad_w24(_screener).merge(
        _parts[["occ", "jd_idx", "jd_fifth"]], on=["occ", "jd_idx"])
    # Control: per fifth these are the values section 12 wrote for its panel
    # (a), which rests on the same cut.
    _saved_a = (df_fig5[(df_fig5["screener"] == _screener)
                        & (df_fig5["panel"] == "a")].sort_values("jd_fifth"))
    _worst_a = float(np.abs(
        _summarise(_frame_b, "a", _screener, "a")
        ["mean_occupations_equal"].to_numpy()
        - _saved_a["mean_occupations_equal"].to_numpy()).max())
    if _worst_a > 1e-9:
        raise AssertionError(
            f"Panel (b) of {_label} does not reproduce the values of "
            f"section 12; it deviates by {_worst_a:.3g}.")
    _fifth_values[_screener] = _mean_fifths(_frame_b, "a")
    # The same quantity without the split by fifth: the mean change per
    # advertisement with the occupations weighted equally. Panel (a) draws the
    # median alone, so the mean is printed below and not drawn.
    _overall_means[_screener] = float(np.mean(
        [_frame_b[_frame_b["occ"] == occ]["a"].mean() for occ in KEPT_13]))
print(f"control panel (b) against section 12: all three screeners reproduce "
      f"the values it wrote for its panel (a)")

print()
print(f"Panel (b): the mean change per fifth of the communal balance of the "
      f"advertisement")
print(f"  {'screener':<16} {'fifth':>6} {'n_ads':>6} {'mean':>8}")
print("  " + "-" * 40)
for _screener, _label in SCREENERS:
    _means, _counts = _fifth_values[_screener]
    for _k in range(N_PARTS):
        print(f"  {_label:<16} {_k + 1:>6} {_counts[_k]:>6} "
              f"{_means[_k]:>+8.3f}")

print()
print(f"Panel (a): the median it draws, and the mean of the same change next "
      f"to the mean of the five points of panel (b)")
print(f"  {'screener':<16} {'median':>8} {'mean':>8} {'mean of (b)':>12} "
      f"{'difference':>11}")
print("  " + "-" * 60)
for _screener, _label in SCREENERS:
    _mean_overall = _overall_means[_screener]
    _mean_of_five = float(_fifth_values[_screener][0].mean())
    print(f"  {_label:<16} {_medians[_screener][0]:>+8.3f} "
          f"{_mean_overall:>+8.3f} {_mean_of_five:>+12.3f} "
          f"{_mean_overall - _mean_of_five:>+11.4f}")
    if abs(_mean_overall - _mean_of_five) > 0.01:
        raise AssertionError(
            f"The mean of {_label} is {_mean_overall:+.4f} and the mean of "
            f"the five points of panel (b) is {_mean_of_five:+.4f}; they are "
            f"more than 0.01 apart, so the two panels do not carry the same "
            f"quantity.")

# Panel (c): the change of the fit slope of a biography. The slope is the one
# of notebook 06 section 14.15: the percentile rank of the biography over the
# advertisements of its occupation against their balance, centred within the
# occupation. For Arm 1 the embeddings notebook 06 cached carry the ranks after
# the edit, for Arm 2 the embeddings of notebook 06a, and for Arm 3 the table
# notebook 07 wrote. Nothing is embedded here.


def _slope_of(matrix, balance):
    """The slope of each row of a matrix of percentile ranks over the
    advertisements, against their balance centred within the occupation."""
    _centred = balance - balance.mean()
    _denominator = float(_centred @ _centred)
    if _denominator < 1e-12:
        raise AssertionError(
            "The balance of the advertisements does not vary, so no slope can "
            "be formed on it.")
    return ((matrix - matrix.mean(axis=1, keepdims=True)) @ _centred
            / _denominator)


def _delta_slopes(screener):
    """Per changed biography of the gate the change of its fit slope."""
    if screener == "arm3":
        _pairs = pd.read_csv(ARM3_PCT_BY_JD)
        _pairs = _pairs[_pairs["arm"] == "communion"].copy()
        _pairs["jd_idx"] = [int(str(_j).rsplit("jd", 1)[1])
                            for _j in _pairs["jd_id"]]
        _pairs = _pairs.merge(_changed_gate[["occ", "bio_idx"]],
                              on=["occ", "bio_idx"])
        _rows = []
        for occ in KEPT_13:
            _here = _pairs[_pairs["occ"] == occ]
            _balance = (_jd_parts[_jd_parts["occ"] == occ]
                        .set_index("jd_idx")["balance"])
            _slopes = {}
            for _column, _name in (("pct_before", "before"),
                                   ("pct_after", "after")):
                _wide = _here.pivot_table(index="bio_idx", columns="jd_idx",
                                          values=_column)
                if _wide.isna().to_numpy().any():
                    raise AssertionError(
                        f"{occ}: a biography was not ranked for every "
                        f"advertisement, so its slope is not the one of "
                        f"notebook 06 section 14.15.")
                _slopes[_name] = _slope_of(
                    _wide.to_numpy(float),
                    _balance.reindex(_wide.columns).to_numpy(float))
            for _k, _bio_idx in enumerate(_wide.index):
                _rows.append({"occ": occ, "bio_idx": int(_bio_idx),
                              "before": float(_slopes["before"][_k]),
                              "after": float(_slopes["after"][_k])})
        _frame = pd.DataFrame(_rows)
        _frame["delta_s"] = _frame["after"] - _frame["before"]
        return _frame.merge(df_bio[["occ", "bio_idx", "female"]],
                            on=["occ", "bio_idx"], how="inner")
    _rows = []
    for occ in KEPT_13:
        _balance = (df_jd_level[df_jd_level["occ"] == occ]
                    .sort_values("jd_idx")["balance"].to_numpy(float))
        if screener == "arm1":
            _jd_cache = np.load(ARM1_CACHE / f"{occ}_jds.npy",
                                allow_pickle=True).item()
            _orig = np.load(ARM1_CACHE / f"{occ}_pool_orig.npy",
                            allow_pickle=True).item()["mean"]
            _manip = np.load(ARM1_CACHE / ARM1_MANIP.format(occ=occ),
                             allow_pickle=True).item()["mean"]
            _so = study2_core.jd_similarity(_orig, _jd_cache, how=JD_REPR)
            _sm = study2_core.jd_similarity(_manip, _jd_cache, how=JD_REPR)
        else:
            _so = emb_orig[occ] @ emb_jd[occ].T
            _sm = emb_manip[occ]["communion"] @ emb_jd[occ].T
        _before = _slope_of(study2_core.pool_pct_per_jd(_so), _balance)
        _after = _slope_of(study2_core.indiv_cf_pct_per_jd(_so, _sm), _balance)
        for _bio_idx in (_changed_gate[_changed_gate["occ"] == occ]["bio_idx"]
                         .to_numpy(int)):
            _rows.append({"occ": occ, "bio_idx": int(_bio_idx),
                          "before": float(_before[_bio_idx]),
                          "after": float(_after[_bio_idx])})
    _frame = pd.DataFrame(_rows)
    _frame["delta_s"] = _frame["after"] - _frame["before"]
    return _frame.merge(df_bio[["occ", "bio_idx", "female", "s_bal"]],
                        on=["occ", "bio_idx"], how="inner")


_slope_change = {}
for _screener, _label in SCREENERS:
    _frame_c = _delta_slopes(_screener)
    if _screener == "arm1":
        # Control: the slope before the edit is the one notebook 06 published.
        _worst_c = float(np.abs(_frame_c["before"] - _frame_c["s_bal"]).max())
        print()
        print(f"control the slope of Arm 1 before the edit against "
              f"study2_fit_bio_slopes.csv: largest deviation {_worst_c:.3g} "
              f"(expected below 1e-09)")
        if _worst_c > 1e-9:
            raise AssertionError(
                f"The slopes of Arm 1 read from the cache deviate from the "
                f"published ones by {_worst_c:.3g}.")
    _slope_change[_screener] = {}
    for _value, _gender in GENDERS:
        _side = _frame_c[_frame_c["female"] == _value]["delta_s"].dropna()
        _slope_change[_screener][_gender] = (
            _side.to_numpy(float),
            _boot_median(_side.to_numpy(float), np.random.default_rng(SEED)))

print()
print(f"Panel (c): the median change of the fit slope after the edit, "
      f"communal arm, gate {PRIMARY_GATE}")
print(f"  {'screener':<16} {'group':<7} {'n_bios':>7} {'median':>9} "
      f"{'ci_lo':>9} {'ci_hi':>9} {'printed':>9}")
print("  " + "-" * 72)
for _screener, _label in SCREENERS:
    for _position, (_value, _gender) in enumerate(GENDERS):
        _values, (_median, _lo, _hi) = _slope_change[_screener][_gender]
        _printed = MEDIANS_SLOPE[_screener][_position]
        print(f"  {_label:<16} {_gender:<7} {len(_values):>7} "
              f"{_median:>+9.4f} {_lo:>+9.4f} {_hi:>+9.4f} {_printed:>+9.4f}")
        if abs(_median - _printed) > 5e-5:
            raise AssertionError(
                f"The median change of the slope of {_label} for {_gender} is "
                f"{_median:+.5f} and does not meet the {_printed:+.4f} that "
                f"notebook 06 section 14.15, section 9 and section 24 print.")

# 13 cm at 300 dpi, the width of every figure of the thesis
# except Figures 1 and 4.
fig6 = plt.figure(figsize=(5.12, 3.3))
_grid6 = fig6.add_gridspec(2, 2, height_ratios=(1.0, 0.95), hspace=0.62,
                           wspace=0.1)
ax_a = fig6.add_subplot(_grid6[0, 0])
ax_b = fig6.add_subplot(_grid6[0, 1], sharey=ax_a)
ax_c = fig6.add_subplot(_grid6[1, :])
_x6 = np.arange(1, N_PARTS + 1)
_positions = np.arange(1, len(SCREENERS) + 1)
for _position, (_screener, _label) in zip(_positions, SCREENERS):
    _median, _lo, _hi = _medians[_screener]
    ax_a.errorbar([_position], [_median],
                  yerr=[[_median - _lo], [_hi - _median]], fmt="o",
                  color=COLOURS[_screener], ms=2.8, elinewidth=0.7,
                  capsize=1.6, label=FIG_LABELS[_screener])
for _screener, _label in SCREENERS:
    _means, _counts = _fifth_values[_screener]
    ax_b.plot(_x6, _means, "-", marker="o", color=COLOURS[_screener],
              lw=1.1, ms=2.6)
for _position, (_screener, _label) in zip(_positions, SCREENERS):
    for _offset, (_value, _gender) in zip((-0.12, 0.12), GENDERS):
        _values, (_median, _lo, _hi) = _slope_change[_screener][_gender]
        ax_c.errorbar([_position + _offset], [_median],
                      yerr=[[_median - _lo], [_hi - _median]], fmt="o",
                      color=COLOURS[_screener],
                      markerfacecolor=(COLOURS[_screener] if _value == 1
                                       else "white"),
                      ms=2.8, elinewidth=0.7, capsize=1.6)
ax_a.set_title("(a) H3: change in rank, median", fontsize=6.6, color=INK,
               pad=3)
ax_a.set_xlim(0.5, len(SCREENERS) + 0.5)
ax_a.set_xticks(_positions)
ax_a.set_xticklabels(TICKS_A, fontsize=5.6)
ax_b.set_title("(b) H4a: by the ad's communal balance, mean", fontsize=6.6,
               color=INK, pad=3)
ax_b.set_xticks(_x6)
ax_b.set_xticklabels(TICKS_FIFTHS, fontsize=TICKS_FIFTHS_SIZE)
ax_b.set_xlabel(LABEL_FIFTHS, fontsize=LABEL_FIFTHS_SIZE, color=INK)
ax_b.tick_params(labelleft=False)
ax_c.set_title("(c) H4b: change in fit slope, women's and men's bios, median",
               fontsize=6.6, color=INK, pad=3)
ax_c.set_xlim(0.5, len(SCREENERS) + 0.5)
ax_c.set_xticks(_positions)
ax_c.set_xticklabels([FIG_LABELS[_s] for _s, _ in SCREENERS], fontsize=5.6)
for _ax in (ax_a, ax_b, ax_c):
    _ax.axhline(0, color=GREY, lw=0.7, zorder=1)
    _ax.tick_params(axis="both", labelsize=6.0, colors=GREY, length=2)
    _ax.grid(axis="y", color="#ecebe7", lw=0.5)
    _ax.set_axisbelow(True)
    for _side in ("top", "right"):
        _ax.spines[_side].set_visible(False)
    for _side in ("bottom", "left"):
        _ax.spines[_side].set_color(LIGHT)
# One rank place in both directions, in the two panels that hold a change in
# the percentile rank. The label sits at the right edge of panel (b).
for _ax in (ax_a, ax_b):
    for _sign in (1, -1):
        _ax.axhline(_sign * RANK_PLACE, color=LIGHT, lw=0.6, ls="--",
                    zorder=1)
ax_b.text(0.985, RANK_PLACE, r"$\pm$1 rank place", fontsize=5.0, color=GREY,
          ha="right", va="bottom", transform=ax_b.get_yaxis_transform())
ax_a.set_ylabel(r"Change in percentile rank ($\Delta$)", fontsize=5.8,
                color=INK)
ax_c.set_ylabel("Change in fit slope", fontsize=5.8, color=INK)
ax_c.legend(handles=[Line2D([], [], marker="o", color=GREY, lw=0,
                            markerfacecolor=GREY, ms=2.8,
                            label="women's bios"),
                     Line2D([], [], marker="o", color=GREY, lw=0,
                            markerfacecolor="white", ms=2.8,
                            label="men's bios")],
            loc="lower left", fontsize=5.6, frameon=False, handlelength=1.0,
            handletextpad=0.3, labelspacing=0.2)

_handles6, _labels6 = ax_a.get_legend_handles_labels()
fig6.legend(_handles6, _labels6, loc="lower center", ncol=len(_labels6),
            fontsize=5.8, frameon=False, handlelength=1.4, handletextpad=0.4,
            columnspacing=1.4, bbox_to_anchor=(0.5, 0.0))
# The margins are set by hand. A gridspec whose panels span different numbers
# of columns does not fit tight_layout, which warns and then guesses.
fig6.subplots_adjust(left=0.1, right=0.985, top=0.93, bottom=0.1,
                     hspace=0.95, wspace=0.08)
FIG6_PNG.parent.mkdir(parents=True, exist_ok=True)
fig6.savefig(FIG6_PNG, dpi=300)
plt.show()
print()
print(f"written: {FIG6_PNG}")
```

### 13.2 Figure 6: style fit and the gender gap in the original rankings

- **Question:** How far apart do the biographies sit in the ranking before
  anything is edited, and does that distance grow with the wording of the
  advertisement?
- **Expectation:** the ones of T1 and T2; both read how the lead changes over
  the fifths, and the panels show the lead itself, so its height can be read
  as well as its course.
- **Status:** rewritten on 2026-11-12, revised on 2026-11-13 (one scale:
  communal balance) and on 2026-11-14: median in (a), axis labels; on the
  same day both panels moved from the change since the first fifth to the
  lead itself. It replaces panels (b) and (c) of the old Figure 5. The values
  per advertisement are controlled against section 12, whose gap under the
  language model is the one of notebook 07.
- **Unit and test:** advertisement; the lead per fifth, no test. The two
  panels share one y axis and the zero line, their ticks name the fifth and
  the mean balance of the advertisements in it, and the change from the most
  agentic to the most communal fifth, which the two tests are about, is
  printed under the leads.
- **Output:** `outputs/figures/fig7_original_rankings.png`.

```python
# Figure 6: style fit and the gender gap in the original rankings, with no edit
# anywhere. Panel (a) is the lead
# of the biographies in the most communal fifth of their own balance over those
# in the most agentic fifth, which is what T2 reads; panel (b) is the lead of
# the biographies of women over those of men, which is what T1 reads. Both
# panels show that lead itself over the fifths of the advertisements, on one y
# axis and with the zero line, so the height of a line can be read as well as
# its course. The values per advertisement are the ones of section 12 and the
# fifths are its fifths of the balance; the gap of the language model in
# panel (b) is the one notebook 07 fixed in advance. The change from the most
# agentic to the most communal fifth, which the two tests are about, is printed
# under the leads.
FIG7_PNG = Path("outputs/figures/fig7_original_rankings.png")
# The titles run over two lines, so that the name of the test and the
# quantity it reads both fit over a panel of half the width of the figure.
PANELS_7 = [("b", "(a) T2, H4a: lead of the most communal\n"
                  "over the most agentic fifth of bios"),
            ("c", "(b) T1, H4b: lead of women's bios\n"
                  "over men's bios")]

_levels_7 = {}
for _panel, _title in PANELS_7:
    for _screener, _label in SCREENERS:
        _parts = _arm3_parts if _screener == "arm3" else _jd_parts
        _gaps_7 = _gaps_per_ad(_screener, _parts)
        # Control: per fifth these are the values section 12 wrote.
        _saved_7 = (df_fig5[(df_fig5["screener"] == _screener)
                            & (df_fig5["panel"] == _panel)]
                    .sort_values("jd_fifth"))
        _worst_7 = float(np.abs(
            _summarise(_gaps_7, _panel, _screener, _panel)
            ["mean_occupations_equal"].to_numpy()
            - _saved_7["mean_occupations_equal"].to_numpy()).max())
        if _worst_7 > 1e-9:
            raise AssertionError(
                f"Panel {_panel} of {_label} does not reproduce the values of "
                f"section 12; it deviates by {_worst_7:.3g}.")
        _levels_7[(_panel, _screener)] = _mean_fifths(_gaps_7, _panel)
print(f"control the gaps per advertisement against section 12: all three "
      f"screeners reproduce both panels")

for _panel, _title in PANELS_7:
    print()
    print(f"{_title}")
    print(f"  {'screener':<16} {'fifth':>6} {'n_ads':>6} {'lead':>8}")
    print("  " + "-" * 40)
    for _screener, _label in SCREENERS:
        _means, _counts = _levels_7[(_panel, _screener)]
        for _k in range(N_PARTS):
            print(f"  {_label:<16} {_k + 1:>6} {_counts[_k]:>6} "
                  f"{_means[_k]:>+8.3f}")
    print(f"  the lead in the most agentic and in the most communal fifth, "
          f"and the change between them")
    for _screener, _label in SCREENERS:
        _means, _counts = _levels_7[(_panel, _screener)]
        print(f"  {_label:<16} {_means[0]:>+8.3f} to {_means[-1]:>+8.3f}, "
              f"change {_means[-1] - _means[0]:>+8.3f}")

print()
for _screener, _label in SCREENERS:
    _means, _counts = _levels_7[("c", _screener)]
    _positive = "yes" if bool((_means > 0).all()) else "no"
    print(f"T1 gap positive in every fifth: {_positive} ({_label}, from "
          f"{_means.min():+.3f} to {_means.max():+.3f} percentile points)")

# 13 cm at 300 dpi, the width of every figure of the thesis
# except Figures 1 and 4.
fig7, axes7 = plt.subplots(1, 2, figsize=(5.12, 2.1), sharey=True)
_x7 = np.arange(1, N_PARTS + 1)
for _ax, (_panel, _title) in zip(axes7, PANELS_7):
    for _screener, _label in SCREENERS:
        _means, _counts = _levels_7[(_panel, _screener)]
        _ax.plot(_x7, _means, "-", marker="o",
                 color=COLOURS[_screener], lw=1.1, ms=2.6,
                 label=FIG_LABELS[_screener])
    _ax.axhline(0, color=GREY, lw=0.7, zorder=1)
    _ax.set_title(_title, fontsize=6.6, color=INK, pad=3, linespacing=1.5)
    _ax.set_xticks(_x7)
    _ax.set_xticklabels(TICKS_FIFTHS, fontsize=TICKS_FIFTHS_SIZE)
    _ax.set_xlabel(LABEL_FIFTHS, fontsize=LABEL_FIFTHS_SIZE, color=INK)
    _ax.tick_params(axis="both", labelsize=6.0, colors=GREY, length=2)
    _ax.grid(axis="y", color="#ecebe7", lw=0.5)
    _ax.set_axisbelow(True)
    for _side in ("top", "right"):
        _ax.spines[_side].set_visible(False)
    for _side in ("bottom", "left"):
        _ax.spines[_side].set_color(LIGHT)
axes7[0].set_ylabel("Lead in percentile points", fontsize=5.8, color=INK)

_handles7, _labels7 = axes7[0].get_legend_handles_labels()
fig7.legend(_handles7, _labels7, loc="lower center", ncol=len(SCREENERS),
            fontsize=5.8, frameon=False, handlelength=1.4, handletextpad=0.4,
            columnspacing=1.4, bbox_to_anchor=(0.5, 0.0))
# The margins are set by hand, as in Figure 5, so that the two figures carry
# the same ticks at the same size. The titles and the ticks run over two lines
# each, so the panels sit lower and start higher than in Figure 5.
fig7.subplots_adjust(left=0.105, right=0.985, top=0.84, bottom=0.3,
                     wspace=0.08)
FIG7_PNG.parent.mkdir(parents=True, exist_ok=True)
fig7.savefig(FIG7_PNG, dpi=300)
plt.show()
print()
print(f"written: {FIG7_PNG}")
```

## 14. The signal before and after the substitution

- **Question:** By how much does the substitution lower the dimension it
  treats, and does it leave the other dimension where it was?
- **Expectation:** the treated dimension falls and the other one stays; the
  direction follows from the protocol and is not a test.
- **Status:** added on 2026-11-07, descriptive, no test. Section 7 of notebook
  06 prints the median of the change; this section prints the levels that
  change is the difference of, from the table that section wrote.
- **Unit and test:** biography; medians and means over the biographies of the
  pool the manipulation changed, no gate, as in section 7 of notebook 06.
- **Output:** `study2_signal_levels.csv`.

```python
# The level of both dimensions before and after the substitution. Notebook 06
# section 7 prints the median of the change; this section prints the levels it
# is the difference of, from the table that section wrote. The rows are the
# biographies of the pool that the manipulation changed; no gate is applied,
# as in section 7 of notebook 06.
SIGNAL_FILE = Path(OUT_BASE) / "study2_signal_levels.csv"
MANIP_CHECK_FILE = Path(OUT_BASE) / "manipulation_check.csv"
if not MANIP_CHECK_FILE.exists():
    raise FileNotFoundError(
        f"{MANIP_CHECK_FILE} not found. It is written by section 7 of "
        f"notebook 06 and carries the rate of both dimensions per biography.")

df_manip_check = pd.read_csv(MANIP_CHECK_FILE)
_pool_check = df_manip_check[df_manip_check["sample"] == "pool"]
_changed_check = _pool_check[_pool_check["changed"]]
print(f"pool records of the manipulation check: {len(_pool_check)}, "
      f"changed by the manipulation: {len(_changed_check)}")

# Control: the changed biographies are the ones notebook 06 counted at least
# one substitution for.
_with_subs = df_pool.merge(
    _pool_check[["occ", "bio_idx", "arm", "changed"]],
    on=["occ", "bio_idx", "arm"], how="inner")
_disagree = int((_with_subs["changed"] != (_with_subs["n_subs"] >= 1)).sum())
print(f"control biographies whose 'changed' differs from n_subs >= 1: "
      f"{_disagree} (expected 0)")
if _disagree:
    raise AssertionError(
        f"{_disagree} biographies are marked as changed although notebook 06 "
        f"counted no substitution, or the other way round.")

_signal_rows = []
for arm in ARMS:
    _sub = _changed_check[_changed_check["arm"] == arm]
    _signal_rows.append({
        "arm": arm, "dimension": "treated", "n": len(_sub),
        "median_before": float(_sub["rate_before"].median()),
        "median_after": float(_sub["rate_after"].median()),
        "mean_before": float(_sub["rate_before"].mean()),
        "mean_after": float(_sub["rate_after"].mean()),
        "median_change": float(_sub["delta_rate"].median()),
        "mean_change": float(_sub["delta_rate"].mean())})
    _signal_rows.append({
        "arm": arm, "dimension": "other", "n": len(_sub),
        "median_before": float(_sub["rate_other_before"].median()),
        "median_after": float(_sub["rate_other_after"].median()),
        "mean_before": float(_sub["rate_other_before"].mean()),
        "mean_after": float(_sub["rate_other_after"].mean()),
        "median_change": float(_sub["delta_rate_other"].median()),
        "mean_change": float(_sub["delta_rate_other"].mean())})
df_signal_levels = pd.DataFrame(_signal_rows)

# Control: the level after minus the level before is the change the table
# carries, so the two ways of reading it agree.
_worst_signal = float(np.abs(
    (df_signal_levels["mean_after"] - df_signal_levels["mean_before"])
    - df_signal_levels["mean_change"]).max())
print(f"control mean after minus mean before against the stored change: "
      f"{_worst_signal:.3g} (expected below 1e-09)")
if _worst_signal > 1e-9:
    raise AssertionError(
        f"The levels and the change of the manipulation check disagree by "
        f"{_worst_signal:.3g}.")

print()
print(f"Rate of markers per 100 words, biographies of the pool the "
      f"manipulation changed")
print(f"  {'arm':<10} {'dimension':<10} {'n':>5} {'median before':>14} "
      f"{'median after':>13} {'mean before':>12} {'mean after':>11}")
print("  " + "-" * 79)
for _row in df_signal_levels.itertuples():
    print(f"  {_row.arm:<10} {_row.dimension:<10} {_row.n:>5} "
          f"{_row.median_before:>14.3f} {_row.median_after:>13.3f} "
          f"{_row.mean_before:>12.3f} {_row.mean_after:>11.3f}")

df_signal_levels.to_csv(SIGNAL_FILE, index=False, encoding="utf-8")
print()
print(f"written: {SIGNAL_FILE} ({len(df_signal_levels)} rows)")
```

## 15. Dimension separation for Arm 2 and Arm 3

- **Question:** Does lowering one dimension move the ranking through that
  dimension rather than through the other one, under the other two screeners?
- **Expectation:** the ones of notebook 06 section 14.3; the matching
  dimension carries the change.
- **Status:** added on 2026-11-07. The section repeats a test of notebook 06
  on Arm 2 and on Arm 3 without changing its specification; Arm 1 is repeated
  as a control and has to reproduce `study2_ols.csv`.
- **Unit and test:** biography; OLS of the change on the number of
  substitutions, the rate of the other dimension and the log token count,
  with occupation effects and HC1, at the gate of the main analysis.
- **Output:** `study2_dimension_separation_by_screener.csv`; the rows per
  replaced marker of Table E5.

```python
# Does the manipulated dimension carry the change under the other screeners?
# The specification is the one of notebook 06 section 14.3, unchanged: the
# change of a biography on the number of substitutions, the rate of the other
# dimension and the log token count, with occupation effects and HC1. Section
# 14.3 runs it on the change of Arm 1, which is the column delta of
# study2_delta_r5_pool.csv and comes from the SBERT embeddings notebook 06
# cached. Arm 1 is repeated here as a control, Arm 2 and Arm 3 are new.
ARM3_DELTA_FILE = Path(OUT_BASE) / "arm3" / "main" / "study2_arm3_delta.csv"
OLS_ARM1_FILE = Path(OUT_BASE) / "study2_ols.csv"
DIMSEP_FILE = Path(OUT_BASE) / "study2_dimension_separation_by_screener.csv"
for _path in (ARM3_DELTA_FILE, OLS_ARM1_FILE):
    if not _path.exists():
        raise FileNotFoundError(
            f"{_path} not found. This section reads the change per biography "
            f"of Arm 3 and the regression of notebook 06 section 14.3.")

# The rates and the token count belong to the biography and not to the
# screener, so all three encoders take them from the same table.
_bio_features = df_bio[["occ", "bio_idx", "word_count", "communion_rate",
                        "agency_rate"]]
# Arm 3 carries two changes per biography: delta on the scale of the rating
# and delta_pct in percentile points. Sections 14.2 and 14.3 are in percentile
# points, so delta_pct is the column that answers the same question.
_arm3_delta = pd.read_csv(ARM3_DELTA_FILE)
_pct_gap = float(np.abs(_arm3_delta["pct_after"] - _arm3_delta["pct_before"]
                        - _arm3_delta["delta_pct"]).max())
print(f"control Arm 3: delta_pct against the rank after minus the rank before:"
      f" {_pct_gap:.3g} (expected below 1e-09)")
if _pct_gap > 1e-9:
    raise AssertionError(
        f"delta_pct of Arm 3 is not the change in percentile rank; it "
        f"deviates by {_pct_gap:.3g}.")
_arm3_delta = _arm3_delta.rename(columns={"delta": "delta_rating",
                                          "delta_pct": "delta"})
_no_rank = int(_arm3_delta["delta"].isna().sum())
print(f"Arm 3 records: {len(_arm3_delta)}, without a percentile rank "
      f"{_no_rank}; these cannot enter a test in percentile points and are "
      f"left out")
_arm3_delta = _arm3_delta.dropna(subset=["delta"])
DELTA_SOURCES = [
    ("arm1", "SBERT", df_pool[["occ", "bio_idx", "arm", "delta", "n_subs",
                               "sim_cp"]]),
    ("arm2", "Qwen3-Embedding", df_arm2_pool[["occ", "bio_idx", "arm", "delta",
                                              "n_subs", "sim_cp"]]),
    ("arm3", "language model", _arm3_delta[["occ", "bio_idx", "arm", "delta",
                                            "sim_cp"]].merge(
        df_pool[["occ", "bio_idx", "arm", "n_subs"]],
        on=["occ", "bio_idx", "arm"], how="left")),
]

_dimsep_rows = []
print(f"OLS of notebook 06 section 14.3 (gate={PRIMARY_GATE}): "
      f"delta ~ n_subs + other_rate + log_wc + C(occ), HC1")
for _screener, _label, _frame in DELTA_SOURCES:
    _missing = int(_frame["n_subs"].isna().sum())
    if _missing:
        raise AssertionError(
            f"{_screener}: {_missing} biographies have no number of "
            f"substitutions, so the model cannot be fitted as in notebook 06.")
    _gated = _frame[_frame["sim_cp"] >= PRIMARY_GATE].merge(
        _bio_features, on=["occ", "bio_idx"], how="left")
    print()
    print(f"=== {_screener} ({_label}) ===")
    for arm in ARMS:
        _sub = _gated[_gated["arm"] == arm].copy()
        _sub["log_wc"] = np.log(_sub["word_count"].clip(lower=1))
        _other = "agency_rate" if arm == "communion" else "communion_rate"
        _fit_data = _sub.dropna(subset=["n_subs", _other, "log_wc"])
        _constant = [_c for _c in ("n_subs", _other, "log_wc")
                     if _fit_data[_c].nunique() < 2]
        print(f"  control {arm}: {len(_fit_data)} rows with all predictors, "
              f"occupations {_fit_data['occ'].nunique()}, "
              f"predictors without variation {_constant}")
        if len(_fit_data) < 30 or _constant or _fit_data["occ"].nunique() < 2:
            raise AssertionError(
                f"The regression for {_screener}/{arm} cannot be fitted: "
                f"{len(_fit_data)} rows, {_fit_data['occ'].nunique()} "
                f"occupations, predictors without variation {_constant}.")
        _res = smf.ols(f"delta ~ n_subs + {_other} + log_wc + C(occ)",
                       data=_fit_data).fit(cov_type="HC1")
        print(f"  {arm.upper()}: n={len(_sub)}  R^2={_res.rsquared:.3f}  "
              f"b_nsubs={_res.params['n_subs']:+.4f} "
              f"p={_res.pvalues['n_subs']:.4f}  "
              f"b_{_other}={_res.params[_other]:+.4f} "
              f"p={_res.pvalues[_other]:.4f}")
        for _pname in _res.params.index:
            _dimsep_rows.append({"screener": _screener, "encoder": _label,
                                 "sample": "pool", "arm": arm,
                                 "n": int(len(_fit_data)),
                                 "r_squared": float(_res.rsquared),
                                 "param": _pname,
                                 "coef": float(_res.params[_pname]),
                                 "se": float(_res.bse[_pname]),
                                 "p": float(_res.pvalues[_pname])})
df_arm2_dimension_separation = pd.DataFrame(_dimsep_rows)

# Control: the rows of Arm 1 are the ones notebook 06 section 14.3 wrote.
_saved_ols = pd.read_csv(OLS_ARM1_FILE)
_here_arm1 = df_arm2_dimension_separation[
    df_arm2_dimension_separation["screener"] == "arm1"]
_joined = _saved_ols.merge(_here_arm1, on=["sample", "arm", "param"],
                           suffixes=("_saved", "_here"))
print()
print(f"control rows of Arm 1 matched against {OLS_ARM1_FILE.name}: "
      f"{len(_joined)} of {len(_saved_ols)}")
if len(_joined) != len(_saved_ols):
    raise AssertionError(
        f"{len(_saved_ols) - len(_joined)} rows of {OLS_ARM1_FILE.name} have "
        f"no counterpart here, so the two regressions do not run on the same "
        f"terms.")
_worst_ols = float(np.abs(_joined["coef_saved"] - _joined["coef_here"]).max())
print(f"largest deviation of a coefficient: {_worst_ols:.3g} (expected below "
      f"1e-06)")
if _worst_ols > 1e-6:
    raise AssertionError(
        f"The regression of Arm 1 repeated here deviates from "
        f"{OLS_ARM1_FILE.name} by {_worst_ols:.3g}.")

df_arm2_dimension_separation.to_csv(DIMSEP_FILE, index=False, encoding="utf-8")
print(f"written: {DIMSEP_FILE} ({len(df_arm2_dimension_separation)} rows)")
```

## 16. The change by author gender for Arm 2 and Arm 3

- **Question:** Does the change differ between biographies of women and of
  men under the other two screeners?
- **Expectation:** the ones of notebook 06 section 14.2.
- **Status:** added on 2026-11-07. The section repeats a test of notebook 06
  on Arm 2 and on Arm 3 without changing its specification; Arm 1 is repeated
  as a control and has to reproduce `study2_gender_pool.csv`.
- **Unit and test:** biography; Wilcoxon per group and Mann-Whitney between
  the groups, over the biographies the manipulation changed, at the gate of
  the main analysis.
- **Output:** `study2_gender_by_screener.csv`; the rows by author gender of
  Table E5.

```python
# The change split by the gender of the author, as notebook 06 section 14.2
# splits it for Arm 1: the treatment table per arm and gender, and a
# Mann-Whitney U between women and men over the biographies the manipulation
# changed. The specification is unchanged; only the screener differs. Arm 1 is
# repeated as a control, Arm 2 and Arm 3 are new.
from scipy.stats import mannwhitneyu as _mwu_arm
GENDER_ARM1_FILE = Path(OUT_BASE) / "study2_gender_pool.csv"
GENDER_SCREENER_FILE = Path(OUT_BASE) / "study2_gender_by_screener.csv"
if not GENDER_ARM1_FILE.exists():
    raise FileNotFoundError(
        f"{GENDER_ARM1_FILE} not found. It carries the rows of Arm 1 that "
        f"this section is controlled against.")

# The gender of the author belongs to the biography, so all three screeners
# take it from the table of notebook 06.
_gender_of_bio = df_pool[["occ", "bio_idx", "arm", "female"]].drop_duplicates()
_gender_tables, _gender_rows = {}, []
for _screener, _label, _frame in DELTA_SOURCES:
    _gated = _frame[_frame["sim_cp"] >= PRIMARY_GATE].copy()
    if "female" not in _gated.columns:
        _gated = _gated.merge(_gender_of_bio, on=["occ", "bio_idx", "arm"],
                              how="left")
    _no_gender = int(_gated["female"].isna().sum())
    print(f"=== {_screener} ({_label}, gate={PRIMARY_GATE}) - by gender ===")
    print(f"  control biographies without a gender: {_no_gender} (expected 0)")
    if _no_gender:
        raise AssertionError(
            f"{_no_gender} biographies of {_screener} carry no gender, so the "
            f"split cannot be formed as in notebook 06.")
    _table = study2_core.treatment_table(_gated, ["arm", "female"])
    print(_table[["arm", "female", "n", "share_unchanged", "median_all",
                  "p_wilcoxon_all", "median_treated",
                  "p_wilcoxon_treated"]].to_string(index=False))
    print(f"\n  Mann-Whitney U (women vs men, changed bios, "
          f"gate={PRIMARY_GATE}):")
    for arm in ARMS:
        _sub = _gated[(_gated["arm"] == arm) & (_gated["n_subs"] >= 1)]
        _f = _sub[_sub["female"] == 1]["delta"].to_numpy(float)
        _m = _sub[_sub["female"] == 0]["delta"].to_numpy(float)
        if len(_f) >= 5 and len(_m) >= 5:
            _U, _p = _mwu_arm(_f, _m, alternative="two-sided")
            print(f"  {arm:<10}  nF={len(_f):>4}  nM={len(_m):>4}  "
                  f"medianF={np.median(_f):+.3f}  medianM={np.median(_m):+.3f}"
                  f"  U={_U:.0f}  p={_p:.4f}")
            _gender_rows.append({"screener": _screener, "encoder": _label,
                                 "sample": "pool", "arm": arm,
                                 "n_female": int(len(_f)),
                                 "n_male": int(len(_m)),
                                 "median_female": float(np.median(_f)),
                                 "median_male": float(np.median(_m)),
                                 "U": float(_U), "p_mwu": float(_p)})
        else:
            print(f"  {arm:<10}  insufficient data (nF={len(_f)}, "
                  f"nM={len(_m)})")
    _table.insert(0, "screener", _screener)
    _gender_tables[_screener] = _table
    print()
df_gender_by_screener = pd.DataFrame(_gender_rows)
df_gender_tables = pd.concat(_gender_tables.values(), ignore_index=True)

# Control: the table of Arm 1 is the one of notebook 06 section 14.2.
_saved_gender = pd.read_csv(GENDER_ARM1_FILE)
_here_arm1 = _gender_tables["arm1"]
_joined_g = _saved_gender.merge(_here_arm1, on=["arm", "female"],
                                suffixes=("_saved", "_here"))
_worst_gender = float(np.abs(_joined_g["median_treated_saved"]
                             - _joined_g["median_treated_here"]).max())
print(f"control rows of Arm 1 matched against {GENDER_ARM1_FILE.name}: "
      f"{len(_joined_g)} of {len(_saved_gender)}, largest deviation of the "
      f"median {_worst_gender:.3g} (expected below 1e-06)")
if len(_joined_g) != len(_saved_gender) or _worst_gender > 1e-6:
    raise AssertionError(
        f"The split by gender of Arm 1 repeated here deviates from "
        f"{GENDER_ARM1_FILE.name} by {_worst_gender:.3g}.")

df_gender_by_screener.to_csv(GENDER_SCREENER_FILE, index=False,
                             encoding="utf-8")
print(f"written: {GENDER_SCREENER_FILE} ({len(df_gender_by_screener)} rows)")
```

## 17. Why does a screener lift the edited biographies?

- **Question:** Arm 1 lifts a biography whose communal markers were replaced.
  Where does that gain come from: from the words the protocol puts in, from
  the vocabulary of the advertisements, from a general move towards
  advertisement prose, or from the position the rule applies in?
- **Expectation:** none. The section was written after the results were known
  and no direction was fixed for any of its four parts.
- **Status:** exploratory, added on 2026-11-07 after the results; no
  hypothesis, no direction expected; reported as such. Nothing is embedded:
  the cached embeddings of Arm 1, the embeddings of Arm 2, the protocol and
  the tables of notebook 06 carry everything the four parts need, and
  section 17.2 reads the advertisement texts themselves.
- **Unit and test:** given per part; every part reports Arm 1 and Arm 2 next
  to each other.
- **Output:** `study2_sbert_diag_rules.csv`,
  `study2_sbert_diag_vocab.csv`, `study2_sbert_diag_generic.csv`,
  `study2_sbert_diag_pos.csv`.

Notebook 06 section 14.11 reports the word-level overlap and section 14.12 the
change by the class of the word; neither is a reading per rule, per word of
the advertisements or per position of the rule, so the four parts below do
not repeat them.

### 17.1 The substitution rules

- **Question:** Which rules of the protocol move the ranking, and do the two
  encoders order the rules in the same way?
- **Expectation:** none fixed in advance.
- **Status:** exploratory, added on 2026-11-07 after the results.
- **Unit and test:** rule of the protocol; a rule enters when it was applied
  at least twenty times among the changed biographies of the pool at the
  gate. The means are taken over the biographies with exactly one
  substitution, so that two rules cannot mix, and a rule needs five of them.
  Spearman between the mean per rule of Arm 1 and of Arm 2.
- **Output:** `study2_sbert_diag_rules.csv`.

```python
# D1: which substitution rules move the ranking? The unit is the rule of the
# protocol, the quantity is the change of a biography the rule was applied to.
# A biography with several substitutions would mix the rules, so the means are
# taken over the biographies with exactly one substitution; the count of
# occurrences that decides which rules are large enough is taken over all
# changed biographies of the pool.
RULES_FILE = Path("config/substitution_rules_v2.csv")
DIAG_RULES_FILE = Path(OUT_BASE) / "study2_sbert_diag_rules.csv"
MIN_OCCURRENCES = 20      # per rule, over the changed biographies of the pool
MIN_SINGLE = 5            # per rule, over the biographies with one substitution
N_SHOWN = 15              # rules shown at each end of the ranking
if not RULES_FILE.exists():
    raise FileNotFoundError(
        f"{RULES_FILE} not found. The source word and the class of a rule come "
        f"from the substitution list of notebook 06.")

df_rules_v2 = pd.read_csv(RULES_FILE)
df_log = pd.read_csv(PROTOCOL_FILE)
_treated = df_log[(df_log["sample"] == "pool")
                  & (df_log["action_taken"].isin(["replaced", "deleted"]))]
_treated = _treated.merge(
    df_pool[["occ", "bio_idx", "arm", "n_subs", "sim_cp", "delta"]],
    on=["occ", "bio_idx", "arm"], how="inner")
_treated = _treated[_treated["sim_cp"] >= PRIMARY_GATE]
print(f"treated occurrences of the pool at the gate of {PRIMARY_GATE}: "
      f"{len(_treated)} in "
      f"{_treated[['occ', 'bio_idx', 'arm']].drop_duplicates().shape[0]} "
      f"biographies, rules used {_treated['rule_id'].nunique()}")

# Control: a biography with one substitution carries exactly one occurrence.
_single = _treated[_treated["n_subs"] == 1]
_per_bio = _single.groupby(["occ", "bio_idx", "arm"]).size()
print(f"control biographies with exactly one substitution: {len(_per_bio)}, "
      f"occurrences per biography {sorted(_per_bio.unique())} (expected [1])")
if list(_per_bio.unique()) != [1]:
    raise AssertionError(
        f"A biography with n_subs == 1 carries "
        f"{sorted(_per_bio.unique())} occurrences in the protocol; the means "
        f"per rule would mix two rules.")

# The change of the same biography under both encoders.
_delta_arm2 = df_arm2_pool[["occ", "bio_idx", "arm", "delta"]].rename(
    columns={"delta": "delta_arm2"})
_single = _single.rename(columns={"delta": "delta_arm1"}).merge(
    _delta_arm2, on=["occ", "bio_idx", "arm"], how="left")
_absent = int(_single["delta_arm2"].isna().sum())
print(f"control occurrences without a change of Arm 2: {_absent} (expected 0)")
if _absent:
    raise AssertionError(
        f"{_absent} biographies of the protocol have no change under Arm 2.")

_counts = _treated.groupby(["rule_id", "arm"]).size().rename("n_occurrences")
_large = _counts[_counts >= MIN_OCCURRENCES].reset_index()
print(f"rules with at least {MIN_OCCURRENCES} occurrences: {len(_large)} of "
      f"{len(_counts)}")

_grouped = _single.groupby(["rule_id", "arm"])
df_diag_rules = pd.DataFrame({
    "n_single": _grouped.size(),
    "mean_arm1": _grouped["delta_arm1"].mean(),
    "median_arm1": _grouped["delta_arm1"].median(),
    "mean_arm2": _grouped["delta_arm2"].mean(),
    "median_arm2": _grouped["delta_arm2"].median(),
    "target": _grouped["target"].agg(
        lambda s: s.dropna().mode().iat[0] if s.notna().any() else "deleted"),
}).reset_index().merge(_large, on=["rule_id", "arm"], how="inner")
df_diag_rules = df_diag_rules.merge(
    df_rules_v2[["rule_id", "source_word", "word_class", "position",
                 "action"]],
    on="rule_id", how="left")
df_diag_rules = df_diag_rules[df_diag_rules["n_single"] >= MIN_SINGLE]
df_diag_rules = df_diag_rules.sort_values("mean_arm1", ascending=False)
print(f"rules with at least {MIN_SINGLE} biographies of one substitution: "
      f"{len(df_diag_rules)}")

_rho_rules, _p_rules = spearmanr(df_diag_rules["mean_arm1"],
                                 df_diag_rules["mean_arm2"])
print(f"Spearman between the mean per rule of Arm 1 and of Arm 2: "
      f"rho={_rho_rules:+.3f} p={_p_rules:.4g} over {len(df_diag_rules)} "
      f"rules")

_columns = ["rule_id", "arm", "source_word", "target", "word_class",
            "n_occurrences", "n_single", "mean_arm1", "mean_arm2"]
for _label, _part in [(f"the {N_SHOWN} rules with the largest mean change in "
                       f"Arm 1", df_diag_rules.head(N_SHOWN)),
                      (f"the {N_SHOWN} rules with the smallest mean change in "
                       f"Arm 1", df_diag_rules.tail(N_SHOWN))]:
    print()
    print(_label)
    print(_part[_columns].to_string(index=False,
                                    float_format=lambda v: f"{v:+.3f}"))

df_diag_rules.to_csv(DIAG_RULES_FILE, index=False, encoding="utf-8")
print()
print(f"written: {DIAG_RULES_FILE} ({len(df_diag_rules)} rows)")
```

### 17.2 Closeness to the vocabulary of the advertisements

- **Question:** Does a substitution move a biography towards the words the
  advertisements of its occupation use?
- **Expectation:** none fixed in advance.
- **Status:** exploratory, added on 2026-11-07 after the results. The
  score orders the biographies against each other; it carries nothing about
  the level of the change, and the correlation is not read as the reason why
  one screener lifts the edited biographies.
- **Unit and test:** biography; the new word and the replaced word are
  counted in the advertisements of the own occupation per 10,000 tokens and
  their difference is summed over the occurrences of a biography. Spearman
  with the change, within the occupations, with the permutation test of
  notebook 06: 5,000 permutations stratified by occupation, seed 42.
- **Output:** `study2_sbert_diag_vocab.csv`.

```python
# D2: does a substitution move a biography towards the vocabulary of the
# advertisements? For every occurrence the new word and the replaced word are
# counted in the advertisements of the same occupation, per 10,000 tokens, and
# the difference is summed over the occurrences of a biography. The score is
# then correlated with the change, within the occupations, with the
# permutation test of notebook 06.
import re as _re
from collections import Counter as _Counter

JD_RAW_PATH = Path("data/job_descriptions/job_postings.csv")
DIAG_VOCAB_FILE = Path(OUT_BASE) / "study2_sbert_diag_vocab.csv"
RANDOM_STATE = 42         # the seed of the draw in notebook 06
N_JDS_FULL = 250          # the cap per occupation in notebook 06
PER_TOKENS = 10000        # the frequencies are counted per 10,000 tokens
OCC_KEYWORDS = {
    "nurse": [r"\bnurse\b", r"\bnursing\b", r"\brn\b",
              r"\bnurse practitioner\b"],
    "software_engineer": [r"\bsoftware engineer\b", r"\bsoftware developer\b",
                          r"\bprogrammer\b", r"\bfull.?stack\b",
                          r"\bbackend engineer\b", r"\bfrontend engineer\b"],
    "accountant": [r"\baccountant\b", r"\bcpa\b", r"\bbookkeeper\b",
                   r"\baccounting\b"],
    "architect": [r"\barchitect\b", r"\barchitecture\b"],
    "psychologist": [r"\bpsychologist\b", r"\btherapist\b", r"\bcounselor\b"],
    "attorney": [r"\battorney\b", r"\blawyer\b", r"\bcounsel\b"],
    "teacher": [r"\bteacher\b", r"\beducator\b", r"\bk-12\b"],
    "surgeon": [r"\bsurgeon\b", r"\bsurgery\b", r"\bsurgical\b"],
    "physician": [r"\bphysician\b", r"\bmedical doctor\b", r"\binternist\b"],
    "professor": [r"\bprofessor\b", r"\bfaculty\b", r"\blecturer\b"],
    "dentist": [r"\bdentist\b", r"\bdental\b", r"\bdds\b"],
    "paralegal": [r"\bparalegal\b", r"\blegal assistant\b"],
    "dietitian": [r"\bdietitian\b", r"\bnutritionist\b",
                  r"\bregistered dietitian\b"],
}
if not JD_RAW_PATH.exists():
    raise FileNotFoundError(
        f"{JD_RAW_PATH} not found. The vocabulary of the advertisements is "
        f"counted on the texts themselves, which notebook 06 reads from this "
        f"file.")

# The draw of notebook 06, repeated: the title decides the occupation, rows
# without a text are dropped, and the rest is drawn with the same seed.
_jd_raw = pd.read_csv(JD_RAW_PATH, usecols=["title", "description"],
                      low_memory=False)
_title_lower = _jd_raw["title"].fillna("").str.lower()
_jd_raw["occupation"] = None
for _occ, _patterns in OCC_KEYWORDS.items():
    _jd_raw.loc[_title_lower.str.contains("|".join(_patterns), regex=True,
                                          na=False), "occupation"] = _occ
_jd_matched = _jd_raw[_jd_raw["occupation"].notna()]
jd_texts_by_occ = {}
for occ in KEPT_13:
    _sub = _jd_matched[_jd_matched["occupation"] == occ][["description"]].dropna()
    jd_texts_by_occ[occ] = (_sub.sample(min(N_JDS_FULL, len(_sub)),
                                        random_state=RANDOM_STATE)
                            .reset_index(drop=True)["description"].tolist())
del _jd_raw, _jd_matched

# Control: the texts drawn here are the ones notebook 06 counted, so their
# token count has to be the one of study2_jd_eda.csv for every advertisement.
_word_counts = pd.DataFrame(
    [{"occ": occ, "jd_idx": _j, "n_words_here": max(len(_t.split()), 1)}
     for occ in KEPT_13 for _j, _t in enumerate(jd_texts_by_occ[occ])])
_check_jd = _word_counts.merge(df_jd_eda[["occ", "jd_idx", "n_words"]],
                               on=["occ", "jd_idx"], how="outer")
_bad_jd = int((_check_jd["n_words_here"] != _check_jd["n_words"]).sum())
print(f"advertisements drawn: {len(_word_counts)}, token counts that differ "
      f"from study2_jd_eda.csv: {_bad_jd} (expected 0)")
if _bad_jd:
    raise AssertionError(
        f"{_bad_jd} advertisements differ from the ones notebook 06 counted, "
        f"so the draw was not repeated correctly.")

# One count of the word forms per occupation; a target of several words is
# counted as a phrase, where three dots stand for one word.
_counts_by_occ, _tokens_by_occ, _texts_joined = {}, {}, {}
for occ in KEPT_13:
    _lower = [_t.lower() for _t in jd_texts_by_occ[occ]]
    _texts_joined[occ] = "\n".join(_lower)
    _counts_by_occ[occ] = _Counter(
        _re.findall(r"[a-z0-9']+", _texts_joined[occ]))
    _tokens_by_occ[occ] = int(_word_counts[_word_counts["occ"] == occ]
                              ["n_words_here"].sum())
print(f"tokens of the advertisements: {sum(_tokens_by_occ.values()):,} in "
      f"{len(KEPT_13)} occupations")


def _frequency(occ, phrase):
    """Occurrences of a word or a phrase per 10,000 tokens of the ads."""
    if not isinstance(phrase, str) or not phrase.strip():
        return 0.0
    _words = phrase.lower().split()
    if len(_words) == 1 and _re.fullmatch(r"[a-z0-9']+", _words[0]):
        _n = _counts_by_occ[occ][_words[0]]
    else:
        _parts = [r"\S+" if _w == "..." else _re.escape(_w) for _w in _words]
        _n = len(_re.findall(r"\b" + r"\s+".join(_parts) + r"\b",
                             _texts_joined[occ]))
    return _n / _tokens_by_occ[occ] * PER_TOKENS


# The score per occurrence: the new word minus the replaced word. A deletion
# has no new word and counts as the loss of the replaced one alone.
_cache = {}
_occurrences = _treated.copy()
_gain = []
for _row in _occurrences.itertuples():
    _value = 0.0
    for _word, _sign in ((_row.target, 1.0), (_row.token, -1.0)):
        _key = (_row.occ, _word)
        if _key not in _cache:
            _cache[_key] = _frequency(_row.occ, _word)
        _value += _sign * _cache[_key]
    _gain.append(_value)
_occurrences["vocab_gain"] = _gain
_per_bio_score = (_occurrences.groupby(["occ", "bio_idx", "arm"])
                  ["vocab_gain"].sum().rename("vocab_gain").reset_index())
print(f"biographies with a score: {len(_per_bio_score)}, mean gain "
      f"{_per_bio_score['vocab_gain'].mean():+.3f} per 10,000 tokens, "
      f"share above zero {(_per_bio_score['vocab_gain'] > 0).mean():.3f}")

_vocab_rows = []
print()
print(f"  {'screener':<10} {'arm':<10} {'n':>5} {'rho':>8} {'p_perm':>10}")
print("  " + "-" * 47)
for _screener, _label, _frame in DELTA_SOURCES[:2]:
    _delta = _frame[["occ", "bio_idx", "arm", "delta"]]
    _joined_v = _per_bio_score.merge(_delta, on=["occ", "bio_idx", "arm"],
                                     how="inner")
    for arm in ARMS:
        _sub = _joined_v[_joined_v["arm"] == arm]
        _rho, _p, _, _ = study2_core.strat_perm_p(
            _sub["occ"].to_numpy(), _sub["vocab_gain"].to_numpy(float),
            _sub["delta"].to_numpy(float), np.zeros(len(_sub)),
            n_perm=N_PERM, seed=SEED)
        print(f"  {_screener:<10} {arm:<10} {len(_sub):>5} {_rho:>+8.3f} "
              f"{_p:>10.4f}")
        _vocab_rows.append({"screener": _screener, "encoder": _label,
                            "arm": arm, "level": "pooled", "occ": "POOLED",
                            "n": int(len(_sub)), "rho": float(_rho),
                            "p": float(_p)})
        for occ in KEPT_13:
            _occ_sub = _sub[_sub["occ"] == occ]
            _rho_occ, _p_occ = spearmanr(_occ_sub["vocab_gain"],
                                         _occ_sub["delta"])
            _vocab_rows.append({"screener": _screener, "encoder": _label,
                                "arm": arm, "level": "occupation",
                                "occ": occ, "n": int(len(_occ_sub)),
                                "rho": float(_rho_occ), "p": float(_p_occ)})
df_diag_vocab = pd.DataFrame(_vocab_rows)
_occ_level = df_diag_vocab[df_diag_vocab["level"] == "occupation"]
print()
print(f"occupations with a negative rho: "
      f"{int((_occ_level['rho'] < 0).sum())} of {len(_occ_level)}")

df_diag_vocab.to_csv(DIAG_VOCAB_FILE, index=False, encoding="utf-8")
print(f"written: {DIAG_VOCAB_FILE} ({len(df_diag_vocab)} rows)")
```

### 17.3 A general approach or a closer fit?

- **Question:** Does the edited biography move towards the advertisements of
  its own occupation, or towards advertisement prose in general?
- **Expectation:** none fixed in advance. If the similarity to the
  advertisements of the other occupations rises as much as the similarity to
  the own ones, the gain is not a fit to the own advertisement.
- **Status:** exploratory, added on 2026-11-07 after the results.
- **Unit and test:** biography; the mean cosine similarity before and after
  the substitution, once against the advertisements of the own occupation and
  once against those of the other twelve, with a Wilcoxon of the change
  against zero and the share of biographies whose similarity rises.
- **Output:** `study2_sbert_diag_generic.csv`, Table E6.

```python
# D3: a general approach to advertisement prose, or a closer fit to the own
# advertisements? For every changed biography the mean cosine similarity is
# taken before and after the substitution, once against the advertisements of
# its own occupation and once against the 2,725 advertisements of the other
# twelve occupations. If the second rises as much as the first, the gain is
# not a fit to the own advertisement.
from scipy.stats import wilcoxon as _wilcoxon_d3
ARM1_CACHE_D3 = Path("outputs/study2_cache/sentence_transformers_all_mpnet_base_v2")
DIAG_GENERIC_FILE = Path(OUT_BASE) / "study2_sbert_diag_generic.csv"
_needed_d3 = [ARM1_CACHE_D3 / f"{occ}_{arm}_manip_pool_run5.npy"
              for occ in KEPT_13 for arm in ARMS]
_needed_d3 += [ARM1_CACHE_D3 / f"{occ}_pool_orig.npy" for occ in KEPT_13]
_needed_d3 += [ARM1_CACHE_D3 / f"{occ}_jds.npy" for occ in KEPT_13]
for _path in _needed_d3:
    if not _path.exists():
        raise FileNotFoundError(
            f"{_path} not found. Arm 1 is read from the embeddings notebook "
            f"06 cached; nothing is embedded here.")

# The embeddings of both encoders, in the order of the pool.
_emb_arm1_orig, _emb_arm1_manip, _emb_arm1_jd = {}, {}, {}
for occ in KEPT_13:
    _emb_arm1_orig[occ] = np.load(ARM1_CACHE_D3 / f"{occ}_pool_orig.npy",
                                  allow_pickle=True).item()["mean"]
    _emb_arm1_jd[occ] = np.load(ARM1_CACHE_D3 / f"{occ}_jds.npy",
                                allow_pickle=True).item()["mean"]
    _emb_arm1_manip[occ] = {
        arm: np.load(ARM1_CACHE_D3 / f"{occ}_{arm}_manip_pool_run5.npy",
                     allow_pickle=True).item()["mean"] for arm in ARMS}

# Control: the ranks these embeddings give are the ones of notebook 06.
_worst_d3 = 0.0
for occ in KEPT_13:
    _so = _emb_arm1_orig[occ] @ _emb_arm1_jd[occ].T
    _base = study2_core.percentile_ranks(_so)
    for arm in ARMS:
        _sm = _emb_arm1_manip[occ][arm] @ _emb_arm1_jd[occ].T
        _delta_here = study2_core.individual_cf_ranks(_so, _sm) - _base
        _saved = (df_pool[(df_pool["occ"] == occ) & (df_pool["arm"] == arm)]
                  .sort_values("bio_idx")["delta"].to_numpy(float))
        _worst_d3 = max(_worst_d3,
                        float(np.abs(_delta_here - _saved).max()))
print(f"control change of Arm 1 from the cache against study2_delta_r5_pool."
      f"csv: largest deviation {_worst_d3:.3g} (expected below 1e-06)")
if _worst_d3 > 1e-6:
    raise AssertionError(
        f"The cached embeddings of Arm 1 do not give the change of notebook "
        f"06; they deviate by {_worst_d3:.3g}.")

ENCODERS_D3 = [("arm1", "SBERT", _emb_arm1_orig, _emb_arm1_manip,
                _emb_arm1_jd),
               ("arm2", "Qwen3-Embedding", emb_orig,
                {occ: emb_manip[occ] for occ in KEPT_13}, emb_jd)]
_changed_d3 = df_pool[(df_pool["sim_cp"] >= PRIMARY_GATE)
                      & (df_pool["n_subs"] >= 1)]
_generic_rows, _per_bio_d3 = [], []
for _screener, _label, _orig, _manip, _jd in ENCODERS_D3:
    _jd_all = {occ: _jd[occ] for occ in KEPT_13}
    for occ in KEPT_13:
        _other = np.vstack([_jd_all[_o] for _o in KEPT_13 if _o != occ])
        for arm in ARMS:
            _idx = (_changed_d3[(_changed_d3["occ"] == occ)
                                & (_changed_d3["arm"] == arm)]["bio_idx"]
                    .to_numpy(int))
            for _scope, _ads in (("own", _jd_all[occ]), ("other", _other)):
                _before = (_orig[occ][_idx] @ _ads.T).mean(axis=1)
                _after = (_manip[occ][arm][_idx] @ _ads.T).mean(axis=1)
                _per_bio_d3.append(pd.DataFrame(
                    {"screener": _screener, "occ": occ, "arm": arm,
                     "bio_idx": _idx, "scope": _scope,
                     "before": _before, "after": _after}))
df_generic_per_bio = pd.concat(_per_bio_d3, ignore_index=True)
df_generic_per_bio["change"] = (df_generic_per_bio["after"]
                                - df_generic_per_bio["before"])
_measured = df_generic_per_bio[df_generic_per_bio["screener"] == "arm1"]
print(f"changed biographies measured: "
      f"{_measured[['occ', 'arm', 'bio_idx']].drop_duplicates().shape[0]}, "
      f"rows {len(df_generic_per_bio)} (two encoders, two sets of ads)")

print()
print(f"  {'screener':<10} {'arm':<10} {'ads':<6} {'n':>5} "
      f"{'median before':>14} {'median after':>13} {'median change':>14} "
      f"{'p_wilcoxon':>11} {'share up':>9}")
print("  " + "-" * 97)
for _screener, _label, _, _, _ in ENCODERS_D3:
    for arm in ARMS:
        for _scope in ("own", "other"):
            _sub = df_generic_per_bio[
                (df_generic_per_bio["screener"] == _screener)
                & (df_generic_per_bio["arm"] == arm)
                & (df_generic_per_bio["scope"] == _scope)]
            _change = _sub["change"].to_numpy(float)
            _p = float(study2_core._safe_wilcoxon(_change))
            print(f"  {_screener:<10} {arm:<10} {_scope:<6} {len(_sub):>5} "
                  f"{_sub['before'].median():>14.4f} "
                  f"{_sub['after'].median():>13.4f} "
                  f"{np.median(_change):>+14.5f} {_p:>11.3g} "
                  f"{(_change > 0).mean():>9.3f}")
            _generic_rows.append({
                "screener": _screener, "encoder": _label, "arm": arm,
                "ads": _scope, "n": int(len(_sub)),
                "median_before": float(_sub["before"].median()),
                "median_after": float(_sub["after"].median()),
                "median_change": float(np.median(_change)),
                "mean_change": float(_change.mean()),
                "p_wilcoxon": _p,
                "share_up": float((_change > 0).mean())})
df_diag_generic = pd.DataFrame(_generic_rows)

df_diag_generic.to_csv(DIAG_GENERIC_FILE, index=False, encoding="utf-8")
print()
print(f"written: {DIAG_GENERIC_FILE} ({len(df_diag_generic)} rows)")
```

### 17.4 The position the rule applies in

- **Question:** Does the change depend on the position of the rule the
  protocol applied?
- **Expectation:** none fixed in advance.
- **Status:** exploratory, added on 2026-11-07 after the results. The
  protocol carries no part of speech from a parser, so what the section reads
  is the position of the rule in the substitution list; the printed column is
  labelled kind and holds that position. Not reported in the thesis
  (decision 2026-11-08).
- **Unit and test:** biography with exactly one substitution; mean and median
  change per position of the rule, with a Kruskal-Wallis over the positions
  that reach ten occurrences.
- **Output:** `study2_sbert_diag_pos.csv`.

```python
# D4: does the change depend on the position of the rule? For
# all-mpnet-base-v2 the similarity of two texts rests above all on the nouns
# they share (Nikolaev and Pado 2023), so the kind of the replaced word is
# worth looking at. The protocol carries no part of speech from a parser; the
# substitution list carries the position a rule applies in, and that position
# is read as the kind of the word. Only biographies with exactly one
# substitution enter, so the kind is the one of that single occurrence.
from scipy.stats import kruskal as _kruskal_d4
DIAG_POS_FILE = Path(OUT_BASE) / "study2_sbert_diag_pos.csv"
MIN_GROUP = 10            # occurrences a kind needs to enter the test
WORD_KIND = {"N": "noun", "N+DET": "noun", "V": "verb", "V+O": "verb",
             "V+V": "verb", "V+O+V": "verb", "ADJ": "adjective",
             "ADV": "adverb", "PH": "phrase", "M": "modifier", "*": "any"}

_positions = (df_rules_v2.rename(columns={"position": "pos_code"})
              [["rule_id", "pos_code"]])
_unknown = sorted(set(_positions["pos_code"].dropna()) - set(WORD_KIND))
print(f"positions of the substitution list: "
      f"{sorted(set(_positions['pos_code'].dropna()))}, without a kind "
      f"{_unknown} (expected [])")
if _unknown:
    raise AssertionError(
        f"The substitution list uses the positions {_unknown}, which this "
        f"section cannot read as a kind of word.")

_single_pos = _single.merge(_positions, on="rule_id", how="left")
_single_pos["word_kind"] = _single_pos["pos_code"].map(WORD_KIND)
_no_kind = int(_single_pos["word_kind"].isna().sum())
print(f"occurrences with one substitution: {len(_single_pos)}, without a kind "
      f"{_no_kind} (expected 0)")
if _no_kind:
    raise AssertionError(
        f"{_no_kind} occurrences carry a rule that is not in the substitution "
        f"list, so their kind of word is unknown.")

_pos_rows = []
print()
print(f"  {'screener':<10} {'arm':<10} {'kind':<10} {'n':>5} {'mean':>9} "
      f"{'median':>9}")
print("  " + "-" * 57)
for _screener, _column in (("arm1", "delta_arm1"), ("arm2", "delta_arm2")):
    for arm in ARMS:
        _sub = _single_pos[_single_pos["arm"] == arm]
        _groups, _names = [], []
        for _kind in sorted(_sub["word_kind"].unique()):
            _values = _sub[_sub["word_kind"] == _kind][_column].to_numpy(float)
            print(f"  {_screener:<10} {arm:<10} {_kind:<10} {len(_values):>5} "
                  f"{_values.mean():>+9.3f} {np.median(_values):>+9.3f}")
            _pos_rows.append({"screener": _screener, "arm": arm,
                              "word_kind": _kind, "n": int(len(_values)),
                              "mean_delta": float(_values.mean()),
                              "median_delta": float(np.median(_values)),
                              "in_test": bool(len(_values) >= MIN_GROUP),
                              "kruskal_h": np.nan, "kruskal_p": np.nan})
            if len(_values) >= MIN_GROUP:
                _groups.append(_values)
                _names.append(_kind)
        if len(_groups) < 2:
            raise AssertionError(
                f"{_screener}/{arm}: fewer than two kinds of word reach "
                f"{MIN_GROUP} occurrences, so the test has nothing to compare.")
        _h, _p_kw = _kruskal_d4(*_groups)
        print(f"  {_screener:<10} {arm:<10} Kruskal-Wallis over {len(_groups)} "
              f"kinds {_names}: H={_h:.3f} p={_p_kw:.4g}")
        _pos_rows.append({"screener": _screener, "arm": arm,
                          "word_kind": "all kinds of the test",
                          "n": int(sum(len(_g) for _g in _groups)),
                          "mean_delta": np.nan, "median_delta": np.nan,
                          "in_test": True, "kruskal_h": float(_h),
                          "kruskal_p": float(_p_kw)})
df_diag_pos = pd.DataFrame(_pos_rows)

df_diag_pos.to_csv(DIAG_POS_FILE, index=False, encoding="utf-8")
print(f"written: {DIAG_POS_FILE} ({len(df_diag_pos)} rows)")
```

## 18. Figure 4: the design of Study 2 in three steps

- **Question:** none; the figure draws the design of the study.
- **Expectation:** none.
- **Status:** added on 2026-11-06, the layout renewed on 2026-11-11, revised
  on 2026-11-13 and on 2026-11-15: equal box heights, symmetric fork from
  step 1. Every box is of the same height, so both branches are equally high
  and step 1 sits in the middle between them; one line leaves it, splits half

  and the cell stops unless the fork is symmetric.
- **Unit and test:** none, a figure.
- **Output:** `outputs/figures/fig5_study2_design.png`, Figure 4 of the
  thesis.

```python
# Figure 4: the design of Study 2 in three steps (schematic, no data).
# Version of 2026-11-15 (all boxes of equal height, both branches of equal height and centred on step 1, which forks into
# them with one line; before: 2026-11-14 arrows to the frames, 2026-11-13 LLM and H4a): a horizontal process (inputs, baseline ranking,
# then the two branches); all five process boxes as wide as the box "2 Counterfactual edit"; the observational tests fill the lower branch.
# Box titles follow the text (baseline ranking, counterfactual edit, re-ranking, experimental tests,
# observational analysis); the inputs name the smaller sample of the LLM. The source is ASCII:
# Delta and the not-equal sign are drawn with mathtext.
# Writes outputs/figures/fig5_study2_design.png at 300 dpi. Unlike the other figures (13 cm), this figure
# and Figure 1 are inserted at the full text width of 16 cm (6.30 in). The axes are drawn in inches at the printed width,
# so the font sizes are the printed ones.
from pathlib import Path
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

FIG_DESIGN_PNG = Path("outputs/figures/fig5_study2_design.png")
INK, GREY, LIGHT = "#1a1a1a", "#52514e", "#b5b4ae"
FILL, BLUE, BLUE_FILL = "#f3f2ee", "#2a78d6", "#e7f1fb"
W_IN = 6.30         # full text width, 16 cm
M = 0.06            # inner margin of a box, left and right, in inches
PAD = 0.065         # inner margin of a panel
EDGE = 0.01         # outer margin of the figure
TITLE_FS, BODY_FS, LABEL_FS = 6.0, 5.5, 6.3

plt.rcParams["font.family"] = "serif"
plt.rcParams["mathtext.fontset"] = "dejavuserif"   # Delta and not-equal in the same serif as the text
LINE = BODY_FS * 1.3 / 72                   # height of one body line in inches
H_BOX = 0.20 + 5 * LINE + 0.06              # the five process boxes
H_OBS = H_BOX                               # all boxes have the same height, so both branches are equally high
PANEL_HEAD = 0.17
y_bot = 0.02
h_panel_bot = PAD + H_OBS + PANEL_HEAD
y_top = y_bot + h_panel_bot + 0.08
h_panel_top = PAD + H_BOX + PANEL_HEAD
H_IN = y_top + h_panel_top + 0.02
fig = plt.figure(figsize=(W_IN, H_IN), dpi=300)    # measured at the export resolution
ax = fig.add_axes([0, 0, 1, 1])
ax.set_xlim(0, W_IN)
ax.set_ylim(0, H_IN)
ax.axis("off")
renderer = fig.canvas.get_renderer()

# body lines are plain strings; a tuple (text, "italic") sets one line in italics
BOXES = {
    "inputs": ("Inputs", None, ["13 occupations;", "250 bios and up to", "250 ads each;", "LLM: about 100 bios", "and 20 ads"]),
    "rank": ("Baseline ranking", "1", ["SBERT,", "Qwen3-Embedding,", "LLM: percentile rank", "of each bio per ad"]),
    "edit": ("Counterfactual edit", "2", ["replace or remove the", "communal (agentic)", "markers of one bio;", "check signal and content"]),
    "again": ("Re-ranking", "3", ["against the same ads and", "the unchanged bios of", r"the pool; $\Delta$ = rank after", "minus before"]),
    "tests_exp": ("Experimental tests", None, [r"H3: $\Delta \neq 0$", r"H4a: $\Delta$ by the ad's", "communal balance", "H4b: change in fit slope", "by gender"]),
    "tests_obs": ("Observational tests", None, [("original bios as ranked in step 1, without any edit", "italic"),
                                                "H4a (T2): fit slope by the bio's communal balance",
                                                "H4b (T1): gender gap by the ad's communal balance"]),
}


def text_width(s, size, bold=False, italic=False):
    t = ax.text(0, 0, s, fontsize=size, weight="bold" if bold else "normal", style="italic" if italic else "normal")
    w = t.get_window_extent(renderer).width / fig.dpi
    t.remove()
    return w


def head_of(key):
    title, step, _ = BOXES[key]
    return f"{step}  {title}" if step else title


def line_width(l):
    return text_width(l[0], BODY_FS, italic=True) if isinstance(l, tuple) else text_width(l, BODY_FS)


# every process box is as wide as the box of step 2 needs to be
w = max([text_width(head_of("edit"), TITLE_FS, True)] + [line_width(l) for l in BOXES["edit"][2]]) + 2 * M
gap = (W_IN - 2 * EDGE - 5 * w - 2 * PAD) / 4
assert gap >= 0.09, f"layout too narrow: gap {gap:.3f} in"

x_inputs = EDGE
x_rank = x_inputs + w + gap
x_panel = x_rank + w + gap
x_edit = x_panel + PAD
x_again = x_edit + w + gap
x_tests = x_again + w + gap
w_panel = 3 * w + 2 * gap + 2 * PAD
w_obs = 3 * w + 2 * gap
WIDTHS = {"inputs": w, "rank": w, "edit": w, "again": w, "tests_exp": w, "tests_obs": w_obs}


def box(key, x, y, h, edge=LIGHT, fill=FILL, ink=INK, body=GREY):
    ax.add_patch(FancyBboxPatch((x, y), WIDTHS[key], h, boxstyle="round,pad=0,rounding_size=0.05",
                                linewidth=0.8, edgecolor=edge, facecolor=fill, zorder=2))
    ax.text(x + M, y + h - 0.05, head_of(key), fontsize=TITLE_FS, color=ink, weight="bold", va="top", zorder=3)
    for i, l in enumerate(BOXES[key][2]):
        text, style = l if isinstance(l, tuple) else (l, "normal")
        ax.text(x + M, y + h - 0.20 - i * LINE, text, fontsize=BODY_FS, color=body, va="top", style=style, zorder=3)


def panel(x, y, width, h, label, colour):
    """A framed section with its label inside the frame, the same for both branches."""
    ax.add_patch(FancyBboxPatch((x, y), width, h, boxstyle="round,pad=0,rounding_size=0.06",
                                linewidth=0.7, edgecolor=colour, facecolor="white", zorder=1))
    ax.text(x + M, y + h - 0.04, label, color=colour, fontsize=LABEL_FS, weight="bold", va="top", zorder=3)


def arrow(x0, y0, x1, y1, color=GREY):
    ax.annotate("", xy=(x1, y1), xytext=(x0, y0), zorder=4,
                arrowprops=dict(arrowstyle="-|>", lw=0.8, color=color, mutation_scale=7, shrinkA=0, shrinkB=0))


# branch 1: counterfactual experiment; branch 2: observational analysis of the ranks of step 1
panel(x_panel, y_top, w_panel, h_panel_top, "Counterfactual experiment", BLUE)
panel(x_panel, y_bot, w_panel, h_panel_bot, "Observational analysis", GREY)
yb_top = y_top + PAD
yb_bot = y_bot + PAD
box("edit", x_edit, yb_top, H_BOX, edge=BLUE, fill=BLUE_FILL)
box("again", x_again, yb_top, H_BOX, edge=BLUE, fill=BLUE_FILL)
box("tests_exp", x_tests, yb_top, H_BOX, edge=BLUE, fill=BLUE, ink="white", body="white")
box("tests_obs", x_edit, yb_bot, H_OBS)

# inputs and step 1 in one row, centred between the two branches
y_shared = (y_bot + y_top + h_panel_top) / 2 - H_BOX / 2
box("inputs", x_inputs, y_shared, H_BOX)
box("rank", x_rank, y_shared, H_BOX)

y_mid_top = yb_top + H_BOX / 2
y_mid_bot = yb_bot + H_OBS / 2
arrow(x_inputs + w + 0.01, y_shared + H_BOX / 2, x_rank - 0.01, y_shared + H_BOX / 2)
# step 1 forks into the two branches: one line leaves the middle of step 1, splits at half the gap and enters each
# branch at the middle of its frame; both branches are equally high and step 1 is centred between them, so the fork is symmetric
y_c = y_shared + H_BOX / 2
y_in_top = y_top + h_panel_top / 2
y_in_bot = y_bot + h_panel_bot / 2
x_fork = x_rank + w + gap / 2
assert abs((y_in_top - y_c) - (y_c - y_in_bot)) < 1e-9
ax.plot([x_rank + w + 0.01, x_fork], [y_c, y_c], color=GREY, lw=0.8, zorder=4, solid_capstyle="butt")
ax.plot([x_fork, x_fork], [y_in_bot, y_in_top], color=GREY, lw=0.8, zorder=4, solid_capstyle="butt")
arrow(x_fork, y_in_top, x_panel - 0.01, y_in_top)
arrow(x_fork, y_in_bot, x_panel - 0.01, y_in_bot)
arrow(x_edit + w + 0.01, y_mid_top, x_again - 0.01, y_mid_top, BLUE)
arrow(x_again + w + 0.01, y_mid_top, x_tests - 0.01, y_mid_top, BLUE)

# control: every line keeps at least the left margin on the right
worst = min(WIDTHS[k] - (line_width(l) + M) for k in BOXES for l in BOXES[k][2])
worst = min(worst, min(WIDTHS[k] - (text_width(head_of(k), TITLE_FS, True) + M) for k in BOXES))
print(f"box width: {w:.3f} in; gap: {gap:.3f} in; smallest right margin: {worst:.3f} in (left margin {M:.3f} in)")
assert worst >= M - 1e-6

FIG_DESIGN_PNG.parent.mkdir(parents=True, exist_ok=True)
fig.savefig(FIG_DESIGN_PNG, dpi=300)

print(f"written: {FIG_DESIGN_PNG}")
```

## 19. What the fifths of the balance hold

- **Question:** The x axis of Figure 5 cuts the advertisements of an
  occupation into fifths of the balance, the communal minus the agentic
  density. Does a low fifth mean a low communal density, and how does the
  change per advertisement follow the balance?
- **Expectation:** none fixed in advance; the section describes the axis of
  Figure 5 and fixes no direction of its own.
- **Status:** added on 2026-11-08, descriptive; describes the axis of
  Figure 5. The correlation of the change with the communal density, which
  notebook 06 section 14.4 and section 6 above report, runs as a control and
  has to reproduce both published rows.
- **Unit and test:** advertisement. Means, medians and the share with an
  agentic surplus per fifth, cut within the occupation and with the
  occupations weighted equally as in the figure, and once over all
  advertisements; Pearson and Spearman between the communal and the agentic
  density within the occupations, with the span over the thirteen
  occupations; the change per advertisement against the balance with the test
  of notebook 06 section 14.4 unchanged, 5,000 permutations stratified by
  occupation, seed 42.
- **Output:** `study2_balance_fifths.csv`.

```python
# W19: what sits inside the fifths of the balance? The x axis of Figure 5 cuts
# the advertisements of an occupation into fifths of the balance, the communal
# minus the agentic density. A low fifth is not the same as a low communal
# density, so the section prints both densities per fifth, how strongly the two
# densities move together within an occupation, and the change per
# advertisement against the balance next to the published correlation with the
# communal density.
from scipy.stats import pearsonr as _pearsonr_w19
BALANCE_FIFTHS_FILE = Path(OUT_BASE) / "study2_balance_fifths.csv"
FIG5_VALUES_SAVED = Path(OUT_BASE) / "study2_fig5_values.csv"
JD_LANG_ARM1 = Path(OUT_BASE) / "study2_jd_language_test.csv"
JD_LANG_ARM2 = Path(OUT_BASE) / "study2_arm2_jd_language_test.csv"
N_PARTS_W19 = 5           # the fifths of Figure 5
DENSITIES = (("comm_dens", "communal"), ("ag_dens", "agentic"),
             ("balance", "balance"))

for _path in (FIG5_VALUES_SAVED, JD_LANG_ARM1, JD_LANG_ARM2):
    if not _path.exists():
        raise FileNotFoundError(
            f"{_path} not found. The section is controlled against the counts "
            f"of Figure 5 and against the published correlation with the "
            f"communal density.")


def _fifths_w19(values):
    """The fifths of a quantity, cut within one occupation, as in section 12."""
    return pd.qcut(values.rank(method="first"), N_PARTS_W19, labels=False,
                   duplicates="drop") + 1


_jd_w19 = df_jd_eda[["occ", "jd_idx", "comm_dens", "ag_dens",
                     "balance"]].copy()
_jd_w19["jd_fifth"] = _jd_w19.groupby("occ")["balance"].transform(_fifths_w19)
print(f"advertisements {len(_jd_w19)} in {_jd_w19['occ'].nunique()} "
      f"occupations, fifths cut within the occupation on the balance of "
      f"dictionary {LEX_MAIN}")

# Control: the balance is the one of the advertisement table, and the fifths
# hold the advertisements the saved values of Figure 5 count.
_merged_w19 = _jd_w19.merge(df_jd_level[["occ", "jd_idx", "balance"]],
                            on=["occ", "jd_idx"], suffixes=("", "_level"))
_worst_bal = float((_merged_w19["balance"]
                    - _merged_w19["balance_level"]).abs().max())
_counts_here = (_jd_w19.groupby("jd_fifth").size()
                .reindex(range(1, N_PARTS_W19 + 1)).to_numpy(int))
_saved_fig5 = pd.read_csv(FIG5_VALUES_SAVED)
_counts_saved = (_saved_fig5[(_saved_fig5["screener"] == "arm1")
                             & (_saved_fig5["panel"] == "a")]
                 .sort_values("jd_fifth")["n_ads"].to_numpy(int))
print(f"control balance against study2_jd_level.csv: largest deviation "
      f"{_worst_bal:.3g} (expected below 1e-09); advertisements per fifth "
      f"{list(_counts_here)}, saved for Figure 5 {list(_counts_saved)}")
if _worst_bal > 1e-9 or not np.array_equal(_counts_here, _counts_saved):
    raise AssertionError(
        f"The fifths of this section are not the ones of Figure 5: balance "
        f"deviation {_worst_bal:.3g}, counts {list(_counts_here)} against "
        f"{list(_counts_saved)}.")

# What the fifths hold. The figure weights the occupations equally, so
# the statistic is taken within an occupation first and then averaged over the
# thirteen; the pooled values over all advertisements follow below.
_by_occ_w19 = (_jd_w19.groupby(["occ", "jd_fifth"])
               .agg(n_ads=("balance", "size"),
                    comm_mean=("comm_dens", "mean"),
                    comm_median=("comm_dens", "median"),
                    ag_mean=("ag_dens", "mean"),
                    ag_median=("ag_dens", "median"),
                    balance_mean=("balance", "mean"),
                    balance_median=("balance", "median"),
                    share_agentic=("balance", lambda s: float((s < 0).mean())))
               .reset_index())
_STATS = ["comm_mean", "comm_median", "ag_mean", "ag_median", "balance_mean",
          "balance_median", "share_agentic"]
_fifth_rows = []
for _weighting in ("occupations equal", "all advertisements"):
    print()
    print(f"Densities per 100 words by fifth of the balance ({_weighting})")
    print(f"  {'fifth':>5} {'n_ads':>6} {'comm_mean':>10} {'comm_med':>9} "
          f"{'ag_mean':>8} {'ag_med':>7} {'bal_mean':>9} {'bal_med':>8} "
          f"{'share_bal_below_0':>18}")
    print("  " + "-" * 86)
    for _part in range(1, N_PARTS_W19 + 1):
        if _weighting == "occupations equal":
            _sub = _by_occ_w19[_by_occ_w19["jd_fifth"] == _part]
            _values = {_s: float(_sub[_s].mean()) for _s in _STATS}
            _n = int(_sub["n_ads"].sum())
            _n_occ = int(len(_sub))
        else:
            _sub = _jd_w19[_jd_w19["jd_fifth"] == _part]
            _values = {
                "comm_mean": float(_sub["comm_dens"].mean()),
                "comm_median": float(_sub["comm_dens"].median()),
                "ag_mean": float(_sub["ag_dens"].mean()),
                "ag_median": float(_sub["ag_dens"].median()),
                "balance_mean": float(_sub["balance"].mean()),
                "balance_median": float(_sub["balance"].median()),
                "share_agentic": float((_sub["balance"] < 0).mean())}
            _n = int(len(_sub))
            _n_occ = int(_sub["occ"].nunique())
        print(f"  {_part:>5} {_n:>6} {_values['comm_mean']:>10.3f} "
              f"{_values['comm_median']:>9.3f} {_values['ag_mean']:>8.3f} "
              f"{_values['ag_median']:>7.3f} {_values['balance_mean']:>+9.3f} "
              f"{_values['balance_median']:>+8.3f} "
              f"{_values['share_agentic']:>18.3f}")
        _fifth_rows.append({"block": "fifths", "weighting": _weighting,
                            "jd_fifth": _part, "n_ads": _n,
                            "n_occupations": _n_occ, **_values})

# Do the two densities move together? Within an occupation, because
# the fifths are cut within the occupation; the pooled values centre each
# occupation, the span reports the thirteen occupations themselves.
_corr_rows = []
print()
print(f"Communal against agentic density of the advertisements, "
      f"within the occupations")
print(f"  {'occ':<20} {'n_ads':>6} {'pearson_r':>10} {'spearman_rho':>13}")
print("  " + "-" * 52)
for occ in KEPT_13:
    _sub = _jd_w19[_jd_w19["occ"] == occ]
    _r = float(_pearsonr_w19(_sub["comm_dens"], _sub["ag_dens"])[0])
    _rho = float(spearmanr(_sub["comm_dens"], _sub["ag_dens"])[0])
    print(f"  {occ:<20} {len(_sub):>6} {_r:>+10.3f} {_rho:>+13.3f}")
    _corr_rows.append({"block": "density_correlation", "occ": occ,
                       "n_ads": int(len(_sub)), "pearson_r": _r,
                       "spearman_rho": _rho})
_centred = _jd_w19.copy()
for _column in ("comm_dens", "ag_dens"):
    _centred[f"{_column}_c"] = (_centred.groupby("occ")[_column]
                                .transform(lambda s: s - s.mean()))
    _centred[f"{_column}_rank_c"] = (
        _centred.groupby("occ")[_column]
        .transform(lambda s: s.rank() - s.rank().mean()))
_r_pooled = float(_pearsonr_w19(_centred["comm_dens_c"],
                                _centred["ag_dens_c"])[0])
_rho_pooled = float(_pearsonr_w19(_centred["comm_dens_rank_c"],
                                  _centred["ag_dens_rank_c"])[0])
_r_span = [float(min(_r["pearson_r"] for _r in _corr_rows)),
           float(max(_r["pearson_r"] for _r in _corr_rows))]
_rho_span = [float(min(_r["spearman_rho"] for _r in _corr_rows)),
             float(max(_r["spearman_rho"] for _r in _corr_rows))]
print(f"  pooled within the occupations (each occupation centred): "
      f"pearson_r {_r_pooled:+.3f}, spearman_rho {_rho_pooled:+.3f}, "
      f"advertisements {len(_centred)}")
print(f"  span over the 13 occupations: pearson_r {_r_span[0]:+.3f} to "
      f"{_r_span[1]:+.3f}, spearman_rho {_rho_span[0]:+.3f} to "
      f"{_rho_span[1]:+.3f}")
_corr_rows.append({"block": "density_correlation", "occ": "pooled, centred",
                   "n_ads": int(len(_centred)), "pearson_r": _r_pooled,
                   "spearman_rho": _rho_pooled})

# The change per advertisement against the balance, with the test of
# notebook 06 section 14.4 unchanged: Spearman within the occupations, 5,000
# permutations stratified by occupation, seed 42. The correlation with the
# communal density runs as a control and has to reproduce the published row.
# The change of Arm 1 is read from the saved advertisement table. The values
# are the ones notebook 06 computed, but writing and reading them moves the
# last bits, and where two advertisements carry the same change the ranks
# behind the correlation can then come out in another order. That shows in the
# sixth decimal of rho, so the control allows 1e-05 and not 1e-06.
CONTROL_TOL = 1e-5
_lang_sources = [("arm1", "SBERT", df_jd_level),
                 ("arm2", "Qwen3-Embedding", df_arm2_jd_level)]
_lang_rows = []
print()
print(f"  {'screener':<10} {'arm':<10} {'ad quantity':<13} {'n_ads':>6} "
      f"{'rho':>7} {'p_rho':>8} {'diff hi-lo':>11} {'p_diff':>8}")
print("  " + "-" * 79)
for _screener, _label, _frame in _lang_sources:
    for arm in ARMS:
        _delta_by_occ, _dens_by_occ = {}, {}
        for occ in KEPT_13:
            _level = _frame[_frame["occ"] == occ].sort_values("jd_idx")
            _delta_by_occ[occ] = _level[f"dj_pool_{arm}_rank"].to_numpy(float)
            _dens_by_occ[occ] = {
                "balance": _level["balance"].to_numpy(float),
                "comm_dens": _level["comm_dens"].to_numpy(float)}
        for _quantity in ("balance", "comm_dens"):
            if _quantity == "comm_dens" and arm != "communion":
                continue
            _, _pool = study2_core.jd_language_test(
                _delta_by_occ, {occ: _dens_by_occ[occ][_quantity]
                                for occ in KEPT_13},
                n_perm=N_PERM, seed=SEED)
            print(f"  {_screener:<10} {arm:<10} {_quantity:<13} "
                  f"{int(_pool['n_jds']):>6} {_pool['rho']:>+7.3f} "
                  f"{_pool['p_rho']:>8.4f} {_pool['diff_hi_lo']:>+11.3f} "
                  f"{_pool['p_diff']:>8.4f}")
            _lang_rows.append({
                "block": "delta_vs_ad_quantity", "screener": _screener,
                "encoder": _label, "arm": arm, "jd_quantity": _quantity,
                "n_occupations": int(_pool["n_occ"]),
                "n_ads": int(_pool["n_jds"]), "rho": float(_pool["rho"]),
                "p_rho": float(_pool["p_rho"]),
                "diff_hi_lo": float(_pool["diff_hi_lo"]),
                "p_diff": float(_pool["p_diff"])})

_saved_arm1 = pd.read_csv(JD_LANG_ARM1)
_saved_arm1 = _saved_arm1[_saved_arm1["test_key"]
                          == "pool_delta_rank_arm=communion_jd=communion"]
_saved_arm2 = pd.read_csv(JD_LANG_ARM2)
_saved_arm2 = _saved_arm2[(_saved_arm2["delta"] == "delta_rank")
                          & (_saved_arm2["arm"] == "communion")
                          & (_saved_arm2["jd_dimension"] == "communion")]
_worst_lang = 0.0
for _screener, _saved in (("arm1", _saved_arm1), ("arm2", _saved_arm2)):
    _here = [_r for _r in _lang_rows if _r["screener"] == _screener
             and _r["arm"] == "communion"
             and _r["jd_quantity"] == "comm_dens"][0]
    for _column in ("rho", "p_rho", "diff_hi_lo", "p_diff"):
        _worst_lang = max(_worst_lang, abs(_here[_column]
                                           - float(_saved[_column].iloc[0])))
print()
print(f"control correlation with the communal density against "
      f"study2_jd_language_test.csv and study2_arm2_jd_language_test.csv: "
      f"largest deviation {_worst_lang:.3g} (expected below {CONTROL_TOL:.0e})")
if _worst_lang > CONTROL_TOL:
    raise AssertionError(
        f"The test of this section does not reproduce the published "
        f"correlation with the communal density; it deviates by "
        f"{_worst_lang:.3g}.")

df_balance_fifths = pd.DataFrame(_fifth_rows + _corr_rows + _lang_rows)
df_balance_fifths.to_csv(BALANCE_FIFTHS_FILE, index=False, encoding="utf-8")
print(f"written: {BALANCE_FIFTHS_FILE} ({len(df_balance_fifths)} rows)")
```

## 20. Two counts the write-up uses

- **Question:** How many occurrences did the protocol replace and how many did
  it delete in a changed biography of the pool, how many markers of the
  manipulated dimension did those biographies hold and how many of them did
  the edit reach, and what do the thirteen occupations of the experiment show
  under the search encoder of Arm 2?
- **Expectation:** none fixed in advance; the section prints counts the
  write-up quotes and fixes no direction of its own.
- **Status:** added on 2026-11-09, descriptive, the share of the markers the
  edit reached added on 2026-11-14. The counts of the protocol are controlled
  against the columns of `study2_delta_r5_pool.csv` and against
  `manipulation_check.csv`.
- **Unit and test:** occurrence of the protocol, and advertisement. Counts,
  sums, shares and means; no test of its own. The denominator of the share is
  the protocol itself, which holds one row per marker of that dimension the
  engine saw. Where a rule or a fixed term covers more than one word, the
  protocol writes one row for the span while the counter of dictionary A
  counts every entry inside it, so the protocol can hold fewer rows than the
  counter finds markers, never more. The occupation rows are the ones
  section 11 wrote, and their p values are nominal.
- **Output:** none; the section prints.

```python
# Three counts the write-up uses. First, how much the protocol changed in a
# biography: the replaced and the deleted occurrences of the pool per arm, and
# their sum per changed biography, once over every changed biography and once
# at the gate of the main analysis. Second, how many markers of the
# manipulated dimension those biographies held and how many of them the edit
# reached, with the untreated ones split by the reason the protocol gives.
# Third, the thirteen occupation rows of the experiment under the search
# encoder of Arm 2 against the communal density of the advertisement, with the
# count of the negative ones.
ARM2_PER_OCC_FILE = Path(OUT_BASE) / "study2_arm2_jd_language_test_per_occ.csv"
ACTIONS = ("replaced", "deleted")
# The reasons the write-up names one by one; every other reason of the
# protocol is counted as the rest.
REASONS_NAMED = ("proper_name", "no_rule_for_word", "fixed_term",
                 "protect_comp", "validation")
if not ARM2_PER_OCC_FILE.exists():
    raise FileNotFoundError(
        f"{ARM2_PER_OCC_FILE} not found. Section 11 of this notebook writes "
        f"the occupation rows of Arm 2.")

# The occurrences of the protocol that changed something, joined to the pool so
# that each one carries the gate of its biography.
_log_w21 = pd.read_csv(PROTOCOL_FILE)
_occ_w21 = _log_w21[(_log_w21["sample"] == "pool")
                    & (_log_w21["action_taken"].isin(ACTIONS))].merge(
    df_pool[["occ", "bio_idx", "arm", "n_subs", "n_replaced", "n_deleted",
             "sim_cp"]], on=["occ", "bio_idx", "arm"], how="inner")
print(f"occurrences of the pool that changed something: {len(_occ_w21)} in "
      f"{_occ_w21[['occ', 'bio_idx', 'arm']].drop_duplicates().shape[0]} "
      f"biographies")

# Control: the protocol and the pool table have to agree on how many
# occurrences a biography carries.
_per_bio_w21 = (_occ_w21.groupby(["occ", "bio_idx", "arm"])
                .agg(n_log=("rule_id", "size"), n_subs=("n_subs", "first"),
                     n_replaced=("n_replaced", "first"),
                     n_deleted=("n_deleted", "first")).reset_index())
_gap_w21 = int((_per_bio_w21["n_log"] - _per_bio_w21["n_subs"]).abs().max())
_gap_sum = int((_per_bio_w21["n_subs"] - _per_bio_w21["n_replaced"]
                - _per_bio_w21["n_deleted"]).abs().max())
print(f"control occurrences per biography, protocol against "
      f"study2_delta_r5_pool.csv: largest deviation {_gap_w21} (expected 0); "
      f"n_subs against n_replaced + n_deleted: {_gap_sum} (expected 0)")
if _gap_w21 or _gap_sum:
    raise AssertionError(
        f"The protocol and the pool table disagree on the number of "
        f"occurrences per biography: {_gap_w21} and {_gap_sum}.")

print()
print(f"Changed biographies of the pool and the occurrences the protocol "
      f"changed in them")
print(f"  {'sample':<20} {'arm':<10} {'bios':>6} {'replaced':>9} "
      f"{'deleted':>8} {'sum':>7} {'per_bio_mean':>13} {'per_bio_median':>15}")
print("  " + "-" * 93)
for _label, _subset in (("every changed bio", _occ_w21),
                        (f"gate {PRIMARY_GATE}",
                         _occ_w21[_occ_w21["sim_cp"] >= PRIMARY_GATE])):
    for arm in ARMS:
        _arm_rows = _subset[_subset["arm"] == arm]
        _bios = _arm_rows.groupby(["occ", "bio_idx"]).size()
        _n_rep = int((_arm_rows["action_taken"] == "replaced").sum())
        _n_del = int((_arm_rows["action_taken"] == "deleted").sum())
        print(f"  {_label:<20} {arm:<10} {len(_bios):>6} {_n_rep:>9} "
              f"{_n_del:>8} {_n_rep + _n_del:>7} "
              f"{(_n_rep + _n_del) / len(_bios):>13.4f} "
              f"{float(_bios.median()):>15.1f}")

# The markers of the manipulated dimension in the changed biographies of the
# gate, and how many of them the edit reached. The denominator is the protocol
# itself: it holds one row per occurrence of a marker of that dimension that
# the engine saw, whether it changed the word or not, so the share below is
# the share of the markers the edit removed from a biography it changed.
_all_w34 = _log_w21[_log_w21["sample"] == "pool"].merge(
    df_pool[["occ", "bio_idx", "arm", "sim_cp", "n_subs"]],
    on=["occ", "bio_idx", "arm"], how="inner")
_all_w34 = _all_w34[(_all_w34["sim_cp"] >= PRIMARY_GATE)
                    & (_all_w34["n_subs"] >= 1)]
_all_w34 = _all_w34.assign(
    treated=_all_w34["action_taken"].isin(ACTIONS).astype(int))
_check_w34 = df_manip_check[df_manip_check["sample"] == "pool"].merge(
    df_pool[["occ", "bio_idx", "arm", "sim_cp", "n_subs"]],
    on=["occ", "bio_idx", "arm"], how="inner")
_check_w34 = _check_w34[(_check_w34["sim_cp"] >= PRIMARY_GATE)
                        & (_check_w34["n_subs"] >= 1)]

print()
print(f"Markers of the manipulated dimension in the changed biographies at "
      f"the gate {PRIMARY_GATE}")
print(f"  {'arm':<10} {'bios':>6} {'markers':>8} {'replaced':>9} "
      f"{'deleted':>8} {'treated':>8} {'share':>7} {'median_bio':>11} "
      f"{'mean_bio':>9}")
print("  " + "-" * 84)
_rows_w34 = {}
for arm in ARMS:
    _arm_rows = _all_w34[_all_w34["arm"] == arm]
    _rows_w34[arm] = _arm_rows
    _per_bio = (_arm_rows.groupby(["occ", "bio_idx"])
                .agg(n=("treated", "size"), t=("treated", "sum")))
    _share_bio = _per_bio["t"] / _per_bio["n"]
    print(f"  {arm:<10} {len(_per_bio):>6} {len(_arm_rows):>8} "
          f"{int((_arm_rows['action_taken'] == 'replaced').sum()):>9} "
          f"{int((_arm_rows['action_taken'] == 'deleted').sum()):>8} "
          f"{int(_arm_rows['treated'].sum()):>8} "
          f"{_arm_rows['treated'].mean():>7.4f} "
          f"{float(_share_bio.median()):>11.4f} "
          f"{float(_share_bio.mean()):>9.4f}")

print()
print(f"The markers the edit left in place, by the reason the protocol gives")
print(f"  {'arm':<10} {'untreated':>10} " + " ".join(
    f"{_reason:>17}" for _reason in REASONS_NAMED) + f" {'rest':>6}")
print("  " + "-" * 118)
for arm in ARMS:
    _untreated = _rows_w34[arm][_rows_w34[arm]["treated"] == 0]
    _counts = [int((_untreated["reason_untreated"] == _reason).sum())
               for _reason in REASONS_NAMED]
    print(f"  {arm:<10} {len(_untreated):>10} "
          + " ".join(f"{_n:>17}" for _n in _counts)
          + f" {len(_untreated) - sum(_counts):>6}")

# Control: the manipulation check counts the markers of the same biographies
# with the counter of dictionary A. Where a rule or a fixed term covers more
# than one word, the protocol writes one row for the span while the counter
# counts every dictionary word inside it, so the protocol holds one row fewer
# in a few biographies. The number of markers the edit treated is the same in
# the protocol, in the check and in the pool table, and the control stops the
# run if a biography carries more rows than the counter finds markers or if
# the two disagree on what was treated.
print()
print(f"The same two counts from the manipulation check of notebook 06 "
      f"section 7")
print(f"  {'arm':<10} {'protocol rows':>14} {'markers of the counter':>23} "
      f"{'bios that differ':>17} {'treated in the check':>21}")
print("  " + "-" * 90)
for arm in ARMS:
    _arm_rows = _rows_w34[arm]
    _arm_check = _check_w34[_check_w34["arm"] == arm]
    _per_bio = (_arm_rows.groupby(["occ", "bio_idx"])
                .agg(n=("treated", "size"), t=("treated", "sum"))
                .reset_index()
                .merge(_arm_check[["occ", "bio_idx", "hits_before",
                                   "n_replaced", "n_deleted"]],
                       on=["occ", "bio_idx"]))
    _more = int((_per_bio["n"] > _per_bio["hits_before"]).sum())
    _differ = int((_per_bio["n"] != _per_bio["hits_before"]).sum())
    _treated_gap = int((_per_bio["t"] - _per_bio["n_replaced"]
                        - _per_bio["n_deleted"]).abs().max())
    _treated_check = int((_arm_check["n_replaced"]
                          + _arm_check["n_deleted"]).sum())
    print(f"  {arm:<10} {len(_arm_rows):>14} "
          f"{int(_arm_check['hits_before'].sum()):>23} {_differ:>17} "
          f"{_treated_check:>21}")
    if _more or _treated_gap:
        raise AssertionError(
            f"{arm}: the protocol holds more rows than the counter finds "
            f"markers in {_more} biographies, and it disagrees with the check "
            f"on the number treated by {_treated_gap}.")

# The occupation rows of the experiment under Arm 2, against the communal
# density of the advertisement: the same quantity notebook 06 section 14.4
# reports for Arm 1.
_arm2_occ_w21 = pd.read_csv(ARM2_PER_OCC_FILE)
_arm2_occ_w21 = _arm2_occ_w21[(_arm2_occ_w21["arm"] == "communion")
                              & (_arm2_occ_w21["jd_dimension"] == "communion")]
_arm2_occ_w21 = _arm2_occ_w21.set_index("occ").reindex(KEPT_13).reset_index()
if _arm2_occ_w21["rho"].isna().any():
    raise AssertionError(
        f"The occupation rows of Arm 2 are not complete: "
        f"{list(_arm2_occ_w21[_arm2_occ_w21['rho'].isna()]['occ'])}.")

print()
print(f"Arm 2, the change per advertisement against the communal density, "
      f"by occupation")
print(f"  {'occ':<20} {'n_ads':>6} {'rho':>7} {'p_rho':>8}")
print("  " + "-" * 45)
for _, _row in _arm2_occ_w21.iterrows():
    print(f"  {_row['occ']:<20} {int(_row['n_jds']):>6} {_row['rho']:>+7.3f} "
          f"{_row['p_rho']:>8.4f}")
_n_neg_w21 = int((_arm2_occ_w21["rho"] < 0).sum())
_n_neg_sig = int(((_arm2_occ_w21["rho"] < 0)
                  & (_arm2_occ_w21["p_rho"] < 0.05)).sum())
print(f"  occupations with a negative rho: {_n_neg_w21} of "
      f"{len(_arm2_occ_w21)}, of them with p_rho below 0.05: {_n_neg_sig}")
print(f"  the p values are nominal; they carry no correction over the "
      f"thirteen occupations")
```

## 21. Figure of Appendix E: the other arm and the other sample

- **Question:** Section 12 holds the change of the biographies the
  communal manipulation changed, by fifth of the communal balance of the
  advertisement, for the pool. What does the same picture look like for the
  agentic arm and for the focal sample?
- **Expectation:** none fixed in advance; the section repeats a picture and
  fixes no direction of its own.
- **Status:** added on 2026-11-10, descriptive; Appendix E,
  revised on 2026-11-13 (one scale: communal balance). The route is controlled against the values of section 12,
  which it has to reproduce.
- **Unit and test:** advertisement; the mean change of the changed
  biographies per advertisement, averaged over the fifths with the
  occupations weighted equally, as in section 12. No test. The three panels
  share one y axis, but the pool and the focal sample divide the percentile by
  a different number of competitors, 249 and 250, so the panels are read for
  their course over the fifths and not for their height against each other.
- **Output:** `outputs/figures/figE1_experiment_by_arm_sample.png`, Figure E1.

```python
# A3: the figure of Appendix E. Section 12 holds the mean change of the
# biographies the communal manipulation changed, by fifth of the communal
# balance of the advertisement, for the pool. The same picture is drawn here
# for the agentic arm and for the focal sample, with whatever screener has
# values for that combination. Nothing is computed but these means: the change
# per advertisement is the one notebook 06 and section 11 of this notebook
# wrote, and the fifths are cut as in section 12.
FIG_E1_PNG = Path("outputs/figures/figE1_experiment_by_arm_sample.png")
FOCAL_FILE = Path(OUT_BASE) / "study2_delta_r5_focal.csv"
ARM3_BY_JD = Path(OUT_BASE) / "arm3" / "main" / "study2_arm3_pct_by_jd.csv"
# The pool and the focal sample divide the percentile by a different number
# of competitors, 249 and 250, so the level of a line of one sample is not the
# level of the other. The three panels share one y axis, and they are read for
# their course over the fifths, not for their height against each other.
PANELS_E1 = [("agency", "pool", "(a) Agentic arm, mean"),
             ("communion", "focal", "(b) Communal arm, focal sample, mean"),
             ("agency", "focal", "(c) Agentic arm, focal sample, mean")]
if not FOCAL_FILE.exists():
    raise FileNotFoundError(
        f"{FOCAL_FILE} not found. The number of changed biographies behind "
        f"the focal panels is read from it.")

df_focal_e1 = pd.read_csv(FOCAL_FILE)


def _series_e1(arm, sample, screener):
    """The change per advertisement for one arm, sample and screener, or None
    if that combination was never ranked."""
    _column = f"dj_{sample}_{arm}_rank"
    if screener == "arm1":
        _frame, _parts = df_jd_level, _jd_parts
    elif screener == "arm2":
        _frame, _parts = df_arm2_jd_level, _jd_parts
    else:
        if sample != "pool" or not ARM3_BY_JD.exists():
            return None
        _pairs = pd.read_csv(ARM3_BY_JD)
        _pairs = _pairs[_pairs["arm"] == arm].copy()
        _pairs["jd_idx"] = [int(str(_j).rsplit("jd", 1)[1])
                            for _j in _pairs["jd_id"]]
        _changed = df_pool[(df_pool["sim_cp"] >= PRIMARY_GATE)
                           & (df_pool["n_subs"] >= 1)
                           & (df_pool["arm"] == arm)]
        _pairs = _pairs.merge(_changed[["occ", "bio_idx"]],
                              on=["occ", "bio_idx"])
        _frame = (_pairs.groupby(["occ", "jd_idx"])["delta_pct"].mean()
                  .reset_index().rename(columns={"delta_pct": _column}))
        _parts = _arm3_parts
    if _column not in _frame.columns:
        return None
    return (_frame[["occ", "jd_idx", _column]]
            .rename(columns={_column: "a"})
            .merge(_parts[["occ", "jd_idx", "jd_fifth"]],
                   on=["occ", "jd_idx"]))


# Control: the same route has to give the values of section 12 back.
_check_e1 = _summarise(_series_e1("communion", "pool", "arm1"), "a", "arm1",
                       "a")
_saved_e1 = (df_fig5[(df_fig5["screener"] == "arm1")
                     & (df_fig5["panel"] == "a")].sort_values("jd_fifth"))
_worst_e1 = float(np.abs(_check_e1["mean_occupations_equal"].to_numpy()
                         - _saved_e1["mean_occupations_equal"].to_numpy()).max())
print(f"control the communal arm of the pool against the values of section "
      f"12: largest deviation {_worst_e1:.3g} (expected below 1e-09)")
if _worst_e1 > 1e-9:
    raise AssertionError(
        f"The means of this section do not reproduce the values of "
        f"section 12; they deviate by {_worst_e1:.3g}.")

# The biographies behind a panel, at the gate of the main analysis.
_bios_e1 = {}
for _arm, _sample, _ in PANELS_E1:
    _source = df_pool if _sample == "pool" else df_focal_e1
    _bios_e1[(_arm, _sample)] = int(((_source["arm"] == _arm)
                                     & (_source["sim_cp"] >= PRIMARY_GATE)
                                     & (_source["n_subs"] >= 1)).sum())

_rows_e1, _drawn_e1 = [], []
for _arm, _sample, _title in PANELS_E1:
    _lines = []
    for _screener, _label in SCREENERS:
        _frame = _series_e1(_arm, _sample, _screener)
        if _frame is None:
            print(f"{_title}: no values for {_label}; the line is left out")
            continue
        _lines.append((_screener, _summarise(_frame, "a", _screener, _title)))
    if not _lines:
        print(f"{_title}: no screener has values; the panel is left out")
        continue
    _drawn_e1.append((_arm, _sample, _title, _lines))
    for _screener, _table in _lines:
        _table = _table.assign(arm=_arm, sample=_sample,
                               n_bios=_bios_e1[(_arm, _sample)])
        _rows_e1.append(_table)
df_figE1 = pd.concat(_rows_e1, ignore_index=True)

for _arm, _sample, _title, _lines in _drawn_e1:
    print()
    print(f"{_title}: changed biographies at the gate of {PRIMARY_GATE}: "
          f"{_bios_e1[(_arm, _sample)]}")
    print(f"  {'screener':<16} {'n_ads':>6} " + " ".join(
        f"{'fifth ' + str(_p):>10}" for _p in range(1, N_PARTS + 1)))
    print("  " + "-" * 78)
    for _screener, _table in _lines:
        _table = _table.sort_values("jd_fifth")
        print(f"  {LABELS[_screener]:<16} "
              f"{int(_table['n_ads'].sum()):>6} " + " ".join(
                  f"{_v:>+10.3f}"
                  for _v in _table["mean_occupations_equal"]))

fig_e1, axes_e1 = plt.subplots(1, len(_drawn_e1), figsize=(5.12, 2.1),
                               sharey=True, squeeze=False)
_x_e1 = np.arange(1, N_PARTS + 1)
for _ax, (_arm, _sample, _title, _lines) in zip(axes_e1[0], _drawn_e1):
    for _screener, _table in _lines:
        _table = _table.sort_values("jd_fifth")
        _ax.plot(_table["jd_fifth"].to_numpy(),
                 _table["mean_occupations_equal"].to_numpy(), "-o",
                 color=COLOURS[_screener], lw=1.1, ms=2.6,
                 label=FIG_LABELS[_screener])
    _ax.axhline(0, color=GREY, lw=0.7, zorder=1)
    # Three panels share the width of one figure here, so the titles are
    # set smaller than in Figure 5 and Figure 6. The ticks are the ones of
    # those two figures, which name the fifth and the mean balance of the
    # advertisements in it, at the size they carry there.
    _ax.set_title(_title, fontsize=5.6, color=INK, pad=3)
    _ax.set_xticks(_x_e1)
    _ax.set_xticklabels(TICKS_FIFTHS, fontsize=TICKS_FIFTHS_SIZE)
    _ax.tick_params(axis="both", labelsize=6.0, colors=GREY, length=2)
    _ax.grid(axis="y", color="#ecebe7", lw=0.5)
    _ax.set_axisbelow(True)
    for _side in ("top", "right"):
        _ax.spines[_side].set_visible(False)
    for _side in ("bottom", "left"):
        _ax.spines[_side].set_color(LIGHT)
axes_e1[0][0].set_ylabel(r"Change in percentile rank ($\Delta$)",
                         fontsize=5.8, color=INK)
# One label for the three panels: it is wider than a single panel.
fig_e1.supxlabel(LABEL_FIFTHS, fontsize=LABEL_FIFTHS_SIZE, color=INK,
                 y=0.1)

_handles_e1, _labels_e1 = axes_e1[0][0].get_legend_handles_labels()
fig_e1.legend(_handles_e1, _labels_e1, loc="lower center",
              ncol=len(_labels_e1), fontsize=5.8, frameon=False,
              handlelength=1.4, handletextpad=0.4, columnspacing=1.4,
              bbox_to_anchor=(0.5, 0.0))
# The margins are set by hand, as in Figure 5 and Figure 6.
fig_e1.subplots_adjust(left=0.085, right=0.988, top=0.88, bottom=0.26,
                       wspace=0.04)
FIG_E1_PNG.parent.mkdir(parents=True, exist_ok=True)
fig_e1.savefig(FIG_E1_PNG, dpi=300)
plt.show()
print()
print(f"written: {FIG_E1_PNG}")
print(f"panels drawn: {[_t for _, _, _t, _ in _drawn_e1]}")
```

## 22. How the three screeners are configured

- **Question:** With what settings did the three screeners rank? Appendix D
  reports them, so they have to stand in print.
- **Expectation:** none fixed in advance; the section reads settings and fixes
  no direction of its own.
- **Status:** added on 2026-11-11, descriptive; Appendix D. No model is
  loaded and no weights are read. The settings come from the metadata the
  runs wrote, from the manifest of notebook 06, from the configuration files
  of the model in the local Hugging Face cache, and, for the form of the
  instruction of Arm 2, from the two functions of notebook 06a. What is not
  in any of these is named as missing and not reconstructed.
- **Unit and test:** none; the section reads settings.
- **Output:** none; the section prints the configuration of Table D1.

```python
# B2: how the three screeners are configured. Appendix D needs the settings in
# print: for the two encoders the pooling, the normalisation and the sequence
# length, for Arm 2 the instruction and the form it is put in front of the
# advertisement, and for Arm 3 the two prompts. Nothing is loaded here. The
# settings are read from the metadata the runs wrote, from the manifest of
# notebook 06, from the configuration files the local cache of Hugging Face
# holds for the model, and, for the form of the instruction, from the two
# functions of notebook 06a.
from huggingface_hub import try_to_load_from_cache

MANIFEST_FILE = Path(OUT_BASE) / "run_manifest.json"
ARM3_META_FILE = Path(OUT_BASE) / "llm" / "main" / "llm_run_metadata.json"
ARM2_NOTEBOOK = Path("notebooks/06a_arm2_embeddings_kaggle.ipynb")
POOLING_MODES = {"pooling_mode_cls_token": "CLS token",
                 "pooling_mode_mean_tokens": "mean of the tokens",
                 "pooling_mode_max_tokens": "maximum over the tokens",
                 "pooling_mode_mean_sqrt_len_tokens":
                     "mean divided by the square root of the length",
                 "pooling_mode_lasttoken": "last token"}
for _path in (MANIFEST_FILE, ARM3_META_FILE, ARM2_NOTEBOOK):
    if not _path.exists():
        raise FileNotFoundError(
            f"{_path} not found. The section prints the settings of the three "
            f"screeners from the files the runs wrote.")

manifest = json.loads(MANIFEST_FILE.read_text(encoding="utf-8"))
meta_arm3 = json.loads(ARM3_META_FILE.read_text(encoding="utf-8"))


def _cached_config(repo, name):
    """A configuration file of a model in the local cache, or None if the model
    was never downloaded on this machine."""
    _path = try_to_load_from_cache(repo, name)
    if not isinstance(_path, str) or not Path(_path).exists():
        return None
    return json.loads(Path(_path).read_text(encoding="utf-8"))


def _pooling_of(config):
    """The pooling modes a Sentence-Transformers pooling configuration sets."""
    return [_text for _key, _text in POOLING_MODES.items()
            if config.get(_key)] or ["none of the known modes"]


def _verbatim(label, text):
    """A stored string, once line by line and once as it is stored."""
    print(f"  {label}")
    for _line in text.split("\n"):
        print(f"    | {_line}")
    print(f"    stored as: {text!r}")


print("Arm 1: the search encoder of the main analysis")
_model_1 = manifest["primary_model"]
_pool_1 = _cached_config(_model_1, "1_Pooling/config.json")
_sbert_1 = _cached_config(_model_1, "sentence_bert_config.json")
_modules_1 = _cached_config(_model_1, "modules.json")
print(f"  model                 : {_model_1}")
if _pool_1 is None or _sbert_1 is None or _modules_1 is None:
    print(f"  pooling and sequence length: the configuration files of this "
          f"model are not in the local cache, so they are not printed here")
else:
    print(f"  pooling               : {', '.join(_pooling_of(_pool_1))} "
          f"(1_Pooling/config.json)")
    print(f"  modules               : "
          f"{[_m['type'].rsplit('.', 1)[-1] for _m in _modules_1]}")
    print(f"  max_seq_length        : {_sbert_1['max_seq_length']} "
          f"(sentence_bert_config.json), {manifest['max_seq_length']} in the "
          f"manifest of notebook 06")
print(f"  normalisation         : the model ends in a Normalize module, and "
      f"notebook 06 calls encode with normalize_embeddings=True")
print(f"  long texts            : notebook 06 embeds with encode_long in "
      f"windows of {manifest['encode_long_window']} tokens and averages the "
      f"windows; the gate of section 10 uses the plain encode, which cuts at "
      f"{manifest['max_seq_length']}")

print()
print("Arm 2: the search encoder of the supplement")
_model_2 = meta_arm2["model"]
_pool_2 = _cached_config(_model_2, "1_Pooling/config.json")
print(f"  model                 : {_model_2}")
print(f"  revision              : {meta_arm2['revision']}")
print(f"  max_seq_length        : {meta_arm2['max_seq_length']}, set from the "
      f"longest text of {meta_arm2['longest_text_tokens']} tokens; texts cut: "
      f"{meta_arm2['n_truncated']}")
print(f"  normalisation         : normalize_embeddings="
      f"{meta_arm2['normalised']}, so every vector has length one")
if _pool_2 is None:
    print(f"  pooling               : this model ran on Kaggle and is not in "
          f"the local cache, so its pooling configuration is not available "
          f"here and is not printed")
else:
    print(f"  pooling               : {', '.join(_pooling_of(_pool_2))} "
          f"(1_Pooling/config.json)")
_verbatim("instruction, as arm2_metadata.json stores it:",
          meta_arm2["instruction"])

# The form the instruction is put in: the two functions of notebook 06a decide
# what is encoded, so they are printed from that notebook and not described.
_nb_arm2 = json.loads(ARM2_NOTEBOOK.read_text(encoding="utf-8"))
_as_query = [_source for _source in
             ("".join(_cell["source"]) for _cell in _nb_arm2["cells"])
             if "def as_query(" in _source]
if len(_as_query) != 1:
    raise AssertionError(
        f"{ARM2_NOTEBOOK} holds {len(_as_query)} cells that define as_query; "
        f"the form of the instruction cannot be read from it.")
_block = _as_query[0].split("def as_query(")[1].split("tokenizer =")[0]
print("  what is encoded, from the cell of notebook 06a:")
for _line in ("def as_query(" + _block).rstrip().split("\n"):
    print(f"    | {_line}")
print(f"  so the advertisement replaces {{jd}} in the instruction and the "
      f"biography is encoded as it stands, without any instruction")

print()
print("Arm 3: the language model")
print(f"  model                 : {meta_arm3['model']}")
print(f"  digest                : {meta_arm3['model_digest']}")
print(f"  runtime               : {meta_arm3['ollama_version']}")
print(f"  temperature {meta_arm3['temperature']}  seed {meta_arm3['seed']}  "
      f"num_predict {meta_arm3['num_predict']}  num_ctx "
      f"{meta_arm3['num_ctx']}  top_logprobs {meta_arm3['top_logprobs']}  "
      f"think {meta_arm3['think']}")
print(f"  score                 : {meta_arm3['score']}")
_verbatim("system prompt, as llm_run_metadata.json stores it:",
          meta_arm3["system_prompt"])
_verbatim("user message, as llm_run_metadata.json stores it:",
          meta_arm3["user_template"])
_verbatim("the part of the user message that two pairs share:",
          meta_arm3["prefix_template"])
```

## 23. The length of the biographies before and after the substitution

- **Question:** Section 14 shows the rate of the other dimension a little
  lower after the substitution, 1.731 to 1.718 in the communal arm and 3.776
  to 3.756 in the agentic one. No rule puts a word of dictionary A or of
  dictionary B into a biography, so where does that movement come from?
- **Expectation:** the number of markers of the other dimension stays and the
  number of words moves, so the rate moves through its denominator. The
  direction follows from the protocol and is not a test.
- **Status:** added on 2026-11-12, descriptive; explains the small change of
  the other dimension in section 14. The counting is the one of section 14:
  the word counts are the ones `manipulation_check.csv` carries, and the
  markers of the other dimension are read back from the rate and the word
  count, which has to give whole numbers.
- **Unit and test:** biography; means, medians and counts over the
  biographies of the pool the manipulation changed, at the gate of the main
  analysis, with the row of every changed biography beside it so that the
  levels of section 14 can be found again. No test.
- **Output:** none; the section prints.

```python
# W22: why the rate of the other dimension falls a little in section 14. No
# rule inserts a word of dictionary A or B, so the rate of the other dimension
# can only move through its numerator, the markers of that dimension, or
# through its denominator, the number of words. Both stand in
# manipulation_check.csv, which notebook 06 section 7 wrote, and the counting
# is the one of section 14: words_before is the stored word count of the
# biography, words_after is that count moved by the change in whitespace
# tokens, and a rate is the number of markers of the dimension per 100 words.
# The markers of the other dimension are not stored as a count, so they are
# read back from the rate and the word count, which has to give whole numbers.
WORDS_ACTIONS = ("replaced", "deleted")

_mc_w22 = df_manip_check[(df_manip_check["sample"] == "pool")
                         & df_manip_check["changed"]].merge(
    df_pool[["occ", "bio_idx", "arm", "sim_cp"]],
    on=["occ", "bio_idx", "arm"], how="inner")
print(f"changed biographies of the pool: {len(_mc_w22)}, of them at the gate "
      f"{PRIMARY_GATE}: {int((_mc_w22['sim_cp'] >= PRIMARY_GATE).sum())}")

# Control: the rate times the word count gives the number of markers of the
# other dimension, so that number can be read back as a whole number.
for _rate, _words in (("rate_other_before", "words_before"),
                      ("rate_other_after", "words_after")):
    _raw = _mc_w22[_rate] * _mc_w22[_words] / 100
    _off = float(np.abs(_raw - _raw.round()).max())
    print(f"control {_rate} times {_words}: largest distance to a whole "
          f"number {_off:.3g} (expected below 1e-06)")
    if _off > 1e-6:
        raise AssertionError(
            f"{_rate} times {_words} is not a count of markers; it misses a "
            f"whole number by {_off:.3g}.")
_mc_w22["other_before"] = (_mc_w22["rate_other_before"]
                           * _mc_w22["words_before"] / 100).round().astype(int)
_mc_w22["other_after"] = (_mc_w22["rate_other_after"]
                          * _mc_w22["words_after"] / 100).round().astype(int)
# What the rate would be if the markers of the other dimension had stayed and
# only the length had moved: it separates the denominator from the numerator.
_mc_w22["rate_other_denom"] = (_mc_w22["other_before"]
                               / _mc_w22["words_after"] * 100)

_gated_w22 = _mc_w22[_mc_w22["sim_cp"] >= PRIMARY_GATE]
print()
print(f"Changed biographies of the pool at the gate {PRIMARY_GATE}: the words "
      f"and the markers of the other dimension")
print(f"  {'arm':<10} {'when':<7} {'n':>5} {'words_mean':>11} "
      f"{'words_median':>13} {'other_mean':>11} {'other_median':>13}")
print("  " + "-" * 77)
for arm in ARMS:
    _arm_rows = _gated_w22[_gated_w22["arm"] == arm]
    for _when, _words, _other in (("before", "words_before", "other_before"),
                                  ("after", "words_after", "other_after")):
        print(f"  {arm:<10} {_when:<7} {len(_arm_rows):>5} "
              f"{_arm_rows[_words].mean():>11.3f} "
              f"{float(_arm_rows[_words].median()):>13.1f} "
              f"{_arm_rows[_other].mean():>11.3f} "
              f"{float(_arm_rows[_other].median()):>13.1f}")

print()
print(f"The rate of the other dimension, and what it would be if only the "
      f"length had moved")
print(f"  {'arm':<10} {'subset':<20} {'n':>5} {'before':>8} {'after':>8} "
      f"{'markers_held':>13}")
print("  " + "-" * 68)
for _label, _subset in (("every changed bio", _mc_w22),
                        (f"gate {PRIMARY_GATE}", _gated_w22)):
    for arm in ARMS:
        _arm_rows = _subset[_subset["arm"] == arm]
        print(f"  {arm:<10} {_label:<20} {len(_arm_rows):>5} "
              f"{_arm_rows['rate_other_before'].mean():>8.3f} "
              f"{_arm_rows['rate_other_after'].mean():>8.3f} "
              f"{_arm_rows['rate_other_denom'].mean():>13.3f}")

print()
print(f"Biographies whose count of markers of the other dimension moved, at "
      f"the gate {PRIMARY_GATE}")
for arm in ARMS:
    _arm_rows = _gated_w22[_gated_w22["arm"] == arm]
    _move = _arm_rows["other_after"] - _arm_rows["other_before"]
    print(f"  {arm:<10} unchanged {int((_move == 0).sum()):>5}, one marker "
          f"fewer {int((_move == -1).sum()):>4}, more than one fewer "
          f"{int((_move < -1).sum()):>4}, more than before "
          f"{int((_move > 0).sum()):>4}")

# How the protocol can move the length at all: a replacement puts a target of
# more than one word in the place of one word, and a deletion takes a marker
# out. The substitution list matches a phrase for some rules, as in
# "a member of" against "part of", so the words of the target do not give the
# change in length on their own; the length itself comes from the word counts
# above.
_log_w22 = pd.read_csv(PROTOCOL_FILE)
_occ_w22 = _log_w22[(_log_w22["sample"] == "pool")
                    & (_log_w22["action_taken"].isin(WORDS_ACTIONS))].merge(
    _gated_w22[["occ", "bio_idx", "arm"]], on=["occ", "bio_idx", "arm"],
    how="inner")
_occ_w22["target_words"] = _occ_w22["target"].fillna("").str.split().str.len()
print()
print(f"The occurrences behind it, at the gate {PRIMARY_GATE}")
print(f"  {'arm':<10} {'replaced':>9} {'target>1 word':>14} "
      f"{'target words 2':>15} {'3':>5} {'deleted':>8} {'words_net':>10}")
print("  " + "-" * 76)
for arm in ARMS:
    _arm_rows = _occ_w22[_occ_w22["arm"] == arm]
    _rep = _arm_rows[_arm_rows["action_taken"] == "replaced"]
    _del = _arm_rows[_arm_rows["action_taken"] == "deleted"]
    _net = int(_gated_w22[_gated_w22["arm"] == arm]["delta_words"].sum())
    print(f"  {arm:<10} {len(_rep):>9} {int((_rep['target_words'] > 1).sum()):>14} "
          f"{int((_rep['target_words'] == 2).sum()):>15} "
          f"{int((_rep['target_words'] == 3).sum()):>5} {len(_del):>8} "
          f"{_net:>+10}")
print(f"  every replaced occurrence is one word in the biography; words_net "
      f"is the sum of delta_words over the biographies of that arm")
```

## 24. The fit by gender under the language model

- **Question:** Table 2 carries the experimental test of H4b for the two
  encoders but not for the language model. What does the same test give for
  Arm 3?
- **Expectation:** none fixed in advance for this screener; the section runs
  the test of notebook 06 section 14.15 on the ranks of notebook 07 and is
  read as a description.
- **Status:** added on 2026-11-12, after the other results; the test of
  notebook 06 section 14.15, not specified in advance for the language model.
  The slope before the edit is controlled against the slope of the original
  biography.
- **Unit and test:** biography; the slope of its percentile rank over the
  twenty advertisements of its occupation against their balance, the signed
  rank test on the change of that slope per gender, and the ordinary least
  squares of the slope on gender with occupation effects and HC1 standard
  errors on the original biographies. The communal arm at the gate of the
  main analysis, as in section 9.
- **Output:** none; the section prints.

```python
# W23: the fit by gender under the language model. Table 2 carries the
# experimental test of H4b for the two encoders but not for Arm 3. The test is
# the one of notebook 06 section 14.15, which section 9 of this notebook runs
# on Arm 2: the slope of a biography is the slope of its percentile rank over
# the advertisements of its occupation against their balance, and the test is
# the change of that slope between the original and the edited biography. The
# ranks are the ones notebook 07 wrote; the percentiles after the edit are the
# ones of the change there, so the biography is ranked against the unchanged
# others. Nothing is rated here.
ARM3_SLOPE_ROUNDS = (("pct_before", "s_before"), ("pct_after", "s_after"))
for _path in (ARM3_ORIGINALS, ARM3_PCT_BY_JD):
    if not _path.exists():
        raise FileNotFoundError(
            f"{_path} not found. This section reads the ranks of the language "
            f"model from the outputs of notebook 07.")

_a3_pairs = pd.read_csv(ARM3_PCT_BY_JD)
_a3_pairs["jd_idx"] = [int(str(_j).rsplit("jd", 1)[1])
                       for _j in _a3_pairs["jd_id"]]
_a3_female = _arm3_ranks[["occ", "bio_idx", "female"]].drop_duplicates()


def _slopes_arm3(frame, column, name):
    """Per biography the slope of its percentile rank over the advertisements
    of its occupation against their balance, formed as in notebook 06 section
    14.15: the balance is centred within the occupation."""
    _rows = []
    for occ in KEPT_13:
        _wide = (frame[frame["occ"] == occ]
                 .pivot_table(index="bio_idx", columns="jd_idx",
                              values=column))
        _balance = (_jd_parts[_jd_parts["occ"] == occ]
                    .set_index("jd_idx")["balance"]
                    .reindex(_wide.columns).to_numpy(float))
        if np.isnan(_balance).any():
            raise AssertionError(
                f"{occ}: an advertisement the language model rated has no "
                f"balance, so no slope can be formed on it.")
        _centred = _balance - _balance.mean()
        _matrix = _wide.to_numpy(float)
        if np.isnan(_matrix).any():
            raise AssertionError(
                f"{occ}: a biography was not ranked for every advertisement, "
                f"so its slope is not the one of section 14.15.")
        _slope = ((_matrix - _matrix.mean(axis=1, keepdims=True)) @ _centred
                  / float(_centred @ _centred))
        for _bio_idx, _value in zip(_wide.index, _slope):
            _rows.append({"occ": occ, "bio_idx": int(_bio_idx),
                          name: float(_value)})
    return pd.DataFrame(_rows)


# Model 1 runs on the original biographies, on every biography the language
# model ranked for all twenty advertisements of its occupation.
_a3_orig_slopes = _slopes_arm3(_arm3_ranks, "pct", "s_bal").merge(
    _a3_female, on=["occ", "bio_idx"], how="inner")
print(f"original biographies with a slope: {len(_a3_orig_slopes)}  women "
      f"{int((_a3_orig_slopes['female'] == 1).sum())}  men "
      f"{int((_a3_orig_slopes['female'] == 0).sum())}  occupations "
      f"{_a3_orig_slopes['occ'].nunique()}")
_m1_arm3 = smf.ols("s_bal ~ female + C(occ)",
                   data=_a3_orig_slopes).fit(cov_type="HC1")
_ci_arm3 = _m1_arm3.conf_int().loc["female"]
print(f"  model 1 (s_bal ~ female + C(occ)): "
      f"b={float(_m1_arm3.params['female']):>+.4f}  "
      f"SE={float(_m1_arm3.bse['female']):.4f}  "
      f"95%CI=[{float(_ci_arm3[0]):>+.4f}, {float(_ci_arm3[1]):>+.4f}]  "
      f"p={float(_m1_arm3.pvalues['female']):.4f}  n={int(_m1_arm3.nobs)}")

# The change of the slope, on the biographies the communal manipulation
# changed, at the gate of the main analysis and among those the language model
# rated.
_a3_changed = _a3_pairs[_a3_pairs["arm"] == "communion"].merge(
    _changed_gate[["occ", "bio_idx"]], on=["occ", "bio_idx"], how="inner")
print()
print(f"changed biographies at the gate {PRIMARY_GATE} the language model "
      f"rated: {_a3_changed[['occ', 'bio_idx']].drop_duplicates().shape[0]} of "
      f"{len(_changed_gate)}, pairs {len(_a3_changed)}")
_a3_slopes = None
for _column, _name in ARM3_SLOPE_ROUNDS:
    _part = _slopes_arm3(_a3_changed, _column, _name)
    _a3_slopes = _part if _a3_slopes is None else _a3_slopes.merge(
        _part, on=["occ", "bio_idx"])
_a3_slopes = _a3_slopes.merge(_a3_female, on=["occ", "bio_idx"], how="inner")
_a3_slopes["delta_s"] = _a3_slopes["s_after"] - _a3_slopes["s_before"]

# Control: the slope before the edit is the slope of the original biography.
_checked = _a3_slopes.merge(_a3_orig_slopes[["occ", "bio_idx", "s_bal"]],
                            on=["occ", "bio_idx"], how="inner")
_worst_slope = float(np.abs(_checked["s_before"] - _checked["s_bal"]).max())
print(f"control the slope before the edit against the slope of the original "
      f"biography: largest deviation {_worst_slope:.3g} (expected below "
      f"1e-09), rows {len(_checked)}")
if _worst_slope > 1e-9 or len(_checked) != len(_a3_slopes):
    raise AssertionError(
        f"The slopes before the edit do not match the slopes of the original "
        f"biographies: deviation {_worst_slope:.3g}, {len(_checked)} of "
        f"{len(_a3_slopes)} rows matched.")

print()
print(f"The slope over the advertisements, communal arm, gate {PRIMARY_GATE}")
print(f"  {'group':<8} {'n':>5} {'median_before':>14} {'median_after':>13} "
      f"{'median_change':>14} {'p_wilcoxon':>11}")
print("  " + "-" * 70)
for _value, _label in ((1, "women"), (0, "men")):
    _side = _a3_slopes[_a3_slopes["female"] == _value]
    _change = _side["delta_s"].dropna()
    if len(_change) < 5 or not (_change != 0).any():
        print(f"  {_label:<8} {len(_change):>5}  test left out, "
              f"{len(_change)} changes")
        continue
    _, _p_arm3 = wilcoxon(_change)
    print(f"  {_label:<8} {len(_change):>5} "
          f"{float(_side['s_before'].median()):>+14.4f} "
          f"{float(_side['s_after'].median()):>+13.4f} "
          f"{float(_change.median()):>+14.4f} {float(_p_arm3):>11.4f}")
print(f"  the two tests are nominal; the language model was not part of the "
      f"test fixed in advance")
```

## 25. One scale: the experimental test of H4a on the balance

- **Question:** Notebook 06 section 14.4 fixes the experimental test of H4a on
  the communal density of the advertisement. What does the same test give on
  the communal balance, the scale on which every other test of Study 1 and
  Study 2 is reported?
- **Expectation:** the one of H4a for the arm that matches the dimension, and
  the one of T1; the section changes no direction and no procedure, only the
  quantity of the advertisement the change is held against.
- **Status:** added on 2026-11-13, the control of T1 added on 2026-11-14;
  one scale: the experimental test of H4a is reported on the communal balance
  like all other tests, and the version on the communal density fixed in
  notebook 06 section 14.4 is kept for Appendix E. T1 of the language model
  is the one row another notebook fixed in advance, and it is held against
  `arm3/main/study2_arm3_t1.csv` in rho and in p.
- **Unit and test:** advertisement; the test of notebook 06 section 14.4
  unchanged, Spearman within the occupation with a permutation test stratified
  by occupation, 5,000 permutations, the seed of the run, two-sided. The
  change per advertisement is the mean over the biographies the arm changed at
  the gate, in percentile points as in Figure 5; T1 is the gap between the
  biographies of women and of men per advertisement, on the original
  biographies and with no edit anywhere. The language model rates on a scale
  of its own, so its change enters in percentile points like the figure, and
  the control of the experiment against notebook 07 is run on the scale that
  notebook used. The occupations enter the permutation in the order of the
  notebook that fixed the test: notebook 06 section 14.4 builds its
  dictionary over the thirteen occupations in the order of the run, notebook
  07 builds its own with a groupby and therefore in alphabetical order. The
  permutation is stratified by occupation, so its draws depend on that order,
  while the rank correlation itself does not.
- **Output:** `study2_scale_table.csv`, Table E7.

```python
# One scale. The experimental test of H4a is fixed in notebook 06 section
# 14.4 on the communal density of the advertisement; the write-up reports it on
# the communal balance, the scale every other test uses, and keeps the version
# on the density for Appendix E. The test itself is the one of section 14.4,
# unchanged: the mean change per advertisement over the biographies the arm
# changed at the gate, against a quantity of the advertisement, Spearman within
# the occupation with a pooled permutation test. T1 of the original
# biographies, the gap between women and men per advertisement, is run against
# the same three quantities. The controls hold the values the earlier sections
# print, which are given to three decimals, so the control is on the third
# decimal and not on 1e-05. T1 of the language model is the one test of this
# section that another notebook fixed in advance, and the control at the end
# holds it against notebook 07 in rho and in p.
SCALE_TABLE = Path(OUT_BASE) / "study2_scale_table.csv"
ARM3_DELTA_BY_JD = ARM3_DIR / "study2_arm3_delta_by_jd.csv"
ARM3_CONDITIONAL = ARM3_DIR / "study2_arm3_conditional.csv"
ARM3_T1 = ARM3_DIR / "study2_arm3_t1.csv"
JD_QUANTITIES = (("communal balance", "balance"),
                 ("communal density", "comm_dens"),
                 ("agentic density", "ag_dens"))
SCALE_TOLERANCE = 5e-4
# The values the earlier sections print, as (screener, row, quantity) -> rho.
# The experiment of Arm 3 is held against notebook 07 separately, because that
# notebook runs the test on the scale of the language model and not in
# percentile points.
CONTROLS_SCALE = {
    ("arm1", "communion", "comm_dens"): -0.115,
    ("arm1", "communion", "balance"): -0.055,
    ("arm2", "communion", "comm_dens"): -0.141,
    ("arm2", "communion", "balance"): -0.113,
    ("arm2", "agency", "ag_dens"): -0.128,
    ("arm1", "T1", "balance"): 0.086,
    ("arm2", "T1", "balance"): 0.100,
    ("arm3", "T1", "balance"): 0.132,
    ("arm1", "T1", "comm_dens"): 0.119,
    ("arm1", "T1", "ag_dens"): 0.030,
}

# The occupations enter the permutation in the order of the notebook that
# fixed the test. Notebook 06 section 14.4 builds its dictionary over KEPT_13;
# notebook 07 builds its own with a groupby and therefore in alphabetical
# order. The permutation is stratified by occupation, so its draws depend on
# that order, and a row reproduces the p value its own notebook published only
# in the order that notebook used. The rank correlation itself does not depend
# on it.
T1_ORDER = {"arm1": KEPT_13, "arm2": KEPT_13, "arm3": sorted(KEPT_13)}

for _path in (ARM3_DELTA_BY_JD, ARM3_CONDITIONAL, ARM3_T1):
    if not _path.exists():
        raise FileNotFoundError(
            f"{_path} not found. The control of Arm 3 reads the change on the "
            f"scale of the language model and the test notebook 07 ran on it.")

_jd_quantities = df_jd_eda[["occ", "jd_idx"]
                           + [_c for _, _c in JD_QUANTITIES]].copy()


def _changed_at_gate(arm):
    """The biographies one arm changed and that pass the gate."""
    return df_pool[(df_pool["sim_cp"] >= PRIMARY_GATE)
                   & (df_pool["n_subs"] >= 1)
                   & (df_pool["arm"] == arm)][["occ", "bio_idx"]]


def _delta_per_ad(screener, arm, column="delta_pct"):
    """The mean change per advertisement over the changed biographies of the
    gate, in percentile points, as section 12 forms it."""
    _name = f"dj_pool_{arm}_rank"
    if screener == "arm1":
        _frame = df_jd_level
    elif screener == "arm2":
        _frame = df_arm2_jd_level
    else:
        _file = (ARM3_PCT_BY_JD if column == "delta_pct"
                 else ARM3_DELTA_BY_JD)
        _pairs = pd.read_csv(_file)
        _pairs = _pairs[_pairs["arm"] == arm].copy()
        _pairs["jd_idx"] = [int(str(_j).rsplit("jd", 1)[1])
                            for _j in _pairs["jd_id"]]
        _pairs = _pairs.merge(_changed_at_gate(arm), on=["occ", "bio_idx"])
        return (_pairs.groupby(["occ", "jd_idx"])[column].mean()
                .reset_index().rename(columns={column: "y"}))
    return (_frame[["occ", "jd_idx", _name]].rename(columns={_name: "y"}))


def _spearman_within(frame, xcolumn, order):
    """The test of notebook 06 section 14.4 on one pair of columns, with the
    occupations entering in the given order."""
    _y, _x = {}, {}
    for occ in order:
        _here = (frame[frame["occ"] == occ].sort_values("jd_idx")
                 .dropna(subset=["y", xcolumn]))
        if _here.empty:
            continue
        _y[occ] = _here["y"].to_numpy(float)
        _x[occ] = _here[xcolumn].to_numpy(float)
    _table, _pool = study2_core.jd_language_test(_y, _x, n_perm=N_PERM,
                                                 seed=SEED)
    if _table.empty:
        raise AssertionError(
            f"No occupation carries enough advertisements for the test on "
            f"{xcolumn}.")
    return _table, _pool


def _rows_for(screener, row_label, per_ad, sign, order):
    """One row per quantity of the advertisement."""
    _frame = per_ad.merge(_jd_quantities, on=["occ", "jd_idx"])
    _out = []
    for _name, _column in JD_QUANTITIES:
        _table, _pool = _spearman_within(_frame, _column, order)
        _counted = int((_table["rho"] < 0).sum() if sign == "-"
                       else (_table["rho"] > 0).sum())
        _out.append({"screener": LABELS[screener], "row": row_label,
                     "quantity": _name, "n_ads": int(_pool["n_jds"]),
                     "rho": float(_pool["rho"]),
                     "p": float(_pool["p_rho"]),
                     "sign_counted": sign,
                     "n_occupations": int(_pool["n_occ"]),
                     "n_occupations_with_sign": _counted})
        _target = CONTROLS_SCALE.get((screener, row_label, _column))
        if _target is not None:
            _deviation = abs(float(_pool["rho"]) - _target)
            if _deviation > SCALE_TOLERANCE:
                raise AssertionError(
                    f"{LABELS[screener]}, {row_label}, {_name}: rho is "
                    f"{float(_pool['rho']):+.5f} and misses the "
                    f"{_target:+.3f} the earlier sections print, by "
                    f"{_deviation:.1e}.")
    return _out

# The experiment, both arms, and T1 on the original biographies.
_scale_rows = []
for _screener, _label in SCREENERS:
    for _arm in ARMS:
        _scale_rows += _rows_for(_screener, _arm,
                                 _delta_per_ad(_screener, _arm), "-",
                                 KEPT_13)
for _screener, _label in SCREENERS:
    _parts = _arm3_parts if _screener == "arm3" else _jd_parts
    _gaps = (_gaps_per_ad(_screener, _parts)[["occ", "jd_idx", "c"]]
             .rename(columns={"c": "y"}))
    _scale_rows += _rows_for(_screener, "T1", _gaps, "+",
                             T1_ORDER[_screener])
df_scale = pd.DataFrame(_scale_rows)

# Control: on the scale of the language model the experiment of Arm 3 has to
# give back the test notebook 07 ran in its section 6.
_saved_a3 = pd.read_csv(ARM3_CONDITIONAL)
_saved_a3 = _saved_a3[_saved_a3["subset"] == "gate"].set_index(["arm", "x"])
_worst_a3 = 0.0
for _arm in ARMS:
    _frame_a3 = (_delta_per_ad("arm3", _arm, column="delta")
                 .merge(_jd_quantities, on=["occ", "jd_idx"]))
    for _name, _column in JD_QUANTITIES:
        _, _pool_a3 = _spearman_within(_frame_a3, _column, KEPT_13)
        _worst_a3 = max(_worst_a3, abs(float(_pool_a3["rho"])
                                       - float(_saved_a3.loc[(_arm, _column),
                                                             "rho"])))
print(f"control the experiment of Arm 3 on the scale of the language model "
      f"against study2_arm3_conditional.csv: largest deviation "
      f"{_worst_a3:.3g} (expected below 1e-09)")
if _worst_a3 > 1e-9:
    raise AssertionError(
        f"The change of Arm 3 per advertisement does not reproduce the test "
        f"of notebook 07; it deviates by {_worst_a3:.3g}.")
print(f"control the ten values the earlier sections print, to the third "
      f"decimal: every one is met")

# Control with a stop: on the communal balance T1 of the language model is the
# test notebook 07 fixed in advance. Section 12 forms the gap per
# advertisement as that notebook forms it, so the two have to agree in rho and
# in p and not only to the third decimal.
_saved_t1 = pd.read_csv(ARM3_T1)
_saved_t1 = _saved_t1[_saved_t1["measure"] == "percentile"]
_row_t1 = df_scale[(df_scale["screener"] == LABELS["arm3"])
                   & (df_scale["row"] == "T1")
                   & (df_scale["quantity"] == "communal balance")]
if len(_saved_t1) != 1 or len(_row_t1) != 1:
    raise AssertionError(
        f"The control of T1 needs one row on each side; "
        f"{ARM3_T1.name} holds {len(_saved_t1)} for the percentile rank and "
        f"this section {len(_row_t1)} for the language model.")
_saved_t1, _row_t1 = _saved_t1.iloc[0], _row_t1.iloc[0]
_dev_rho_t1 = abs(float(_row_t1["rho"]) - float(_saved_t1["rho_within_occ"]))
_dev_p_t1 = abs(float(_row_t1["p"]) - float(_saved_t1["p_within_occ"]))
print(f"control T1 of the language model against {ARM3_T1.name}: rho "
      f"{float(_row_t1['rho']):+.6f} against "
      f"{float(_saved_t1['rho_within_occ']):+.6f}, p "
      f"{float(_row_t1['p']):.6f} against "
      f"{float(_saved_t1['p_within_occ']):.6f}, largest deviation "
      f"{max(_dev_rho_t1, _dev_p_t1):.3g} (expected below 1e-06)")
if max(_dev_rho_t1, _dev_p_t1) > 1e-6:
    raise AssertionError(
        f"T1 of the language model is not the test notebook 07 fixed in "
        f"advance: rho deviates by {_dev_rho_t1:.3g} and p by "
        f"{_dev_p_t1:.3g}.")

print()
print(f"The experiment and T1 on the three quantities of the advertisement, "
      f"gate {PRIMARY_GATE}")
print(f"  {'screener':<16} {'arm or test':<11} {'quantity':<17} {'n_ads':>6} "
      f"{'rho':>7} {'p':>7} {'sign':>5} {'occupations':>12}")
print("  " + "-" * 88)
for _, _row in df_scale.iterrows():
    _counted = (f"{_row['n_occupations_with_sign']} of "
                f"{_row['n_occupations']}")
    print(f"  {_row['screener']:<16} {_row['row']:<11} "
          f"{_row['quantity']:<17} {_row['n_ads']:>6} "
          f"{_row['rho']:>+7.3f} {_row['p']:>7.4f} "
          f"{_row['sign_counted']:>5} {_counted:>12}")

df_scale.to_csv(SCALE_TABLE, index=False, encoding="utf-8")
print()
print(f"written: {SCALE_TABLE} ({len(df_scale)} rows)")
```

## 26. Does the communal difference rest on markers the edit cannot reach?

- **Question:** The biographies of women carry more communal wording than
  those of men. How much of that difference sits in words the substitution
  replaces or deletes, and how much in proper names, fixed terms and words
  for which the list holds no rule?
- **Expectation:** none fixed in advance; the section splits a quantity of
  Study 1 into the parts of the protocol and fixes no direction of its own.
- **Status:** added on 2026-11-14, descriptive; whether the difference in communal words
  rests on markers the edit cannot reach. The parts are controlled against
  the counter of dictionary A, which they cannot exceed.
- **Unit and test:** biography; every biography of the pool enters, also the
  ones the manipulation did not change. The parts are rates per 100 words and
  add up to the rate of the whole biography; the difference between women and
  men is read as in Study 1, an OLS of the rate on female with occupation
  effects and HC1 standard errors, with the interval of that coefficient.
- **Output:** none; the section prints.

```python
# Does the communal difference between the biographies of women and of men
# rest on markers the edit cannot reach? Every biography of the pool enters,
# not only the ones the manipulation changed. The protocol of the communal arm
# holds one row per communal marker the engine saw: a row counts as treated
# when the edit replaced or deleted the word, and otherwise under the reason
# the protocol gives. A biography without a communal marker carries a zero in
# every part. The parts are rates per 100 words, so they add up to the rate of
# the whole biography, and the difference between women and men is read as in
# Study 1: an OLS of the rate on female with occupation effects and HC1
# standard errors.
ACTIONS_W35 = ("replaced", "deleted")
PARTS_W35 = (("treated", "replaced or deleted"),
             ("proper_name", "untreated: proper names"),
             ("fixed_term", "untreated: fixed terms"),
             ("no_rule", "untreated: no rule for the word"),
             ("rest", "untreated: rest"))
# The protocol names the position of a word as a second reason for having no
# rule, so both reasons are counted as one part.
NO_RULE_REASONS = ("no_rule_for_word", "no_rule_for_position")

_log_w35 = pd.read_csv(PROTOCOL_FILE)
_log_w35 = _log_w35[(_log_w35["sample"] == "pool")
                    & (_log_w35["arm"] == "communion")].copy()


def _part_of_w35(action, reason):
    """The part of the communal markers of a biography that one row of the
    protocol counts towards."""
    if action in ACTIONS_W35:
        return "treated"
    if reason in ("proper_name", "fixed_term"):
        return reason
    if reason in NO_RULE_REASONS:
        return "no_rule"
    return "rest"


_log_w35["part"] = [_part_of_w35(_a, _r) for _a, _r
                    in zip(_log_w35["action_taken"],
                           _log_w35["reason_untreated"])]
_counts_w35 = (_log_w35.groupby(["occ", "bio_idx", "part"]).size()
               .unstack("part"))
_frame_w35 = df_bio[["occ", "bio_idx", "female", "word_count",
                     "communion_rate"]].copy()
_frame_w35 = _frame_w35.merge(_counts_w35, on=["occ", "bio_idx"], how="left")
for _part, _ in PARTS_W35:
    if _part not in _frame_w35.columns:
        _frame_w35[_part] = 0.0
_frame_w35 = _frame_w35.fillna(0.0)
_frame_w35["untreated"] = _frame_w35[[_p for _p, _ in PARTS_W35
                                      if _p != "treated"]].sum(axis=1)
_frame_w35["total"] = _frame_w35["treated"] + _frame_w35["untreated"]
for _column in (("total", "treated", "untreated")
                + tuple(_p for _p, _ in PARTS_W35 if _p != "treated")):
    _frame_w35[f"rate_{_column}"] = (100.0 * _frame_w35[_column]
                                     / _frame_w35["word_count"])
_frame_w35["rate_counter"] = _frame_w35["communion_rate"]
print(f"biographies {len(_frame_w35)}, of them with at least one communal "
      f"marker {int((_frame_w35['total'] > 0).sum())}; communal markers in "
      f"all {int(_frame_w35['total'].sum())}, of them replaced or deleted "
      f"{int(_frame_w35['treated'].sum())}")

# Control: the counter of dictionary A counts the markers of the same
# biographies. Section 20 shows that the protocol writes one row for a span
# that covers more than one word, so it can hold fewer rows than the counter
# finds markers, never more.
_over_w35 = int((_frame_w35["rate_total"]
                 > _frame_w35["rate_counter"] + 1e-9).sum())
_differ_w35 = int((_frame_w35["rate_total"] - _frame_w35["rate_counter"])
                  .abs().gt(1e-9).sum())
print(f"control the protocol against the counter of dictionary {LEX_MAIN}: "
      f"biographies with a lower rate in the protocol {_differ_w35}, with a "
      f"higher one {_over_w35} (expected 0)")
if _over_w35:
    raise AssertionError(
        f"The protocol holds more communal markers than the counter finds in "
        f"{_over_w35} biographies, so the parts do not sit inside the whole.")

print()
print(f"The difference between the biographies of women and of men per 100 "
      f"words, with occupation effects and HC1, n = {len(_frame_w35)}")
print(f"  {'quantity':<32} {'women':>7} {'men':>7} {'b_female':>9} "
      f"{'ci_lo':>8} {'ci_hi':>8} {'p':>8} {'share':>7}")
print("  " + "-" * 98)
_b_whole_w35 = None
for _column, _label in ((("total", "all communal markers"),
                         ("treated", "replaced or deleted"),
                         ("untreated", "untreated, together"))
                        + tuple((_p, _l) for _p, _l in PARTS_W35
                                if _p != "treated")
                        + (("counter", f"counter of dictionary {LEX_MAIN}"),)):
    _y = f"rate_{_column}"
    _fit = smf.ols(f"{_y} ~ female + C(occ)",
                   data=_frame_w35).fit(cov_type="HC1")
    _b = float(_fit.params["female"])
    _lo, _hi = [float(_v) for _v in _fit.conf_int().loc["female"]]
    if _column == "total":
        _b_whole_w35 = _b
    print(f"  {_label:<32} "
          f"{_frame_w35[_frame_w35['female'] == 1][_y].mean():>7.3f} "
          f"{_frame_w35[_frame_w35['female'] == 0][_y].mean():>7.3f} "
          f"{_b:>+9.4f} {_lo:>+8.4f} {_hi:>+8.4f} "
          f"{float(_fit.pvalues['female']):>8.4f} "
          f"{_b / _b_whole_w35:>7.3f}")
print(f"  the share is the coefficient divided by the one of all communal "
      f"markers; the parts add up to the whole")
```

## 27. Why the mean change lies further from zero than the median

- **Question:** Table 2 reports the median change for H3, panel (b) of
  Figure 5 draws means, and the mean lies about twice as far from zero as the
  median in all three screeners. Where does that distance come from?
- **Expectation:** none fixed in advance; the section describes a
  distribution and fixes no direction of its own.
- **Status:** added on 2026-11-14, descriptive; why the mean change lies
  further from zero than the median.
- **Unit and test:** biography. The change of a biography is the one of H3:
  the mean over all advertisements of its occupation of the change in its
  percentile rank. Quantiles, shares, means and medians, and an OLS of that
  change on what the biography carries, with occupation effects and HC1; it
  describes and tests none of the hypotheses. The biographies are the ones
  Table 2 counts, and the section reproduces its three medians before it
  starts. Part D forms panel (b) of Figure 5 a second time, with the median
  biography of a fifth in place of the mean advertisement, and its first
  version has to reproduce the line the figure draws.
- **Output:** none; the section prints.

```python
# Why does the mean change lie further from zero than the median? For H3 the
# change of a biography is the mean over all advertisements of its occupation
# of the change in its percentile rank, which is what
# study2_core.individual_cf_ranks minus study2_core.percentile_ranks gives,
# and Table 2 reports its median over the biographies the communal
# manipulation changed that pass the gate. The section takes the distance
# between the mean and that median apart: the shape of the distribution (A),
# what the biographies at the strong end carry (B), whether the distance
# already sits inside an occupation (C), and what panel (b) of Figure 5 looks
# like when the median biography carries it instead of the mean
# advertisement (D).
QUANTILES_W37 = (0.05, 0.10, 0.25, 0.50, 0.75, 0.90, 0.95)
TRIM_W37 = 0.05           # the share cut off at each end of the trimmed mean
TENTH_W37 = 0.10          # the strong tenth of part B
PCT_MIDDLE = (25.0, 75.0)  # the middle of the ranking in part B
# The strong end is the one the median points to: upwards under SBERT,
# downwards under the other two.
STRONG_SIDE = {"arm1": "high", "arm2": "low", "arm3": "low"}


def _matrices_w37(screener, occ):
    """Per occupation the percentile rank of every biography before the edit
    and the change of that rank, one row per biography and one column per
    advertisement. Nothing is embedded; Arm 1 reads the cache of notebook 06,
    Arm 2 the embeddings of notebook 06a and Arm 3 the table of notebook 07."""
    if screener == "arm3":
        _pairs = pd.read_csv(ARM3_PCT_BY_JD)
        _pairs = _pairs[(_pairs["arm"] == "communion")
                        & (_pairs["occ"] == occ)].copy()
        _pairs["jd_idx"] = [int(str(_j).rsplit("jd", 1)[1])
                            for _j in _pairs["jd_id"]]
        _before = _pairs.pivot(index="bio_idx", columns="jd_idx",
                               values="pct_before")
        _change = _pairs.pivot(index="bio_idx", columns="jd_idx",
                               values="delta_pct")
        if _before.isna().to_numpy().any() or _change.isna().to_numpy().any():
            raise AssertionError(
                f"{occ}: a biography of Arm 3 was not rated against every "
                f"advertisement, so its change is not the one of Table 2.")
        return (_before.index.to_numpy(int), _before.columns.to_numpy(int),
                _before.to_numpy(float), _change.to_numpy(float))
    if screener == "arm1":
        _jd_cache = np.load(ARM1_CACHE / f"{occ}_jds.npy",
                            allow_pickle=True).item()
        _orig = np.load(ARM1_CACHE / f"{occ}_pool_orig.npy",
                        allow_pickle=True).item()["mean"]
        _manip = np.load(ARM1_CACHE / ARM1_MANIP.format(occ=occ),
                         allow_pickle=True).item()["mean"]
        _so = study2_core.jd_similarity(_orig, _jd_cache, how=JD_REPR)
        _sm = study2_core.jd_similarity(_manip, _jd_cache, how=JD_REPR)
    else:
        _so = emb_orig[occ] @ emb_jd[occ].T
        _sm = emb_manip[occ]["communion"] @ emb_jd[occ].T
    _before = study2_core.pool_pct_per_jd(_so)
    _change = study2_core.indiv_cf_pct_per_jd(_so, _sm) - _before
    return (np.arange(_before.shape[0]), np.arange(_before.shape[1]),
            _before, _change)


# The biographies Table 2 counts: changed by the communal manipulation and at
# the gate. Arm 3 rated a sample of them, so its set is the one the table of
# notebook 07 carries with a percentile change.
_gated_w37 = {}
_a3_gate_w37 = pd.read_csv(ARM3_DELTA_MAIN)
_a3_gate_w37 = _a3_gate_w37[(_a3_gate_w37["arm"] == "communion")
                            & (_a3_gate_w37["sim_cp"] >= PRIMARY_GATE)
                            & (_a3_gate_w37["delta_pct"].notna())]
for _screener, _label in SCREENERS:
    _gated_w37[_screener] = (
        _a3_gate_w37[["occ", "bio_idx"]] if _screener == "arm3"
        else _changed_gate[["occ", "bio_idx"]])

# The size of the ranking each screener places a biography in: 250 for the two
# embedding models, the rated comparison set for the language model. One rank
# place is 100 divided by that number less one.
_pool_size_w37 = {}
for occ in KEPT_13:
    _pool_size_w37[("arm3", occ)] = int(
        _arm3_ranks[_arm3_ranks["occ"] == occ]["bio_idx"].nunique())
    for _screener in ("arm1", "arm2"):
        _pool_size_w37[(_screener, occ)] = N_BIOS_POOL

_bio_w37, _fifth_w37 = {}, {}
for _screener, _label in SCREENERS:
    _rows, _fifth_rows = [], []
    _parts = _arm3_parts if _screener == "arm3" else _jd_parts
    for occ in KEPT_13:
        _bios, _ads, _before, _change = _matrices_w37(_screener, occ)
        _fifth_of_ad = (_parts[_parts["occ"] == occ]
                        .set_index("jd_idx")["jd_fifth"])
        _keep_ad = [_k for _k, _j in enumerate(_ads)
                    if _j in _fifth_of_ad.index]
        _place = 100.0 / (_pool_size_w37[(_screener, occ)] - 1)
        for _i, _bio_idx in enumerate(_bios):
            _rows.append({"occ": occ, "bio_idx": int(_bio_idx),
                          "delta": float(_change[_i].mean()),
                          "pct_before": float(_before[_i].mean()),
                          "rank_place": _place})
        for _part in range(1, N_PARTS + 1):
            _columns = [_k for _k in _keep_ad
                        if int(_fifth_of_ad.loc[_ads[_k]]) == _part]
            if not _columns:
                raise AssertionError(
                    f"{occ}: no advertisement of {_label} sits in fifth "
                    f"{_part}, so the fifth cannot be formed.")
            for _i, _bio_idx in enumerate(_bios):
                _fifth_rows.append(
                    {"occ": occ, "bio_idx": int(_bio_idx), "jd_fifth": _part,
                     "delta": float(_change[_i, _columns].mean())})
    _bio_w37[_screener] = (pd.DataFrame(_rows)
                           .merge(_gated_w37[_screener],
                                  on=["occ", "bio_idx"], how="inner"))
    _fifth_w37[_screener] = (pd.DataFrame(_fifth_rows)
                             .merge(_gated_w37[_screener],
                                    on=["occ", "bio_idx"], how="inner"))

# Control: these are the biographies and the medians of Table 2.
print(f"Biographies behind the median of Table 2, communal arm, gate "
      f"{PRIMARY_GATE}")
print(f"  {'screener':<16} {'n_bios':>7} {'median':>8} {'Table 2':>8}")
print("  " + "-" * 44)
for _screener, _label in SCREENERS:
    _delta = _bio_w37[_screener]["delta"]
    print(f"  {_label:<16} {len(_delta):>7} {float(_delta.median()):>+8.3f} "
          f"{MEDIANS_TABLE2[_screener]:>+8.3f}")
    if abs(float(_delta.median()) - MEDIANS_TABLE2[_screener]) > 5e-4:
        raise AssertionError(
            f"The median of {_label} is {float(_delta.median()):+.4f} and "
            f"does not meet the {MEDIANS_TABLE2[_screener]:+.3f} of Table 2, "
            f"so this section does not rest on the same biographies.")

# A. The shape of the distribution.
print()
print(f"A. The change per biography: where the mean and the median sit")
print(f"  {'screener':<16} {'n':>6} {'median':>8} {'mean':>8} "
      f"{'mean of occ means':>18} {'mean of occ medians':>20} "
      f"{'trimmed mean':>13}")
print("  " + "-" * 94)
for _screener, _label in SCREENERS:
    _frame = _bio_w37[_screener]
    _by_occ = _frame.groupby("occ")["delta"]
    _sorted = np.sort(_frame["delta"].to_numpy(float))
    _cut = int(round(TRIM_W37 * len(_sorted)))
    _trimmed = float(_sorted[_cut:len(_sorted) - _cut].mean())
    print(f"  {_label:<16} {len(_frame):>6} "
          f"{float(_frame['delta'].median()):>+8.3f} "
          f"{float(_frame['delta'].mean()):>+8.3f} "
          f"{float(_by_occ.mean().mean()):>+18.3f} "
          f"{float(_by_occ.median().mean()):>+20.3f} {_trimmed:>+13.3f}")

print()
print(f"  {'screener':<16} " + " ".join(f"{'q' + str(int(_q * 100)):>8}"
                                        for _q in QUANTILES_W37))
print("  " + "-" * 76)
for _screener, _label in SCREENERS:
    _delta = _bio_w37[_screener]["delta"]
    print(f"  {_label:<16} "
          + " ".join(f"{float(_delta.quantile(_q)):>+8.3f}"
                     for _q in QUANTILES_W37))

print()
print(f"  {'screener':<16} {'share > 0':>10} {'share < 0':>10} "
      f"{'share = 0':>10} {'share below one rank place':>27}")
print("  " + "-" * 78)
for _screener, _label in SCREENERS:
    _frame = _bio_w37[_screener]
    _small = float((_frame["delta"].abs()
                    < _frame["rank_place"]).mean())
    print(f"  {_label:<16} {float((_frame['delta'] > 0).mean()):>10.3f} "
          f"{float((_frame['delta'] < 0).mean()):>10.3f} "
          f"{float((_frame['delta'] == 0).mean()):>10.3f} {_small:>27.3f}")
_a3_sizes = [_pool_size_w37[("arm3", occ)] for occ in KEPT_13]
print(f"  one rank place is {100.0 / (N_BIOS_POOL - 1):.3f} points in the two "
      f"embedding models, and between {100.0 / (max(_a3_sizes) - 1):.3f} and "
      f"{100.0 / (min(_a3_sizes) - 1):.3f} points under the language model, "
      f"which ranks a biography among {min(_a3_sizes)} to {max(_a3_sizes)}")

# B. Who moves far?
# The columns of part B, with the heading each one carries.
_STRONG_COLUMNS = (("n_subs", "markers"), ("word_count", "words"),
                   ("communion_rate", "rate_bef"),
                   ("rate_after", "rate_aft"), ("pct_before", "pct_bef"),
                   ("in_middle", "mid_25_75"), ("female", "women"),
                   ("sim_cp", "cosine"))
_check_w37 = df_manip_check[(df_manip_check["sample"] == "pool")
                            & (df_manip_check["arm"] == "communion")]
_traits_w37 = (df_pool[df_pool["arm"] == "communion"]
               [["occ", "bio_idx", "female", "n_subs", "sim_cp"]]
               .merge(df_bio[["occ", "bio_idx", "word_count",
                              "communion_rate"]], on=["occ", "bio_idx"])
               .merge(_check_w37[["occ", "bio_idx", "rate_after"]],
                      on=["occ", "bio_idx"]))
print()
print(f"B. The strong tenth of the biographies against the rest")
print(f"  {'screener':<16} {'group':<14} {'n':>5} "
      + " ".join(f"{_name:>10}" for _, _name in _STRONG_COLUMNS))
print("  " + "-" * 126)
for _screener, _label in SCREENERS:
    _frame = _bio_w37[_screener].merge(_traits_w37, on=["occ", "bio_idx"])
    _frame["in_middle"] = ((_frame["pct_before"] >= PCT_MIDDLE[0])
                           & (_frame["pct_before"] <= PCT_MIDDLE[1]))
    if STRONG_SIDE[_screener] == "high":
        _limit = float(_frame["delta"].quantile(1 - TENTH_W37))
        _strong = _frame["delta"] >= _limit
    else:
        _limit = float(_frame["delta"].quantile(TENTH_W37))
        _strong = _frame["delta"] <= _limit
    for _name, _mask in (("strong tenth", _strong), ("rest", ~_strong)):
        _group = _frame[_mask]
        print(f"  {_label:<16} {_name:<14} {len(_group):>5} "
              + " ".join(f"{float(_group[_column].mean()):>10.3f}"
                         for _column, _ in _STRONG_COLUMNS))
    _share_strong = (_frame[_strong].groupby("occ").size()
                     / int(_strong.sum()))
    _share_all = _frame.groupby("occ").size() / len(_frame)
    _top = _share_strong.sort_values(ascending=False).head(3)
    print(f"  {_label:<16} the tenth runs to {_limit:+.3f}; the three "
          f"occupations with the largest share of it: "
          + ", ".join(f"{_occ} {float(_value):.3f} (all bios "
                      f"{float(_share_all.loc[_occ]):.3f})"
                      for _occ, _value in _top.items()))

print()
print(f"B. The change on the biography, with occupation effects and HC1")
print(f"  {'screener':<16} {'term':<26} {'b':>10} {'ci_lo':>10} "
      f"{'ci_hi':>10} {'p':>9}")
print("  " + "-" * 86)
_TERMS_W37 = (("n_subs", "replaced markers"),
              ("log_words", "log words before"),
              ("pct_before", "percentile rank before"),
              ("pct_before_sq", "percentile rank squared"))
for _screener, _label in SCREENERS:
    _frame = _bio_w37[_screener].merge(_traits_w37, on=["occ", "bio_idx"])
    _frame["log_words"] = np.log(_frame["word_count"])
    _frame["pct_before_sq"] = _frame["pct_before"] ** 2
    _fit = smf.ols("delta ~ n_subs + log_words + pct_before + pct_before_sq"
                   " + C(occ)", data=_frame).fit(cov_type="HC1")
    for _term, _name in _TERMS_W37:
        _lo, _hi = [float(_v) for _v in _fit.conf_int().loc[_term]]
        print(f"  {_label:<16} {_name:<26} "
              f"{float(_fit.params[_term]):>+10.4f} {_lo:>+10.4f} "
              f"{_hi:>+10.4f} {float(_fit.pvalues[_term]):>9.4f}")

# C. Inside an occupation or only over the occupations?
print()
print(f"C. The change per occupation: the median and the mean side by side")
print(f"  {'occ':<20} " + " ".join(f"{_label:>24}" for _, _label in SCREENERS))
print(f"  {'':<20} " + " ".join(f"{'n':>6} {'median':>8} {'mean':>8}"
                                for _ in SCREENERS))
print("  " + "-" * 96)
for occ in KEPT_13:
    _parts_line = []
    for _screener, _label in SCREENERS:
        _here = _bio_w37[_screener]
        _here = _here[_here["occ"] == occ]["delta"]
        _parts_line.append(f"{len(_here):>6} {float(_here.median()):>+8.3f} "
                           f"{float(_here.mean()):>+8.3f}")
    print(f"  {occ:<20} " + " ".join(_parts_line))
for _screener, _label in SCREENERS:
    _stats = (_bio_w37[_screener].groupby("occ")["delta"]
              .agg(["median", "mean"]))
    _pooled_up = float(_bio_w37[_screener]["delta"].median()) > 0
    _same = int(((_stats["median"] > 0) == (_stats["mean"] > 0)).sum())
    _with_pool = int(((_stats["median"] > 0) == _pooled_up).sum())
    print(f"  {_label:<16} occupations whose median and mean carry the same "
          f"sign: {_same} of {len(_stats)}; whose median carries the sign of "
          f"the pooled median: {_with_pool} of {len(_stats)}")

# D. Two versions of panel (b) of Figure 5.
print()
print(f"D. Panel (b) of Figure 5 in two versions, by fifth of the "
      f"advertisements")
print(f"  {'screener':<16} {'version':<22} "
      + " ".join(f"{'fifth ' + str(_k):>9}" for _k in range(1, N_PARTS + 1))
      + f" {'5 minus 1':>10}")
print("  " + "-" * 112)
for _screener, _label in SCREENERS:
    _frame = _fifth_w37[_screener]
    # (i) the mean advertisement: the mean change per advertisement over the
    # changed biographies, averaged within the occupation and then over the
    # thirteen. It has to reproduce the line the figure draws.
    _mean_version = []
    _median_version = []
    for _part in range(1, N_PARTS + 1):
        _here = _frame[_frame["jd_fifth"] == _part]
        _mean_version.append(
            float(_here.groupby("occ")["delta"].mean().mean()))
        _median_version.append(
            float(_here.groupby("occ")["delta"].median().mean()))
    _saved = _fifth_values[_screener][0]
    _worst = float(np.abs(np.asarray(_mean_version) - _saved).max())
    if _worst > 1e-9:
        raise AssertionError(
            f"Version (i) of {_label} does not reproduce panel (b) of "
            f"Figure 5; it deviates by {_worst:.3g}.")
    for _name, _values in (("(i) mean of the ads", _mean_version),
                           ("(ii) median biography", _median_version)):
        print(f"  {_label:<16} {_name:<22} "
              + " ".join(f"{_v:>+9.3f}" for _v in _values)
              + f" {_values[-1] - _values[0]:>+10.3f}")
print(f"  version (i) reproduces the line panel (b) draws, to below 1e-09")
```

## 28. The advertisements by occupation and by seniority level

- **Question:** How do the advertisements of an occupation fall on the two
  sides of the balance, and does the wording follow the seniority of the post
  that is offered?
- **Expectation:** none fixed in advance. Damelang, Rueckel and Stops (2024)
  find more agentic wording in leading positions in German advertisements,
  which Chapter 6 takes up; the section describes and fixes no direction of
  its own.
- **Status:** added on 2026-11-14, descriptive; ads by occupation and
  seniority, for Section 3.3 and Chapter 6. Notebook 06 is not run again: the
  level is read from the file the advertisements come from, with the draw of
  section 17.2 repeated and controlled against the token counts of
  `study2_jd_eda.csv`.
- **Unit and test:** advertisement; counts, shares and means, an OLS of the
  wording on the level with occupation effects and HC1, and Spearman within
  the occupations with the permutation test of notebook 06 section 14.4,
  5,000 permutations stratified by occupation. The level enters as one number,
  from 1 for an internship to 6 for an executive post; the advertisements
  without a level are counted apart and enter neither the model nor the
  correlation.
- **Output:** none; the section prints.

```python
# The advertisements by occupation and by seniority level. Section 3.3 names
# the spread of the share with an agentic surplus over the occupations, and
# Chapter 6 holds the comparison with the German advertisements of Damelang,
# Rueckel and Stops (2024), who find more agentic wording in leading
# positions. The level sits in the column formatted_experience_level of the
# file the advertisements come from, which notebook 06 does not load, and
# section 17.2 of this notebook reads the file for the texts alone. The draw
# of notebook 06 is therefore repeated here once more, with the keywords, the
# seed and the cap of section 17.2, and the level is taken from the rows that
# draw selects. Notebook 06 is not run again.
LEVELS_W38 = (("Internship", 1), ("Entry level", 2), ("Associate", 3),
              ("Mid-Senior level", 4), ("Director", 5), ("Executive", 6))
LEVEL_COLUMN = "formatted_experience_level"
QUANTITIES_W38 = (("balance", "communal balance"),
                  ("comm_dens", "communal density"),
                  ("ag_dens", "agentic density"))

_raw_w38 = pd.read_csv(JD_RAW_PATH,
                       usecols=["title", "description", LEVEL_COLUMN],
                       low_memory=False)
_title_w38 = _raw_w38["title"].fillna("").str.lower()
_raw_w38["occupation"] = None
for _occ, _patterns in OCC_KEYWORDS.items():
    _raw_w38.loc[_title_w38.str.contains("|".join(_patterns), regex=True,
                                         na=False), "occupation"] = _occ
_matched_w38 = _raw_w38[_raw_w38["occupation"].notna()]
_rows_w38 = []
for occ in KEPT_13:
    # The draw of section 17.2: the column of the text alone decides which
    # rows are dropped, so the level cannot change which advertisements the
    # draw returns. It is read from the rows the draw keeps.
    _sub = _matched_w38[_matched_w38["occupation"] == occ][["description"]]
    _sub = _sub.dropna()
    _drawn = _sub.sample(min(N_JDS_FULL, len(_sub)),
                         random_state=RANDOM_STATE)
    for _jd_idx, (_index, _row) in enumerate(_drawn.iterrows()):
        _rows_w38.append({
            "occ": occ, "jd_idx": _jd_idx,
            "n_words_here": max(len(str(_row["description"]).split()), 1),
            "level": _matched_w38.at[_index, LEVEL_COLUMN]})
df_levels = pd.DataFrame(_rows_w38)

# Control: the advertisements drawn here are the ones notebook 06 counted, so
# every token count has to be the one of study2_jd_eda.csv.
df_levels = df_levels.merge(
    df_jd_eda[["occ", "jd_idx", "n_words", "balance", "comm_dens", "ag_dens"]],
    on=["occ", "jd_idx"], how="outer")
_bad_w38 = int((df_levels["n_words_here"] != df_levels["n_words"]).sum())
print(f"advertisements drawn: {len(df_levels)}, token counts that differ "
      f"from study2_jd_eda.csv: {_bad_w38} (expected 0)")
if _bad_w38 or len(df_levels) != len(df_jd_eda):
    raise AssertionError(
        f"{_bad_w38} advertisements differ from the ones notebook 06 counted "
        f"and the draw holds {len(df_levels)} of {len(df_jd_eda)}, so the "
        f"level cannot be attached to them.")
del _raw_w38, _matched_w38

print()
print(f"Advertisements per occupation and the side their balance falls on")
print(f"  {'occ':<20} {'n_ads':>6} {'agentic':>9} {'equal':>9} "
      f"{'communal':>9}")
print("  " + "-" * 58)
for occ in KEPT_13 + ["all occupations"]:
    _here = (df_levels if occ == "all occupations"
             else df_levels[df_levels["occ"] == occ])
    print(f"  {occ:<20} {len(_here):>6} "
          f"{float((_here['balance'] < 0).mean()):>9.3f} "
          f"{float((_here['balance'] == 0).mean()):>9.3f} "
          f"{float((_here['balance'] > 0).mean()):>9.3f}")

print()
print(f"Advertisements per seniority level")
print(f"  {'level':<18} {'n_ads':>6} {'n_occ':>6} {'balance':>9} "
      f"{'agentic':>9} {'comm_dens':>10} {'ag_dens':>9}")
print("  " + "-" * 72)
for _level, _rank in LEVELS_W38:
    _here = df_levels[df_levels["level"] == _level]
    if _here.empty:
        raise AssertionError(
            f"No advertisement carries the level {_level}; the six levels of "
            f"the file are the ones the table names.")
    print(f"  {_level:<18} {len(_here):>6} {_here['occ'].nunique():>6} "
          f"{float(_here['balance'].mean()):>+9.3f} "
          f"{float((_here['balance'] < 0).mean()):>9.3f} "
          f"{float(_here['comm_dens'].mean()):>10.3f} "
          f"{float(_here['ag_dens'].mean()):>9.3f}")
_not_stated = df_levels[df_levels["level"].isna()]
print(f"  advertisements without a level: {len(_not_stated)}, mean balance "
      f"{float(_not_stated['balance'].mean()):+.3f}; they carry no rank and "
      f"stay out of the model below")

# The level inside the occupation. The rank runs from 1 for an internship to
# 6 for an executive post and enters as one number, so the coefficient is the
# step from one level to the next.
_ranked_w38 = df_levels.copy()
_ranked_w38["level_rank"] = _ranked_w38["level"].map(dict(LEVELS_W38))
_ranked_w38 = _ranked_w38[_ranked_w38["level_rank"].notna()]
print()
print(f"The step from one level to the next, with occupation effects and "
      f"HC1, n = {len(_ranked_w38)}")
print(f"  {'quantity':<20} {'b_per_step':>11} {'ci_lo':>9} {'ci_hi':>9} "
      f"{'p':>9}")
print("  " + "-" * 62)
for _column, _name in QUANTITIES_W38:
    _fit = smf.ols(f"{_column} ~ level_rank + C(occ)",
                   data=_ranked_w38).fit(cov_type="HC1")
    _lo, _hi = [float(_v) for _v in _fit.conf_int().loc["level_rank"]]
    print(f"  {_name:<20} {float(_fit.params['level_rank']):>+11.4f} "
          f"{_lo:>+9.4f} {_hi:>+9.4f} "
          f"{float(_fit.pvalues['level_rank']):>9.4f}")

# The same question without a linear form: Spearman of the level with the
# balance within the occupations, with the permutation test of notebook 06
# section 14.4, stratified by occupation.
_level_by_occ, _balance_by_occ = {}, {}
for occ in KEPT_13:
    _here = _ranked_w38[_ranked_w38["occ"] == occ]
    _level_by_occ[occ] = _here["level_rank"].to_numpy(float)
    _balance_by_occ[occ] = _here["balance"].to_numpy(float)
_table_w38, _pool_w38 = study2_core.jd_language_test(
    _balance_by_occ, _level_by_occ, n_perm=N_PERM, seed=SEED)
print(f"  Spearman of the level with the balance within the occupations: "
      f"rho {float(_pool_w38['rho']):+.3f}, p {float(_pool_w38['p_rho']):.4f} "
      f"from {N_PERM} permutations, advertisements "
      f"{int(_pool_w38['n_jds'])} in {int(_pool_w38['n_occ'])} occupations")
print(f"  occupations with a negative rho: "
      f"{int((_table_w38['rho'] < 0).sum())} of {len(_table_w38)}, "
      f"span {float(_table_w38['rho'].min()):+.3f} to "
      f"{float(_table_w38['rho'].max()):+.3f}")
```

## 29. The advertisements, the share of women and seniority by occupation

- **Question:** Does the wording of the advertisements of an occupation follow
  the share of women in that occupation, and how does the seniority of the
  posts that are offered sit next to the two?
- **Expectation:** none fixed in advance. Section 3.3 reads the wording of an
  occupation next to the share of women in it, and section 28 finds the
  wording turning less communal as the level rises within an occupation; the
  section describes and fixes no direction of its own.
- **Status:** added on 2026-11-15, descriptive; ad wording, share of women and
  seniority by occupation, for Section 3.3. The share of women among those
  employed is read from the file Figure A1 and Table A1 read, the share of
  women in the biographies from the pool of Study 2, and the wording and the
  level from section 28.
- **Unit and test:** occupation, 13 of them. Counts, shares and means per
  occupation, and over the 13 occupations Spearman with a two-sided
  permutation p from 10,000 permutations with a fixed seed, and the first
  pair as a product-moment correlation as well. None of this tests a
  hypothesis, and every correlation rests on 13 points.
- **Output:** `outputs/tables/tableA2_ads_by_occupation.csv`, Table A2
  of the appendix, which holds the columns of part A.

```python
# The wording of the advertisements, the share of women and the seniority of
# the posts, per occupation. Section 3.3 reads the wording of an occupation
# next to the share of women in it, and the two follow each other only
# loosely. Two forces can run together between the occupations: the gender
# typing of the occupation, and the share of senior posts among its
# advertisements, since within an occupation the wording turns less communal
# as the level rises (section 28). The section puts the three quantities side
# by side and describes how they move together. It tests none of the
# hypotheses, and with 13 occupations every correlation here rests on 13
# points.
# The share of women among those employed is the BLS value Figure A1 and
# Table A1 carry, read from the file they read. The share of women in the
# biographies is the one of the pool of Study 2. The wording, the densities
# and the level are the ones of section 28, which this section
# follows. Part A is written out as Table A2 of the appendix.
from scipy.stats import pearsonr   # section 2 imports Spearman alone

OCC_META_PATH = Path("config/occupation_metadata_27.csv")
N_PERM_W42, SEED_W42 = 10000, 42   # permutations and seed of this section
SENIOR_FROM = 4   # rank of Mid-Senior level in LEVELS_W38, which runs 1 to 6

_meta_w42 = pd.read_csv(OCC_META_PATH)
_meta_w42 = _meta_w42[_meta_w42["in_study2"] == "yes"]
if sorted(_meta_w42["occ"]) != sorted(KEPT_13):
    raise AssertionError(
        f"{OCC_META_PATH} marks {len(_meta_w42)} occupations as part of "
        f"Study 2; the 13 of this notebook are the ones the file has to "
        f"cover.")
_bls_w42 = _meta_w42.set_index("occ")["pct_women_bls_2025"]

# The share of women in the pool: one row per biography, so the communal arm
# of the pool table carries each of them once.
_bios_w42 = df_pool[df_pool["arm"] == "communion"]
_n_bios_w42 = _bios_w42.groupby("occ").size()
if int(_n_bios_w42.min()) != int(_n_bios_w42.max()):
    raise AssertionError(
        f"The pool holds between {int(_n_bios_w42.min())} and "
        f"{int(_n_bios_w42.max())} biographies per occupation; the share of "
        f"women is read per occupation, so the sizes have to agree.")
_women_w42 = _bios_w42.groupby("occ")["female"].mean() * 100

# The seniority of the posts: the rank of section 28, and the share of the
# advertisements that name a level from Mid-Senior level upwards. The
# advertisements without a level carry no rank and stay out of that share.
_stated_w42 = df_levels.assign(
    level_rank=df_levels["level"].map(dict(LEVELS_W38)))
_stated_w42 = _stated_w42[_stated_w42["level_rank"].notna()]

_rows_w42 = []
for occ in KEPT_13:
    _ads = df_levels[df_levels["occ"] == occ]
    _with_level = _stated_w42[_stated_w42["occ"] == occ]
    _rows_w42.append({
        "occ": occ,
        "n_ads": len(_ads),
        "bls_women": float(_bls_w42.loc[occ]),
        "bio_women": float(_women_w42.loc[occ]),
        "balance": float(_ads["balance"].mean()),
        "agentic": float((_ads["balance"] < 0).mean()),
        "comm_dens": float(_ads["comm_dens"].mean()),
        "ag_dens": float(_ads["ag_dens"].mean()),
        "senior": float((_with_level["level_rank"] >= SENIOR_FROM).mean()),
        "n_level": len(_with_level)})
df_occ_w42 = pd.DataFrame(_rows_w42).sort_values("bls_women")

print(f"A. The {len(df_occ_w42)} occupations, by the share of women the BLS "
      f"reports, {len(_bios_w42)} biographies and {len(df_levels)} "
      f"advertisements")
print(f"  {'occ':<20} {'n_ads':>6} {'bls_women':>10} {'bio_women':>10} "
      f"{'balance':>9} {'agentic':>8} {'comm_dens':>10} {'ag_dens':>8} "
      f"{'senior':>7} {'n_level':>8}")
print("  " + "-" * 101)
for _row in df_occ_w42.itertuples():
    print(f"  {_row.occ:<20} {_row.n_ads:>6} {_row.bls_women:>10.1f} "
          f"{_row.bio_women:>10.1f} {_row.balance:>+9.3f} "
          f"{_row.agentic:>8.3f} {_row.comm_dens:>10.3f} "
          f"{_row.ag_dens:>8.3f} {_row.senior:>7.3f} {_row.n_level:>8}")

# Table A2 of the appendix: the columns of the table above, without the two
# densities, which the print carries for Section 3.3 alone. The values are the
# ones printed, rounded to the places the table shows; nothing is computed
# again.
TABLE_A2_PATH = Path("outputs/tables/tableA2_ads_by_occupation.csv")
TABLE_A2_PLACES = {"bls_women": 1, "bio_women": 1, "balance": 3,
                   "agentic": 3, "senior": 3}
TABLE_A2_PATH.parent.mkdir(parents=True, exist_ok=True)
df_occ_w42[["occ", "n_ads", "bls_women", "bio_women", "balance", "agentic",
            "senior", "n_level"]].round(TABLE_A2_PLACES).to_csv(
    TABLE_A2_PATH, index=False)
print(f"  written: {TABLE_A2_PATH}, {len(df_occ_w42)} rows")


def _spearman_perm_w42(x, y, n_perm=N_PERM_W42, seed=SEED_W42):
    """Spearman of two values per occupation, with a two-sided permutation
    p-value. The 13 occupations are one stratum, so one of the two series is
    permuted against the other; the count follows the convention of
    study2_core.strat_perm_p, (1 + hits) / (n_perm + 1)."""
    x, y = np.asarray(x, float), np.asarray(y, float)
    _rho = float(spearmanr(x, y)[0])
    _rng = np.random.default_rng(seed)
    _hits = 0
    for _ in range(n_perm):
        _drawn = float(spearmanr(_rng.permutation(x), y)[0])
        _hits += abs(_drawn) >= abs(_rho) - 1e-12
    return _rho, (1 + _hits) / (n_perm + 1)


PAIRS_W42 = (
    ("balance_bls", "mean balance", "balance", "BLS share of women",
     "bls_women"),
    ("balance_bio", "mean balance", "balance", "women in the biographies",
     "bio_women"),
    ("balance_senior", "mean balance", "balance", "share of senior posts",
     "senior"),
    ("bls_senior", "BLS share of women", "bls_women", "share of senior posts",
     "senior"),
)
print()
print(f"B. Over the {len(df_occ_w42)} occupations: Spearman, with p from "
      f"{N_PERM_W42} permutations, seed {SEED_W42}")
print(f"  {'quantity':<20} {'against':<26} {'rho':>8} {'p':>8}")
print("  " + "-" * 64)
_found_w42 = {}
for _key, _name_x, _column_x, _name_y, _column_y in PAIRS_W42:
    _found_w42[_key] = _spearman_perm_w42(df_occ_w42[_column_x],
                                          df_occ_w42[_column_y])
    print(f"  {_name_x:<20} {_name_y:<26} {_found_w42[_key][0]:>+8.3f} "
          f"{_found_w42[_key][1]:>8.4f}")

# The first pair as a product-moment correlation as well: the write-up names
# a Pearson r, and the line below prints it next to the rank measure.
_r_w42, _p_pearson_w42 = pearsonr(df_occ_w42["balance"].to_numpy(float),
                                  df_occ_w42["bls_women"].to_numpy(float))
print(f"  Pearson of the mean balance with the BLS share of women: "
      f"r {float(_r_w42):+.3f}, p {float(_p_pearson_w42):.4f} from the t "
      f"distribution with {len(df_occ_w42) - 2} degrees of freedom")

print()
print(f"C. Between the {len(df_occ_w42)} occupations the wording moves with "
      f"the share of women at rho {_found_w42['balance_bls'][0]:+.3f} "
      f"(p {_found_w42['balance_bls'][1]:.4f}) and with the share of senior "
      f"posts at rho {_found_w42['balance_senior'][0]:+.3f} "
      f"(p {_found_w42['balance_senior'][1]:.4f}), while those two carry rho "
      f"{_found_w42['bls_senior'][0]:+.3f} "
      f"(p {_found_w42['bls_senior'][1]:.4f}) with each other; on "
      f"{len(df_occ_w42)} points neither force can be held apart from the "
      f"other, and the share of women alone leaves the wording of an "
      f"occupation open.")
```

## 30. Figure 1: the research model of the two studies

- **Question:** none; the figure draws the research model the thesis follows.
- **Expectation:** none.
- **Status:** revised on 2026-11-18: arrow label, box texts, colours. The
  arrow carries the finding of Study 1 above it and what Study 2 does with
  that finding below it; the two outer frames are blue and every box inside
  them is grey. Revised on 2026-11-15: observational analysis added, style of
  Figure 4. Study 1 stands on the left, the communal language it measures
  leads to Study 2, and Study 2 holds the two branches the thesis reports.
  The single steps of the experiment are not repeated; Figure 4 draws them.
  No data are read: every box and every line is written out in the cell, the
  widths are measured with the renderer so that no line runs past its box,
  the label of the frame and the first line of the experiment run over two
  lines because on one they would leave the arrow no room, and the cell
  stops unless the fork into the two branches is symmetric and every line
  keeps its margin.
- **Unit and test:** none, a figure.
- **Output:** `outputs/figures/fig1_framework.png`.

```python
# Figure 1: the research model of the two studies (schematic, no data). Study 1
# stands on the left, the finding it reports leads to Study 2 on the right as
# the treatment of the experiment, and Study 2 holds the two branches the
# thesis reports: the counterfactual experiment and the observational analysis
# of the same rankings. The single steps of the experiment are not repeated
# here; Figure 4 draws them.
# The style is the one of Figure 4 in section 18: the same serif, the same
# colours, line widths and rounded boxes, the full text width of 16 cm
# (6.30 in) at 300 dpi, and boxes only as high as their text needs. The two
# outer frames, Study 1 and Study 2, carry a blue outline; every box inside
# them is grey, and nothing is filled blue. The values are repeated here with
# their own names, so that the cell stands on its own and nothing it binds
# reaches the cell of Figure 4. The source is ASCII.
# Writes outputs/figures/fig1_framework.png at 300 dpi.
from matplotlib.patches import FancyBboxPatch as _Box_FW
import matplotlib.pyplot as plt

FIG_FRAMEWORK_PNG = Path("outputs/figures/fig1_framework.png")
INK_FW, GREY_FW, LIGHT_FW = "#1a1a1a", "#52514e", "#b5b4ae"
FILL_FW, BLUE_FW = "#f3f2ee", "#2a78d6"
W_IN_FW = 6.30          # full text width, 16 cm
M_FW = 0.06             # inner margin of a box, left and right, in inches
PAD_FW = 0.065          # inner margin of the frame
EDGE_FW = 0.01          # outer margin of the figure
GAP_V_FW = 0.08         # between the two boxes of Study 2
FORK_BACK_FW = 0.12     # the fork sits this far before the frame
TITLE_FS_FW, BODY_FS_FW, LABEL_FS_FW = 6.0, 5.5, 6.3

plt.rcParams["font.family"] = "serif"
LINE_FW = BODY_FS_FW * 1.3 / 72             # height of one body line in inches
LABEL_LINE_FW = LABEL_FS_FW * 1.3 / 72      # height of one line of the label
# The arrow carries the finding of Study 1 above it and what Study 2 does with
# that finding below it.
ARROW_ABOVE_FW = "finding: women's bios more communal"
ARROW_BELOW_FW = "treatment of Study 2"
# The label of the frame runs over two lines. On one line it would be 3.48 in
# wide and leave the arrow too little room for the label above it; the wording
# is the one of the thesis, only broken.
FRAME_LABEL_FW = ("Study 2 (SRQ2): 13 occupations,",
                  "three AI-based screeners, real job ads")
HEAD_FW = 0.045 + len(FRAME_LABEL_FW) * LABEL_LINE_FW + 0.045
BOXES_FW = {
    "study1": ("Study 1 (SRQ1): bios of women and men",
               ["393,977 bios, 27 occupations",
                "four features of wording",
                "H1a, H1b: communal and agentic language",
                "H2a, H2b: framing of achievements"]),
    # The first line of the experiment box names what the experiment compares;
    # it runs over two lines, because on one it would be 2.57 in wide and no
    # longer fit beside the box of Study 1 and the arrow.
    "experiment": ("Counterfactual experiment",
                   ["baseline ranking vs. ranking after neutralising",
                    "communal markers",
                    "H3: change in rank",
                    "H4a: depends on the ad's communal balance",
                    "H4b: differs between women's and men's bios"]),
    "observation": ("Observational analysis",
                    ["relations in the baseline ranking of unedited bios",
                     "H4a (T2): style fit",
                     "H4b (T1): gender gap by the ad"]),
}


def _height_fw(key):
    """A box is as high as its title and its lines need."""
    return 0.20 + len(BOXES_FW[key][1]) * LINE_FW + 0.06


H_LEFT_FW = _height_fw("study1")
# The two boxes of Study 2 carry a different number of lines, so each one is as
# high as its own text needs; the fork below is placed between their middles
# and is symmetric whatever the two heights are.
H_EXP_FW = _height_fw("experiment")
H_OBS_FW = _height_fw("observation")
H_FRAME_FW = PAD_FW + H_EXP_FW + H_OBS_FW + GAP_V_FW + HEAD_FW
H_IN_FW = 0.02 + H_FRAME_FW + 0.02

fig_fw = plt.figure(figsize=(W_IN_FW, H_IN_FW), dpi=300)
ax_fw = fig_fw.add_axes([0, 0, 1, 1])
ax_fw.set_xlim(0, W_IN_FW)
ax_fw.set_ylim(0, H_IN_FW)
ax_fw.axis("off")
renderer_fw = fig_fw.canvas.get_renderer()


def _width_fw(text, size, bold=False):
    _t = ax_fw.text(0, 0, text, fontsize=size,
                 weight="bold" if bold else "normal")
    _w = _t.get_window_extent(renderer_fw).width / fig_fw.dpi
    _t.remove()
    return _w


def _needed_fw(key):
    """The width the title and the lines of a box ask for, without margins."""
    return max([_width_fw(BOXES_FW[key][0], TITLE_FS_FW, True)]
               + [_width_fw(_l, BODY_FS_FW) for _l in BOXES_FW[key][1]])


# The frame carries its label above the two boxes, and the boxes fill it, so
# the wider of the two decides how wide the frame is.
W_LEFT_FW = _needed_fw("study1") + 2 * M_FW
W_LABEL_FW = max(_width_fw(_l, LABEL_FS_FW, True) for _l in FRAME_LABEL_FW)
W_NEEDED_FW = max(W_LABEL_FW + 2 * M_FW,
                  max(_needed_fw(k) for k in ("experiment", "observation"))
                  + 2 * M_FW + 2 * PAD_FW)
# The arrow between the two columns has to carry the wider of its two labels,
# with the inner margin of a box left and right of it, and the frame then
# takes whatever width is left, so that the figure fills the text width.
_label_fw = max(_width_fw(ARROW_ABOVE_FW, BODY_FS_FW),
                _width_fw(ARROW_BELOW_FW, BODY_FS_FW))
GAP_MIN_FW = _label_fw + 2 * M_FW + FORK_BACK_FW
W_FRAME_FW = max(W_NEEDED_FW,
                 W_IN_FW - 2 * EDGE_FW - W_LEFT_FW - GAP_MIN_FW)
W_BOX_FW = W_FRAME_FW - 2 * PAD_FW
GAP_FW = W_IN_FW - 2 * EDGE_FW - W_LEFT_FW - W_FRAME_FW

x_left_fw = EDGE_FW
x_frame_fw = x_left_fw + W_LEFT_FW + GAP_FW
x_fork_fw = x_frame_fw - FORK_BACK_FW
_stem_fw = x_fork_fw - (x_left_fw + W_LEFT_FW)
if _stem_fw < _label_fw + M_FW:
    raise AssertionError(
        f"The arrow is {_stem_fw:.3f} in long before the fork and does not "
        f"carry its two labels, the wider of which is {_label_fw:.3f} in; the "
        f"two columns leave too little room between them.")

y_frame_fw = 0.02
y_obs_fw = y_frame_fw + PAD_FW
y_exp_fw = y_obs_fw + H_OBS_FW + GAP_V_FW
y_mid_obs_fw = y_obs_fw + H_OBS_FW / 2
y_mid_exp_fw = y_exp_fw + H_EXP_FW / 2
y_arrow_fw = (y_mid_obs_fw + y_mid_exp_fw) / 2
y_left_fw = y_arrow_fw - H_LEFT_FW / 2
if abs((y_mid_exp_fw - y_arrow_fw) - (y_arrow_fw - y_mid_obs_fw)) > 1e-9:
    raise AssertionError("The fork into the two boxes is not symmetric.")
if y_exp_fw + H_EXP_FW > y_frame_fw + H_FRAME_FW - HEAD_FW + 1e-9:
    raise AssertionError(
        "The upper box of Study 2 reaches into the label of the frame.")
if y_left_fw < 0.01 or y_left_fw + H_LEFT_FW > H_IN_FW - 0.01:
    raise AssertionError(
        f"The box of Study 1 runs out of the figure: it sits at "
        f"{y_left_fw:.3f} in and is {H_LEFT_FW:.3f} in high.")


def _draw_box_fw(key, x, y, width, edge, fill, ink, body):
    ax_fw.add_patch(_Box_FW((x, y), width, _height_fw(key),
                         boxstyle="round,pad=0,rounding_size=0.05",
                         linewidth=0.8, edgecolor=edge, facecolor=fill,
                         zorder=2))
    _top = y + _height_fw(key)
    ax_fw.text(x + M_FW, _top - 0.05, BOXES_FW[key][0], fontsize=TITLE_FS_FW,
            color=ink, weight="bold", va="top", zorder=3)
    for _i, _line in enumerate(BOXES_FW[key][1]):
        ax_fw.text(x + M_FW, _top - 0.20 - _i * LINE_FW, _line,
                fontsize=BODY_FS_FW, color=body, va="top", zorder=3)


def _arrow_fw(x0, y0, x1, y1, colour=GREY_FW):
    ax_fw.annotate("", xy=(x1, y1), xytext=(x0, y0), zorder=4,
                arrowprops=dict(arrowstyle="-|>", lw=0.8, color=colour,
                                mutation_scale=7, shrinkA=0, shrinkB=0))


# The frame of Study 2, with its label inside it, as the branches of Figure 4.
# Study 1 and Study 2 are the two outer frames and carry the blue outline; the
# boxes inside Study 2 are grey, like the box of Study 1.
ax_fw.add_patch(_Box_FW((x_frame_fw, y_frame_fw), W_FRAME_FW, H_FRAME_FW,
                     boxstyle="round,pad=0,rounding_size=0.06",
                     linewidth=0.7, edgecolor=BLUE_FW, facecolor="white",
                     zorder=1))
for _i_fw, _line_fw in enumerate(FRAME_LABEL_FW):
    ax_fw.text(x_frame_fw + M_FW,
            y_frame_fw + H_FRAME_FW - 0.045 - _i_fw * LABEL_LINE_FW, _line_fw,
            color=GREY_FW, fontsize=LABEL_FS_FW, weight="bold", va="top",
            zorder=3)
_draw_box_fw("study1", x_left_fw, y_left_fw, W_LEFT_FW, BLUE_FW, FILL_FW,
             INK_FW, GREY_FW)
_draw_box_fw("experiment", x_frame_fw + PAD_FW, y_exp_fw, W_BOX_FW, LIGHT_FW,
             FILL_FW, INK_FW, GREY_FW)
_draw_box_fw("observation", x_frame_fw + PAD_FW, y_obs_fw, W_BOX_FW, LIGHT_FW,
             FILL_FW, INK_FW, GREY_FW)

# One line leaves the middle of Study 1, carries the finding above it and the
# treatment of Study 2 below it, splits before the frame and enters each branch
# at the middle of its box.
ax_fw.plot([x_left_fw + W_LEFT_FW + 0.01, x_fork_fw], [y_arrow_fw, y_arrow_fw],
        color=GREY_FW, lw=0.8, zorder=4, solid_capstyle="butt")
ax_fw.plot([x_fork_fw, x_fork_fw], [y_mid_obs_fw, y_mid_exp_fw], color=GREY_FW,
        lw=0.8, zorder=4, solid_capstyle="butt")
_arrow_fw(x_fork_fw, y_mid_exp_fw, x_frame_fw - 0.01, y_mid_exp_fw)
_arrow_fw(x_fork_fw, y_mid_obs_fw, x_frame_fw - 0.01, y_mid_obs_fw)
_x_label_fw = (x_left_fw + W_LEFT_FW + x_fork_fw) / 2
ax_fw.text(_x_label_fw, y_arrow_fw + 0.025, ARROW_ABOVE_FW,
        fontsize=BODY_FS_FW, color=GREY_FW, ha="center", va="bottom", zorder=4)
ax_fw.text(_x_label_fw, y_arrow_fw - 0.025, ARROW_BELOW_FW,
        fontsize=BODY_FS_FW, color=GREY_FW, ha="center", va="top", zorder=4)

# Control: every line and every title keeps at least the left margin on the
# right, inside its box and inside the frame.
_margins_fw = [W_LEFT_FW - (_needed_fw("study1") + M_FW)]
for _key in ("experiment", "observation"):
    _margins_fw.append(W_BOX_FW - (_needed_fw(_key) + M_FW))
_margins_fw.append(W_FRAME_FW - (W_LABEL_FW + M_FW))
_worst_fw = min(_margins_fw)
print(f"box width: Study 1 {W_LEFT_FW:.3f} in, Study 2 {W_BOX_FW:.3f} in "
      f"inside a frame of {W_FRAME_FW:.3f} in; gap: {GAP_FW:.3f} in; "
      f"smallest right margin: {_worst_fw:.3f} in "
      f"(left margin {M_FW:.3f} in)")
print(f"box height: experiment {H_EXP_FW:.3f} in, observational analysis "
      f"{H_OBS_FW:.3f} in; the fork reaches {y_mid_exp_fw - y_arrow_fw:.3f} in "
      f"up and the same down")
if _worst_fw < M_FW - 1e-6:
    raise AssertionError(
        f"A line runs to {_worst_fw:.3f} in of the right edge of its box, "
        f"less than the left margin of {M_FW:.3f} in.")

FIG_FRAMEWORK_PNG.parent.mkdir(parents=True, exist_ok=True)
fig_fw.savefig(FIG_FRAMEWORK_PNG, dpi=300)

print(f"written: {FIG_FRAMEWORK_PNG}")
```

## 31. Four edited biographies, before and after

- **Question:** none; the section shows what the substitution does to a
  biography and how far the three screeners differ on the same edit.
- **Expectation:** none.
- **Status:** added on 2026-11-16 for Appendix C. Nothing is computed: the
  texts are the ones the language model was given, which notebook 06 exported
  after the substitution; the changed places come from the protocol, and the
  change in percentile rank from the three tables the runs wrote. The examples
  carry no name of a person: a biography is only used when, apart from the
  first word of a sentence, it holds no capitalised word at all, and when both
  of its versions are ASCII. That is stricter than the cleaning of notebook 02,
  because a first name the name list does not know can survive it.
- **Unit and test:** biography, four of them, two of women and two of men in
  four occupations of the communal arm; no test.
- **Output:** none; the section prints the four examples of Table C4.

```python
# Four edited biographies, before and after, for Appendix C. Nothing is
# computed here: the texts, the protocol and the change per screener are read
# from the files the runs wrote. The texts are the ones Arm 3 was given, which
# notebook 06 exported after the substitution, so the edited text is the one
# every screener saw.
# The examples have to carry no name of a person. The biographies were cleaned
# in notebook 02, yet a first name that the name list does not know can remain,
# so the choice here is strict: a biography is only used when, apart from the
# first word of a sentence, it holds no capitalised word at all, and when both
# of its versions are ASCII, as every output of this notebook is. That rules out
# employers and places as well, which makes the examples plainer than an average
# biography; the counts below say how many biographies each step leaves.
import re
import textwrap

BIO_TEXT_FILE_W50 = Path(OUT_BASE) / "llm" / "kaggle_bios.parquet"
ARM2_POOL_FILE_W50 = Path(OUT_BASE) / "study2_delta_r5_pool_arm2.csv"
ARM3_DELTA_FILE_W50 = Path(OUT_BASE) / "arm3" / "main" / "study2_arm3_delta.csv"
ARM_W50 = "communion"       # the communal arm, as the appendix asks
MIN_SUBS_W50 = 3            # at least three changed places, so the edit is visible
N_EXAMPLES_W50 = 4          # two biographies of women, two of men
WRAP_W50 = 74

for _path_w50 in (BIO_TEXT_FILE_W50, ARM2_POOL_FILE_W50, ARM3_DELTA_FILE_W50):
    if not _path_w50.exists():
        raise FileNotFoundError(
            f"{_path_w50} not found. The examples read the texts of the "
            f"language model run and the change per biography of all three "
            f"screeners.")

_texts_w50 = (pd.read_parquet(BIO_TEXT_FILE_W50)
              .set_index(["occ", "bio_idx", "version"])["bio_text"])
_arm2_w50 = pd.read_csv(ARM2_POOL_FILE_W50)
_arm3_w50 = pd.read_csv(ARM3_DELTA_FILE_W50)
_log_w50 = pd.read_csv(PROTOCOL_FILE)

# Arm 1 carries the gate and the number of changed places; Arm 2 and Arm 3 add
# their own change in percentile points. Arm 3 rates 20 advertisements per
# occupation and the two embedding arms up to 250, so its records are fewer.
_cand_w50 = df_pool[(df_pool["arm"] == ARM_W50)
                    & (df_pool["sim_cp"] >= PRIMARY_GATE)
                    & (df_pool["n_subs"] >= MIN_SUBS_W50)]
print(f"changed biographies of the communal arm at the gate "
      f"{PRIMARY_GATE} with at least {MIN_SUBS_W50} changed places: "
      f"{len(_cand_w50)}")
_cand_w50 = _cand_w50.merge(
    _arm2_w50[_arm2_w50["arm"] == ARM_W50][["occ", "bio_idx", "delta"]]
    .rename(columns={"delta": "delta_arm2"}), on=["occ", "bio_idx"])
_a3_w50 = _arm3_w50[(_arm3_w50["arm"] == ARM_W50)
                    & (_arm3_w50["delta_pct"].notna())]
_cand_w50 = _cand_w50.merge(
    _a3_w50[["occ", "bio_idx", "delta_pct", "n_jds"]], on=["occ", "bio_idx"])
print(f"  of them with a change under all three screeners: {len(_cand_w50)}")

_SENTENCE_START_W50 = re.compile(r"(?:^|(?<=[.!?])\s+)")


def _capitalised_inside_w50(text):
    """Capitalised words that do not open a sentence and are not an acronym."""
    _starts = {_m.end() for _m in _SENTENCE_START_W50.finditer(text)}
    return [_m.group(0) for _m in re.finditer(r"[A-Za-z][\w'-]*", text)
            if _m.start() not in _starts and _m.group(0)[:1].isupper()
            and not _m.group(0).isupper()]


_rows_w50 = []
_n_cap_w50 = _n_ascii_w50 = 0
for _row_w50 in _cand_w50.itertuples():
    _key_w50 = (_row_w50.occ, _row_w50.bio_idx)
    _orig_w50 = _texts_w50.loc[(*_key_w50, "original")]
    _edit_w50 = _texts_w50.loc[(*_key_w50, ARM_W50)]
    if not (_orig_w50.isascii() and _edit_w50.isascii()):
        _n_ascii_w50 += 1
        continue
    if _capitalised_inside_w50(_orig_w50) or _capitalised_inside_w50(_edit_w50):
        _n_cap_w50 += 1
        continue
    _rows_w50.append(_row_w50)
print(f"  left out because a version is not ASCII: {_n_ascii_w50}; because a "
      f"capitalised word stands inside a sentence: {_n_cap_w50}")
print(f"  biographies the examples are drawn from: {len(_rows_w50)}")

# The choice: the first two occupations in the order of KEPT_13 that offer a
# biography of a woman, then the first two further occupations that offer one of
# a man, and within an occupation the lowest number. No random draw.
_chosen_w50, _used_w50 = [], set()
for _female_w50 in (1, 1, 0, 0):
    for _occ_w50 in KEPT_13:
        if _occ_w50 in _used_w50:
            continue
        _here_w50 = sorted((_r for _r in _rows_w50
                            if _r.occ == _occ_w50 and _r.female == _female_w50),
                           key=lambda _r: _r.bio_idx)
        if _here_w50:
            _chosen_w50.append(_here_w50[0])
            _used_w50.add(_occ_w50)
            break
if len(_chosen_w50) < N_EXAMPLES_W50:
    raise AssertionError(
        f"Only {len(_chosen_w50)} of {N_EXAMPLES_W50} examples were found: the "
        f"filters leave no biography of a woman or of a man in four different "
        f"occupations.")
print(f"chosen: {[(_r.occ, int(_r.bio_idx)) for _r in _chosen_w50]}")

for _i_w50, _row_w50 in enumerate(_chosen_w50, start=1):
    _key_w50 = (_row_w50.occ, _row_w50.bio_idx)
    _orig_w50 = _texts_w50.loc[(*_key_w50, "original")]
    _edit_w50 = _texts_w50.loc[(*_key_w50, ARM_W50)]
    _places_w50 = _log_w50[(_log_w50["occ"] == _row_w50.occ)
                           & (_log_w50["bio_idx"] == _row_w50.bio_idx)
                           & (_log_w50["arm"] == ARM_W50)
                           & (_log_w50["sample"] == "pool")]
    _replaced_w50 = _places_w50[_places_w50["action_taken"] == "replaced"]
    _deleted_w50 = _places_w50[_places_w50["action_taken"] == "deleted"]
    _untreated_w50 = _places_w50[_places_w50["action_taken"] == "untreated"]
    print()
    print(f"example {_i_w50}: {_row_w50.occ}, biography of a "
          f"{'woman' if _row_w50.female else 'man'}, number "
          f"{int(_row_w50.bio_idx)} of the pool")
    print(f"  content preservation {_row_w50.sim_cp:.4f}, changed places "
          f"{int(_row_w50.n_subs)}, words {len(_orig_w50.split())}")
    for _label_w50, _text_w50 in (("before", _orig_w50), ("after", _edit_w50)):
        print(textwrap.fill(_text_w50, WRAP_W50,
                            initial_indent=f"  {_label_w50:<6}: ",
                            subsequent_indent=" " * 10))
    print("  replaced: " + ", ".join(
        f"{_r.token} -> {_r.target} ({_r.rule_id}, {_r.word_class})"
        for _r in _replaced_w50.itertuples()))
    if len(_deleted_w50):
        print("  deleted : " + ", ".join(
            f"{_r.token} ({_r.rule_id}, {_r.word_class})"
            for _r in _deleted_w50.itertuples()))
    if len(_untreated_w50):
        print("  left in place: " + ", ".join(
            f"{_r.token} ({_r.reason_untreated})"
            for _r in _untreated_w50.itertuples()))
    print(f"  change in percentile rank: SBERT {_row_w50.delta:+.3f}, "
          f"Qwen3-Embedding {_row_w50.delta_arm2:+.3f}, language model "
          f"{_row_w50.delta_pct:+.3f}")
    print(f"  advertisements behind it: "
          f"{int((df_jd_level['occ'] == _row_w50.occ).sum())} under the two "
          f"embedding models, {int(_row_w50.n_jds)} under the language model")
print()
print("the change is the mean over the advertisements of the occupation; a plus "
      "means the edited")
print("biography stands higher in the ranking of that advertisement than the "
      "original one")
```

## 32. The lead per occupation against the share of women and seniority

- **Question:** Does the lead of the biographies of women over those of men in
  the original rankings follow the share of women in the occupation, or the
  seniority of the posts its advertisements offer?
- **Expectation:** none fixed in advance. Notebook 06 section 14.13 relates the
  same lead to the wording of the advertisement and finds a positive rank
  correlation over the occupations; this section keeps the lead and changes
  what it is read against.
- **Status:** added on 2026-11-17, descriptive. Nothing is rated or embedded
  again: the lead of Arm 1 is the column notebook 06 wrote into the
  advertisement table, the lead of Arm 2 is the one section 8 formed from the
  stored embeddings, and the lead of the language model is read from the table
  of notebook 07 in percentile points, so that all three stand on one scale.
  The two predictors are the ones of Table A2 in section 29.
- **Unit and test:** occupation, 13 of them. Spearman with a two-sided
  permutation p from 10,000 permutations with a fixed seed, for three screeners
  against three quantities. None of this was fixed in advance, and every
  correlation rests on 13 points.
- **Output:** none; the section prints the leads of Table E8.

```python
# The lead of the biographies of women over those of men per occupation, read
# against the share of women in the occupation and against the seniority of its
# advertisements. Notebook 06 section 14.13 relates the same lead to the wording
# of the advertisement; this section keeps the lead and changes what it is read
# against, for all three screeners. Nothing is rated or embedded again: Arm 1
# takes the lead per advertisement from the column the run wrote into the
# advertisement table, Arm 2 takes it from section 8, which formed it from the
# stored embeddings, and Arm 3 reads it from the table of notebook 07. The two
# predictors are the ones of Table A2 in section 29.
# The test is the rank correlation over the 13 occupations, with a two-sided
# permutation p from 10,000 permutations and the seed named below, the
# convention of section 29 and of study2_core.strat_perm_p. Thirteen points
# carry little, and none of these correlations was fixed in advance.
N_PERM_W59, SEED_W59 = 10000, 42
ARM3_GAP_FILE_W59 = Path(OUT_BASE) / "arm3" / "main" / "study2_arm3_gender_gap.csv"

if "_gap_arm2" not in globals():
    raise AssertionError(
        "The lead of Arm 2 per advertisement is missing; section 8 forms it and "
        "has to run before this section.")
if "df_occ_w42" not in globals():
    raise AssertionError(
        "The occupation table of section 29 is missing; it carries the share of "
        "women and the share of senior advertisements and has to run first.")
if "gap_f_minus_m" not in df_jd_level.columns:
    raise AssertionError(
        "The advertisement table has no column gap_f_minus_m, so the lead of "
        "Arm 1 cannot be read; notebook 06 writes it in section 14.13.")
if not ARM3_GAP_FILE_W59.exists():
    raise FileNotFoundError(
        f"{ARM3_GAP_FILE_W59} not found; notebook 07 writes the lead per "
        f"advertisement of the language model.")


def _spearman_perm_w59(x, y, n_perm=N_PERM_W59, seed=SEED_W59):
    """Spearman rho and a two-sided permutation p, as section 29 forms it."""
    x, y = np.asarray(x, float), np.asarray(y, float)
    _rho = float(spearmanr(x, y)[0])
    _rng = np.random.default_rng(seed)
    _hits = 0
    for _ in range(n_perm):
        _hits += abs(float(spearmanr(_rng.permutation(x), y)[0])) >= abs(_rho) - 1e-12
    return _rho, (1 + _hits) / (n_perm + 1)


# The lead per occupation, one mean per screener. Arm 1 and Arm 2 cover every
# advertisement of the occupation, Arm 3 the 20 it rated.
_arm3_gap_w59 = pd.read_csv(ARM3_GAP_FILE_W59)
_arm3_gap_w59 = _arm3_gap_w59[_arm3_gap_w59["measure"] == "percentile"]
if _arm3_gap_w59.empty:
    raise AssertionError(
        "The table of the language model holds no lead in percentile points, so "
        "the three screeners cannot be compared on one scale.")
_lead_w59 = {
    "SBERT": df_jd_level.groupby("occ")["gap_f_minus_m"].mean(),
    "Qwen3-Embedding": pd.Series({occ: float(np.mean(_gap_arm2[occ]))
                                  for occ in KEPT_13}),
    "language model": _arm3_gap_w59.groupby("occ")["gap_f_minus_m"].mean(),
}
_n_ads_w59 = {
    "SBERT": df_jd_level.groupby("occ").size(),
    "Qwen3-Embedding": pd.Series({occ: len(_gap_arm2[occ]) for occ in KEPT_13}),
    "language model": _arm3_gap_w59.groupby("occ").size(),
}
_occ_w59 = df_occ_w42.set_index("occ")
_predictors_w59 = {"share of women (BLS)": _occ_w59["bls_women"],
                   "share senior": _occ_w59["senior"],
                   "mean balance": _occ_w59["balance"]}

print(f"The lead of the biographies of women over those of men in the original "
      f"rankings, per occupation,")
print(f"in percentile points; {len(KEPT_13)} occupations, permutations "
      f"{N_PERM_W59}, seed {SEED_W59}")
print(f"\n  {'occ':<20} {'bls_women':>9} {'senior':>7} {'balance':>8}"
      + "".join(f" {_s:>18}" for _s in _lead_w59))
print("  " + "-" * 96)
for _occ in _occ_w59.index:
    print(f"  {_occ:<20} {_occ_w59.loc[_occ, 'bls_women']:>9.1f} "
          f"{_occ_w59.loc[_occ, 'senior']:>7.3f} "
          f"{_occ_w59.loc[_occ, 'balance']:>8.3f}"
          + "".join(f" {_lead_w59[_s][_occ]:>+13.3f} "
                    f"({int(_n_ads_w59[_s][_occ]):>3})" for _s in _lead_w59))

print(f"\n  {'screener':<18} {'read against':<22} {'rho':>8} {'p':>9}")
print("  " + "-" * 60)
_found_w59 = {}
for _screener, _series in _lead_w59.items():
    _lead_vector = [float(_series[_occ]) for _occ in _occ_w59.index]
    for _label, _column in _predictors_w59.items():
        _rho, _p = _spearman_perm_w59(_lead_vector,
                                      [float(_column[_occ])
                                       for _occ in _occ_w59.index])
        _found_w59[(_screener, _label)] = (_rho, _p)
        print(f"  {_screener:<18} {_label:<22} {_rho:>+8.3f} {_p:>9.4f}")

# Control: Arm 1 against the mean balance is the occupation-level correlation of
# notebook 06 section 14.13, which reports it with the p of scipy rather than a
# permutation p; both are printed so the two can be held against each other.
_rho_ctrl_w59 = spearmanr([float(_lead_w59["SBERT"][_occ])
                           for _occ in _occ_w59.index],
                          [float(_occ_w59.loc[_occ, "balance"])
                           for _occ in _occ_w59.index])
print(f"\n  control SBERT against the mean balance: rho "
      f"{float(_rho_ctrl_w59[0]):+.3f}, p of scipy "
      f"{float(_rho_ctrl_w59[1]):.4f}; notebook 06 section 14.13 prints "
      f"rho +0.560, p 0.0463")
if abs(float(_rho_ctrl_w59[0]) - 0.560) > 0.001:
    raise AssertionError(
        f"The lead against the balance gives rho {float(_rho_ctrl_w59[0]):+.3f} "
        f"here and +0.560 in notebook 06; the two do not rest on the same "
        f"numbers.")
print("  the p values are nominal; three screeners and three predictors are "
      "read here, and none")
print("  of these correlations was fixed in advance")
```

## 33. Who lands in the upper tenth of a ranking?

- **Question:** The sections above report the lead in percentile points. Read
  as a shortlist instead: how large is the share of the biographies of women
  among the upper tenth of an advertisement's ranking, and how many of the
  edited biographies does the communal edit carry into that tenth or out of
  it?
- **Expectation:** none fixed in advance. The share of women in the upper
  tenth follows from the lead the sections above report, and the section puts
  that lead on a scale a reader of a shortlist can picture.
- **Status:** added on 2026-11-18, descriptive, not fixed in advance. Nothing
  is embedded or rated again: Arm 1 is read from the embeddings notebook 06
  cached, Arm 2 from the embeddings section 2 loaded, and Arm 3 from the two
  tables notebook 07 wrote. The upper tenth is a cut this notebook draws; no
  screener of Study 2 applies one.
- **Unit and test:** advertisement; shares, no test. The upper tenth holds the
  places the rounded tenth of the pool gives: 25 of the 250 biographies for
  Arm 1 and Arm 2 in every occupation, and 9 or 10 of the 94 to 100
  biographies the language model rated. The shares are averaged over the
  advertisements of an occupation first and then over the occupations, so
  every occupation weighs the same, and they are reported over all
  advertisements and per fifth of the communal balance of the advertisement,
  cut as in section 12 and drawn in Figure 5 and Figure 6. The movement reads
  the biographies the communal manipulation changed at the gate of the main
  analysis, with the individual counterfactual of notebook 06: one biography
  is edited and the rest of the pool keeps its text.
- **Output:** `study2_top_decile.csv`, Table E9.

```python
# Who lands in the upper tenth of an advertisement's ranking? The sections
# above report the lead in percentile points; this one reads the same rankings
# as a shortlist. Nothing is embedded or rated again: Arm 1 comes from the
# embeddings notebook 06 cached, Arm 2 from the embeddings of notebook 06a that
# section 2 loaded, and Arm 3 from the two tables notebook 07 wrote. The fifths
# of the advertisements are the ones of section 12, so the rows below sit on
# the same cut as Figure 5 and Figure 6.
TOP_SHARE_W60 = 0.10
TOP_DECILE_FILE = Path(OUT_BASE) / "study2_top_decile.csv"
ARM3_PCT_BY_JD_W60 = Path(OUT_BASE) / "arm3" / "main" / "study2_arm3_pct_by_jd.csv"
SCREENERS_W60 = (("arm1", "SBERT"), ("arm2", "Qwen3-Embedding"),
                 ("arm3", "language model"))

for _name_w60, _where_w60 in (
        ("_ranks", "section 12, the ranks of Arm 1 and Arm 2 per advertisement"),
        ("_jd_parts", "section 12, the fifths of the advertisements"),
        ("_arm3_parts", "section 12, the fifths of the advertisements of Arm 3"),
        ("_arm3_ranks", "section 12, the ranks of Arm 3"),
        ("_emb_arm1_orig", "section 17.3, the cached embeddings of Arm 1"),
        ("_emb_arm1_manip", "section 17.3, the cached embeddings of Arm 1"),
        ("_emb_arm1_jd", "section 17.3, the cached embeddings of Arm 1"),
        ("df_arm2_delta_bio", "section 3, the change of Arm 2 per biography")):
    if _name_w60 not in globals():
        raise AssertionError(
            f"{_name_w60} is missing; it is formed in {_where_w60} and that "
            f"section has to run before this one.")
if not ARM3_PCT_BY_JD_W60.exists():
    raise FileNotFoundError(
        f"{ARM3_PCT_BY_JD_W60} not found; notebook 07 writes the rank of every "
        f"rated biography before and after the edit.")

# The biographies the communal manipulation changed and that pass the gate are
# the ones section 12 reads for panel (a); the movement below is read on them.
_edited_w60 = df_pool[(df_pool["sim_cp"] >= PRIMARY_GATE)
                      & (df_pool["n_subs"] >= 1)
                      & (df_pool["arm"] == "communion")]
_female_w60 = {occ: df_bio[df_bio["occ"] == occ].sort_values("bio_idx")
               ["female"].to_numpy(int) for occ in KEPT_13}


def _cut_w60(n_pool):
    """The upper tenth of a pool of n_pool biographies: how many places it
    holds, and the percentile rank a biography needs to hold one of them.
    A rank is (n - 1 - above) / (n - 1) * 100, where above counts the original
    biographies ahead of it, so the k-th place carries (n - k) / (n - 1) * 100
    both in the original ranking and after an edit of one biography."""
    _k = int(np.rint(TOP_SHARE_W60 * n_pool))
    if _k < 1:
        raise AssertionError(
            f"A pool of {n_pool} biographies leaves no place in the upper "
            f"tenth.")
    return _k, (n_pool - _k) / (n_pool - 1) * 100


# ---------------------------------------------------------------- the rankings
# Arm 1 and Arm 2 rank the whole pool against every advertisement of their
# occupation; Arm 3 rated about 100 biographies against 20 advertisements.
_orig_w60, _after_w60, _pool_female_w60, _cols_w60 = {}, {}, {}, {}
for _screener_w60, _label_w60 in SCREENERS_W60:
    _orig_w60[_screener_w60], _after_w60[_screener_w60] = {}, {}
    _pool_female_w60[_screener_w60], _cols_w60[_screener_w60] = {}, {}

_worst_w60 = {"arm1": 0.0, "arm2": 0.0, "arm3": 0.0}
for occ in KEPT_13:
    _idx_w60 = (_edited_w60[_edited_w60["occ"] == occ]["bio_idx"]
                .to_numpy(int))
    for _screener_w60, _orig_emb_w60, _manip_emb_w60, _jd_emb_w60 in (
            ("arm1", _emb_arm1_orig, _emb_arm1_manip, _emb_arm1_jd),
            ("arm2", emb_orig, emb_manip, emb_jd)):
        _so_w60 = _orig_emb_w60[occ] @ _jd_emb_w60[occ].T
        _sm_w60 = _manip_emb_w60[occ]["communion"] @ _jd_emb_w60[occ].T
        _pct_orig_w60 = study2_core.pool_pct_per_jd(_so_w60)
        _pct_after_w60 = study2_core.indiv_cf_pct_per_jd(_so_w60, _sm_w60)
        # Control: the rank of the originals is the one section 12 read, and
        # the change of the edited biographies is the one the tables carry.
        _worst_w60[_screener_w60] = max(
            _worst_w60[_screener_w60],
            float(np.abs(_pct_orig_w60 - _ranks[_screener_w60][occ]).max()))
        _here_w60 = (_pct_after_w60[_idx_w60] - _pct_orig_w60[_idx_w60]).mean(axis=1)
        if _screener_w60 == "arm1":
            _saved_w60 = (df_pool[(df_pool["occ"] == occ)
                                  & (df_pool["arm"] == "communion")]
                          .set_index("bio_idx").loc[_idx_w60, "delta"]
                          .to_numpy(float))
        else:
            _saved_w60 = (df_arm2_delta_bio[(df_arm2_delta_bio["occ"] == occ)
                                            & (df_arm2_delta_bio["arm"] == "communion")]
                          .set_index("bio_idx").loc[_idx_w60, "delta"]
                          .to_numpy(float))
        _worst_w60[_screener_w60] = max(
            _worst_w60[_screener_w60],
            float(np.abs(_here_w60 - _saved_w60).max()))
        _orig_w60[_screener_w60][occ] = _pct_orig_w60
        _after_w60[_screener_w60][occ] = _pct_after_w60
        _pool_female_w60[_screener_w60][occ] = _female_w60[occ]
        _cols_w60[_screener_w60][occ] = np.arange(_pct_orig_w60.shape[1])

_pct_by_jd_w60 = pd.read_csv(ARM3_PCT_BY_JD_W60)
_pct_by_jd_w60 = _pct_by_jd_w60[_pct_by_jd_w60["arm"] == "communion"].copy()
_pct_by_jd_w60["jd_idx"] = [int(str(_j).rsplit("jd", 1)[1])
                            for _j in _pct_by_jd_w60["jd_id"]]
_pct_by_jd_w60 = _pct_by_jd_w60.merge(_edited_w60[["occ", "bio_idx"]],
                                      on=["occ", "bio_idx"])
for occ in KEPT_13:
    _wide_w60 = (_arm3_ranks[_arm3_ranks["occ"] == occ]
                 .pivot(index="bio_idx", columns="jd_idx", values="pct")
                 .sort_index())
    _orig_w60["arm3"][occ] = _wide_w60.to_numpy(float)
    _cols_w60["arm3"][occ] = _wide_w60.columns.to_numpy(int)
    _pool_female_w60["arm3"][occ] = (
        _arm3_ranks[_arm3_ranks["occ"] == occ]
        .drop_duplicates("bio_idx").sort_values("bio_idx")["female"]
        .to_numpy(int))
    if np.isnan(_orig_w60["arm3"][occ]).any():
        raise AssertionError(
            f"{occ}: the table of the language model has no rank for every "
            f"pair of biography and advertisement.")
    # Control: the rank before the edit is the one of the table of originals.
    _sub_w60 = _pct_by_jd_w60[_pct_by_jd_w60["occ"] == occ]
    _row_w60 = _wide_w60.index.get_indexer(_sub_w60["bio_idx"])
    _col_w60 = _wide_w60.columns.get_indexer(_sub_w60["jd_idx"])
    if (_row_w60 < 0).any() or (_col_w60 < 0).any():
        raise AssertionError(
            f"{occ}: the two tables of notebook 07 do not cover the same "
            f"biographies and advertisements.")
    _worst_w60["arm3"] = max(_worst_w60["arm3"], float(np.abs(
        _orig_w60["arm3"][occ][_row_w60, _col_w60]
        - _sub_w60["pct_before"].to_numpy(float)).max()))

print(f"control the ranks against the tables they come from, largest "
      f"deviation (expected below 1e-09): "
      + ", ".join(f"{_s} {_worst_w60[_s]:.3g}" for _s, _ in SCREENERS_W60))
if max(_worst_w60.values()) > 1e-9:
    raise AssertionError(
        f"The ranks read here are not the ones of the tables: largest "
        f"deviation {max(_worst_w60.values()):.3g}.")

print()
print(f"The upper {TOP_SHARE_W60:.0%} of the ranking of one advertisement")
print(f"  {'':<20}{'Arm 1 and Arm 2':^41}{'language model':^41}")
print(f"  {'occ':<20}" + 2 * (f" {'bios ranked':>11} {'places':>7} "
                              f"{'rank needed':>12} {'ads':>5}  "))
print("  " + "-" * 102)
_cut_table_w60 = {}
for occ in KEPT_13:
    _line_w60 = f"  {occ:<20}"
    for _screener_w60 in ("arm1", "arm3"):
        _n_w60 = _orig_w60[_screener_w60][occ].shape[0]
        _k_w60, _thresh_w60 = _cut_w60(_n_w60)
        _line_w60 += (f" {_n_w60:>11} {_k_w60:>7} {_thresh_w60:>12.3f} "
                      f"{_orig_w60[_screener_w60][occ].shape[1]:>5}  ")
    print(_line_w60)
for _screener_w60, _ in SCREENERS_W60:
    _cut_table_w60[_screener_w60] = {
        occ: _cut_w60(_orig_w60[_screener_w60][occ].shape[0])
        for occ in KEPT_13}

# ------------------------------------------- who holds the places, per fifth
_fifth_of_ad_w60 = {}
for _screener_w60, _ in SCREENERS_W60:
    _parts_w60 = _arm3_parts if _screener_w60 == "arm3" else _jd_parts
    _fifth_of_ad_w60[_screener_w60] = (_parts_w60.set_index(["occ", "jd_idx"])
                                       ["jd_fifth"])

_rows_w60 = []
for _screener_w60, _label_w60 in SCREENERS_W60:
    for occ in KEPT_13:
        _k_w60, _thresh_w60 = _cut_table_w60[_screener_w60][occ]
        _matrix_w60 = _orig_w60[_screener_w60][occ]
        _flags_w60 = _pool_female_w60[_screener_w60][occ]
        if len(_flags_w60) != _matrix_w60.shape[0]:
            raise AssertionError(
                f"{occ}/{_screener_w60}: {len(_flags_w60)} biographies carry a "
                f"gender and {_matrix_w60.shape[0]} carry a rank.")
        _top_w60 = _matrix_w60 >= _thresh_w60 - 1e-9
        _share_top_w60 = (_flags_w60[:, None] * _top_w60).sum(axis=0) / _top_w60.sum(axis=0)
        for _position_w60, _jd_w60 in enumerate(_cols_w60[_screener_w60][occ]):
            _rows_w60.append({
                "block": "composition", "screener": _screener_w60,
                "encoder": _label_w60, "occ": occ, "jd_idx": int(_jd_w60),
                "jd_fifth": int(_fifth_of_ad_w60[_screener_w60]
                                .loc[(occ, int(_jd_w60))]),
                "n_pool": int(_matrix_w60.shape[0]), "n_top": _k_w60,
                "n_in_band": int(_top_w60[:, _position_w60].sum()),
                "share_women_pool": float(_flags_w60.mean()),
                "share_women_top": float(_share_top_w60[_position_w60])})
df_top_ads_w60 = pd.DataFrame(_rows_w60)
df_top_ads_w60["difference_pp"] = 100 * (df_top_ads_w60["share_women_top"]
                                         - df_top_ads_w60["share_women_pool"])

# ------------------------------- what the edit does to the places, per fifth
_move_rows_w60 = []
for _screener_w60, _label_w60 in SCREENERS_W60:
    for occ in KEPT_13:
        _k_w60, _thresh_w60 = _cut_table_w60[_screener_w60][occ]
        _columns_w60 = _cols_w60[_screener_w60][occ]
        if _screener_w60 == "arm3":
            _sub_w60 = _pct_by_jd_w60[_pct_by_jd_w60["occ"] == occ]
            _pairs_w60 = pd.DataFrame({
                "jd_idx": _sub_w60["jd_idx"].to_numpy(int),
                "before": _sub_w60["pct_before"].to_numpy(float) >= _thresh_w60 - 1e-9,
                "after": _sub_w60["pct_after"].to_numpy(float) >= _thresh_w60 - 1e-9})
        else:
            _idx_w60 = (_edited_w60[_edited_w60["occ"] == occ]["bio_idx"]
                        .to_numpy(int))
            _before_w60 = _orig_w60[_screener_w60][occ][_idx_w60] >= _thresh_w60 - 1e-9
            _after_block_w60 = _after_w60[_screener_w60][occ][_idx_w60] >= _thresh_w60 - 1e-9
            _pairs_w60 = pd.DataFrame({
                "jd_idx": np.repeat(_columns_w60, len(_idx_w60)),
                "before": _before_w60.T.reshape(-1),
                "after": _after_block_w60.T.reshape(-1)})
        _pairs_w60["into"] = (~_pairs_w60["before"]) & _pairs_w60["after"]
        _pairs_w60["out"] = _pairs_w60["before"] & (~_pairs_w60["after"])
        _per_ad_w60 = _pairs_w60.groupby("jd_idx").agg(
            n_edited=("before", "size"), share_into=("into", "mean"),
            share_out=("out", "mean"), share_before=("before", "mean"),
            share_after=("after", "mean")).reset_index()
        _per_ad_w60["occ"] = occ
        _per_ad_w60["block"] = "movement"
        _per_ad_w60["screener"] = _screener_w60
        _per_ad_w60["encoder"] = _label_w60
        _per_ad_w60["jd_fifth"] = [
            int(_fifth_of_ad_w60[_screener_w60].loc[(occ, int(_j))])
            for _j in _per_ad_w60["jd_idx"]]
        _move_rows_w60.append(_per_ad_w60)
df_move_ads_w60 = pd.concat(_move_rows_w60, ignore_index=True)
df_move_ads_w60["share_net"] = (df_move_ads_w60["share_into"]
                                - df_move_ads_w60["share_out"])


def _equal_w60(frame, columns, scope, extra=None):
    """One row per screener: the mean over the advertisements of an occupation
    first, then over the occupations, so every occupation weighs the same."""
    _by_occ = frame.groupby(["screener", "encoder", "occ"])[columns].mean()
    _counts = frame.groupby(["screener", "encoder", "occ"]).size()
    _out = _by_occ.groupby(["screener", "encoder"]).mean().reset_index()
    _out.insert(2, "scope", scope)
    _out.insert(3, "n_ads", _counts.groupby(["screener", "encoder"]).sum().to_numpy())
    _out.insert(4, "n_occupations",
                _counts.groupby(["screener", "encoder"]).size().to_numpy())
    if extra is not None:
        _out.insert(5, "n_pairs", extra)
    return _out


_composition_w60 = [_equal_w60(df_top_ads_w60,
                               ["share_women_pool", "share_women_top",
                                "difference_pp", "n_in_band"], "all")]
for _part_w60 in range(1, N_PARTS + 1):
    _composition_w60.append(_equal_w60(
        df_top_ads_w60[df_top_ads_w60["jd_fifth"] == _part_w60],
        ["share_women_pool", "share_women_top", "difference_pp", "n_in_band"],
        f"fifth {_part_w60}"))
df_top_decile_composition = pd.concat(_composition_w60, ignore_index=True)

_movement_w60 = []
for _scope_w60, _frame_w60 in [("all", df_move_ads_w60)] + [
        (f"fifth {_p}", df_move_ads_w60[df_move_ads_w60["jd_fifth"] == _p])
        for _p in range(1, N_PARTS + 1)]:
    _pairs_sum_w60 = (_frame_w60.groupby(["screener", "encoder"])["n_edited"]
                      .sum().to_numpy())
    _movement_w60.append(_equal_w60(
        _frame_w60, ["n_edited", "share_before", "share_after", "share_into",
                     "share_out", "share_net"], _scope_w60,
        extra=_pairs_sum_w60))
df_top_decile_movement = pd.concat(_movement_w60, ignore_index=True)

print()
print(f"Who holds the places in the upper tenth: the original biographies, "
      f"occupations weighted equally")
print(f"  {'screener':<16} {'ads':<9} {'ads':>5} {'occ':>4} "
      f"{'in the band':>12} {'women in the pool':>18} "
      f"{'women in the top':>17} {'difference, pp':>15}")
print("  " + "-" * 102)
for _, _row_w60 in df_top_decile_composition.iterrows():
    print(f"  {_row_w60['encoder']:<16} {_row_w60['scope']:<9} "
          f"{int(_row_w60['n_ads']):>5} {int(_row_w60['n_occupations']):>4} "
          f"{_row_w60['n_in_band']:>12.2f} "
          f"{_row_w60['share_women_pool']:>18.3f} "
          f"{_row_w60['share_women_top']:>17.3f} "
          f"{_row_w60['difference_pp']:>+15.2f}")

print()
print(f"What the communal edit does to those places: the edited biographies at "
      f"the gate of {PRIMARY_GATE}")
print(f"  {'screener':<16} {'ads':<9} {'ads':>5} {'pairs':>7} "
      f"{'in before':>10} {'in after':>9} {'moves in':>9} {'moves out':>10} "
      f"{'net':>8}")
print("  " + "-" * 92)
for _, _row_w60 in df_top_decile_movement.iterrows():
    print(f"  {_row_w60['encoder']:<16} {_row_w60['scope']:<9} "
          f"{int(_row_w60['n_ads']):>5} {int(_row_w60['n_pairs']):>7} "
          f"{_row_w60['share_before']:>10.4f} {_row_w60['share_after']:>9.4f} "
          f"{_row_w60['share_into']:>9.4f} {_row_w60['share_out']:>10.4f} "
          f"{_row_w60['share_net']:>+8.4f}")

df_top_decile = pd.concat(
    [df_top_decile_composition.assign(block="composition"),
     df_top_decile_movement.assign(block="movement")], ignore_index=True)
df_top_decile = df_top_decile[
    ["block", "screener", "encoder", "scope", "n_ads", "n_occupations",
     "n_pairs", "n_in_band", "share_women_pool", "share_women_top",
     "difference_pp", "n_edited", "share_before", "share_after", "share_into",
     "share_out", "share_net"]]
df_top_decile.to_csv(TOP_DECILE_FILE, index=False, encoding="utf-8")
print()
print(f"written: {TOP_DECILE_FILE} ({len(df_top_decile)} rows)")
print("  the shares are descriptive; no direction was fixed in advance for "
      "any of them,")
print("  and the upper tenth is a cut this notebook draws, not one the "
      "screeners apply")
```

## 34. Do the title keywords reach into neighbouring occupations?

- **Question:** Notebook 06 assigns an advertisement to an occupation by
  keywords in its title. How many of the drawn advertisements name a
  neighbouring occupation in their title, and how many also fit the
  keywords of another occupation? The neighbours the section counts are a
  solutions architect for architect, a physician assistant for physician, a
  dental hygienist for dentist, a surgical technologist for surgeon, a
  physical therapist for psychologist, a nursing assistant for nurse, and
  the word counsel, which stands for a lawyer as well as for an adviser.
- **Expectation:** none fixed in advance. The keywords were fixed before the
  run and are not changed here; the section counts what they caught.
- **Status:** added on 2026-11-18, descriptive check of the assignment; the
  keywords were fixed before the run and are not changed here. The
  draw of notebook 06 is repeated with the keywords, the seed and the cap of
  section 17.2, as section 28 repeats it for the seniority level, and only the
  title is read from the rows that draw keeps. No advertisement is dropped, no
  text is embedded, and no result of Study 2 changes.
- **Unit and test:** advertisement; counts and shares, no test. A pattern is
  matched as a whole word and without regard to case, so counselor is not a
  hit for the keyword counsel of attorney; counselor is a keyword of
  psychologist. An advertisement whose title matches several keyword lists is
  assigned to the last list that matches, in the order of the lists, so it
  appears in one occupation only; part 4 counts how often that happened.
- **Output:** `study2_jd_title_check.csv`.

```python
# Do the title keywords of notebook 06 pull in neighbouring occupations? The
# draw of notebook 06 is repeated here with the keywords, the seed and the cap
# of section 17.2, as section 28 repeats it for the seniority level, and the
# title is read from the rows that draw keeps. Only the titles are read; no
# text is embedded and no advertisement is drawn anew. The patterns below name
# the neighbouring occupations the section counts; they are read
# as whole words, so "counselor" is not a hit for "counsel".
import re as _re_w66
from collections import Counter as _Counter_w66

TITLE_CHECK_FILE = Path(OUT_BASE) / "study2_jd_title_check.csv"
TOP_TITLES_W66 = 10
NEIGHBOURS_W66 = {
    "architect": ["solution architect", "solutions architect",
                  "cloud architect", "data architect", "software architect",
                  "enterprise architect", "security architect",
                  "network architect", "technical architect",
                  "systems architect", "it architect"],
    "physician": ["physician assistant", "physician associate"],
    "dentist": ["dental hygienist", "dental assistant", "dental technician",
                "dental office", "dental receptionist"],
    "surgeon": ["surgical tech", "surgical technologist",
                "surgical technician", "surgical assistant",
                "surgical nurse"],
    "psychologist": ["physical therapist", "occupational therapist",
                     "speech therapist", "massage therapist",
                     "respiratory therapist"],
    "nurse": ["nursing assistant", "certified nursing assistant", "cna",
              "nursing aide"],
}
# The word to look at for attorney is counsel itself, which stands
# both for a lawyer and for an adviser, so the titles are printed instead of
# matched against a list.
COUNSEL_W66 = r"\bcounsel\b"
# Thirty of the 2,975 titles carry a typographic character. They are mapped to
# ASCII here, so that what this cell prints and writes is ASCII; the mapping
# touches punctuation only and no word.
ASCII_TITLE_W66 = {chr(0x2013): "-", chr(0x2014): "-", chr(0x2012): "-",
                   chr(0x2010): "-", chr(0x2018): "'", chr(0x2019): "'",
                   chr(0x201c): '"', chr(0x201d): '"', chr(0x00a0): " ",
                   chr(0x2192): "->", chr(0x2026): "...", chr(0x00ad): ""}

for _name_w66 in ("OCC_KEYWORDS", "N_JDS_FULL", "RANDOM_STATE",
                  "JD_RAW_PATH"):
    if _name_w66 not in globals():
        raise AssertionError(
            f"{_name_w66} is missing; section 17.2 sets the keywords, the cap "
            f"and the seed of the draw and has to run before this section.")


def _ascii_title_w66(text):
    """A title in ASCII: the typographic characters above are mapped, and a
    character that is left over is dropped."""
    for _bad, _good in ASCII_TITLE_W66.items():
        text = text.replace(_bad, _good)
    return text.encode("ascii", "ignore").decode("ascii").strip()


# The draw of notebook 06, repeated as in section 28: the column of the text
# alone decides which rows are dropped, so reading the title cannot change
# which advertisements the draw returns.
_raw_w66 = pd.read_csv(JD_RAW_PATH, usecols=["title", "description"],
                       low_memory=False)
_title_lower_w66 = _raw_w66["title"].fillna("").str.lower()
_raw_w66["occupation"] = None
_matches_w66 = {}
for _occ_w66, _patterns_w66 in OCC_KEYWORDS.items():
    _hit_w66 = _title_lower_w66.str.contains("|".join(_patterns_w66),
                                             regex=True, na=False)
    _matches_w66[_occ_w66] = _hit_w66
    _raw_w66.loc[_hit_w66, "occupation"] = _occ_w66
_matched_w66 = _raw_w66[_raw_w66["occupation"].notna()]
_rows_w66 = []
for occ in KEPT_13:
    _sub_w66 = _matched_w66[_matched_w66["occupation"] == occ][["description"]]
    _sub_w66 = _sub_w66.dropna()
    _drawn_w66 = _sub_w66.sample(min(N_JDS_FULL, len(_sub_w66)),
                                 random_state=RANDOM_STATE)
    for _jd_idx_w66, (_index_w66, _row_w66) in enumerate(
            _drawn_w66.iterrows()):
        _title_w66 = str(_matched_w66.at[_index_w66, "title"])
        _also_w66 = [_other for _other in OCC_KEYWORDS
                     if _other != occ and bool(_matches_w66[_other][_index_w66])]
        _rows_w66.append({
            "occ": occ, "jd_idx": _jd_idx_w66,
            "title": _ascii_title_w66(_title_w66),
            "title_mapped": any(ord(_ch_w66) > 127 for _ch_w66 in _title_w66),
            "n_words_here": max(len(str(_row_w66["description"]).split()), 1),
            "also_matches": ", ".join(sorted(_also_w66))})
df_titles_w66 = pd.DataFrame(_rows_w66)
df_titles_w66["title_lower"] = df_titles_w66["title"].str.lower()
del _raw_w66, _matched_w66, _matches_w66

# Control: these are the advertisements notebook 06 counted, so every token
# count has to be the one of study2_jd_eda.csv.
_check_w66 = df_titles_w66.merge(df_jd_eda[["occ", "jd_idx", "n_words"]],
                                 on=["occ", "jd_idx"], how="outer")
_bad_w66 = int((_check_w66["n_words_here"] != _check_w66["n_words"]).sum())
print(f"advertisements drawn: {len(df_titles_w66)}, token counts that differ "
      f"from study2_jd_eda.csv: {_bad_w66} (expected 0)")
if _bad_w66 or len(df_titles_w66) != len(df_jd_eda):
    raise AssertionError(
        f"{_bad_w66} advertisements differ from the ones notebook 06 counted "
        f"and the draw holds {len(df_titles_w66)} of {len(df_jd_eda)}, so the "
        f"titles do not belong to the advertisements of the run.")
print(f"titles that held a character above 127 and were mapped to ASCII: "
      f"{int(df_titles_w66['title_mapped'].sum())} of "
      f"{len(df_titles_w66)}; punctuation only, no word was changed")

_rows_out_w66 = []

# ----------------------------------------- 1. the neighbouring occupations
print()
print(f"1. Titles that name a neighbouring occupation")
print(f"  {'occ':<16} {'pattern':<30} {'n_ads':>6} {'hits':>6} "
      f"{'share':>8}")
print("  " + "-" * 70)
for occ, _patterns_w66 in NEIGHBOURS_W66.items():
    _here_w66 = df_titles_w66[df_titles_w66["occ"] == occ]
    _any_w66 = pd.Series(False, index=_here_w66.index)
    for _pattern_w66 in _patterns_w66:
        _hit_w66 = _here_w66["title_lower"].str.contains(
            r"\b" + _re_w66.escape(_pattern_w66) + r"\b", regex=True)
        _any_w66 = _any_w66 | _hit_w66
        print(f"  {occ:<16} {_pattern_w66:<30} {len(_here_w66):>6} "
              f"{int(_hit_w66.sum()):>6} {float(_hit_w66.mean()):>8.3f}")
        _rows_out_w66.append({"block": "pattern", "occ": occ,
                              "item": _pattern_w66, "n_ads": len(_here_w66),
                              "n": int(_hit_w66.sum()),
                              "share": float(_hit_w66.mean())})
    print(f"  {occ:<16} {'any of them':<30} {len(_here_w66):>6} "
          f"{int(_any_w66.sum()):>6} {float(_any_w66.mean()):>8.3f}")
    print("  " + "-" * 70)
    _rows_out_w66.append({"block": "any pattern", "occ": occ,
                          "item": "any of the patterns",
                          "n_ads": len(_here_w66), "n": int(_any_w66.sum()),
                          "share": float(_any_w66.mean())})

# ------------------------------------------------- 2. counsel for attorney
_att_w66 = df_titles_w66[df_titles_w66["occ"] == "attorney"]
_counsel_w66 = _att_w66[_att_w66["title_lower"].str.contains(COUNSEL_W66,
                                                             regex=True)]
print()
print(f"2. The word counsel in the titles of attorney: "
      f"{len(_counsel_w66)} of {len(_att_w66)} advertisements "
      f"({len(_counsel_w66) / len(_att_w66):.3f}). The keyword is the whole "
      f"word,")
print(f"   so counselor is not a hit for attorney; it is a keyword of "
      f"psychologist.")
print(f"  {'rank':>4} {'n':>4}  title")
print("  " + "-" * 70)
for _rank_w66, (_title_w66, _n_w66) in enumerate(
        _Counter_w66(_counsel_w66["title"]).most_common(TOP_TITLES_W66), 1):
    print(f"  {_rank_w66:>4} {_n_w66:>4}  {_title_w66[:62]}")
    _rows_out_w66.append({"block": "counsel", "occ": "attorney",
                          "item": _title_w66, "n_ads": len(_att_w66),
                          "n": _n_w66, "share": _n_w66 / len(_att_w66),
                          "rank": _rank_w66})

# ------------------------------------- 3. the most frequent titles per job
print()
print(f"3. The {TOP_TITLES_W66} most frequent titles of the drawn "
      f"advertisements, per occupation")
for occ in KEPT_13:
    _here_w66 = df_titles_w66[df_titles_w66["occ"] == occ]
    print()
    print(f"  {occ} ({len(_here_w66)} advertisements)")
    for _rank_w66, (_title_w66, _n_w66) in enumerate(
            _Counter_w66(_here_w66["title"]).most_common(TOP_TITLES_W66), 1):
        print(f"  {_rank_w66:>4} {_n_w66:>4}  {_title_w66[:66]}")
        _rows_out_w66.append({"block": "title", "occ": occ,
                              "item": _title_w66, "n_ads": len(_here_w66),
                              "n": _n_w66, "share": _n_w66 / len(_here_w66),
                              "rank": _rank_w66})

# -------------------------------- 4. titles that match a second occupation
# An advertisement whose title matches several keyword lists is assigned to
# the last list that matches, in the order of OCC_KEYWORDS, so it appears in
# one occupation only. This part counts how often that happened.
print()
print(f"4. Drawn advertisements whose title also matches the keywords of "
      f"another occupation")
print(f"  {'occ':<18} {'n_ads':>6} {'also matched':>13} {'share':>8}  "
      f"the other occupations")
print("  " + "-" * 86)
for occ in KEPT_13:
    _here_w66 = df_titles_w66[df_titles_w66["occ"] == occ]
    _double_w66 = _here_w66[_here_w66["also_matches"] != ""]
    _others_w66 = _Counter_w66(
        _other for _entry in _double_w66["also_matches"]
        for _other in _entry.split(", "))
    print(f"  {occ:<18} {len(_here_w66):>6} {len(_double_w66):>13} "
          f"{len(_double_w66) / len(_here_w66):>8.3f}  "
          + ", ".join(f"{_o} {_n}" for _o, _n in _others_w66.most_common(3)))
    _rows_out_w66.append({"block": "second occupation", "occ": occ,
                          "item": ", ".join(f"{_o} {_n}" for _o, _n
                                            in sorted(_others_w66.items())),
                          "n_ads": len(_here_w66), "n": len(_double_w66),
                          "share": len(_double_w66) / len(_here_w66)})

df_title_check = pd.DataFrame(_rows_out_w66)
df_title_check.to_csv(TITLE_CHECK_FILE, index=False, encoding="utf-8")
print()
print(f"written: {TITLE_CHECK_FILE} ({len(df_title_check)} rows)")
print("  the shares are descriptive; the draw of notebook 06 is not changed "
      "and no")
print("  advertisement is dropped here")
```

## 35. T1, T2 and the experiment without the advertisements of neighbouring occupations

- **Question:** Section 34 counts the advertisements whose title names a
  neighbouring occupation. Do the three results of Study 2 that rest on the
  wording of the advertisement hold when those advertisements are left out?
- **Expectation:** none fixed in advance. The directions of T1, T2 and H4a
  were fixed before the run and are not restated here; this section asks
  whether the same tests give the same picture on the rest.
- **Status:** added on 2026-11-18, robustness, not fixed in advance. The
  main analysis keeps every advertisement. The assignment of the
  advertisements is the one fixed before the run and is not changed here,
  and no advertisement is dropped from the run itself.
  Nothing is embedded or rated again; the ranks are the ones the sections
  above hold.
- **Unit and test:** an advertisement is marked when its title carries one of
  the patterns of section 34 part 1 or when it also matches the keywords of
  another occupation, as section 34 part 4 counts. The three tests are then
  run on the advertisements that are left, through the same functions as the
  main analysis: the slope of a biography over the advertisements of its
  occupation against the balance of the biography for T2, the gap between the
  biographies of women and of men per advertisement against the balance of the
  advertisement for T1, and the mean change of the communal arm per
  advertisement against the same balance for H4a. Spearman within the
  occupation with a pooled permutation test, 5,000 permutations, seed 42,
  two-sided. A point is a biography for T2 and an advertisement for T1 and
  H4a. An occupation enters only with at least ten points, so the language
  model, which was given twenty advertisements per occupation, loses
  occupations once the marked ones are gone; the printed count says how many
  are left.
- **Output:** `study2_neighbour_robustness.csv`.

```python
# Do T1, T2 and the experimental test of H4a hold when the advertisements
# whose titles reach into a neighbouring occupation are left out? Section 34
# marks those advertisements; this section repeats the three tests on the rest.
# The tests are the ones of the main analysis, run through the same functions:
# _delta_per_ad and _spearman_within of section 25, _gaps_per_ad of section 12,
# and study2_core.jd_language_test with the permutations of the run. Nothing is
# embedded or rated again, no advertisement is dropped from the run itself, and
# the main analysis keeps every advertisement.
ROBUST_FILE = Path(OUT_BASE) / "study2_neighbour_robustness.csv"
SCOPE_ALL_W69 = "all advertisements"
SCOPE_KEPT_W69 = "without neighbouring titles"

for _name_w69, _where_w69 in (
        ("df_titles_w66", "section 34, the titles of the drawn advertisements"),
        ("NEIGHBOURS_W66", "section 34, the patterns of the neighbouring jobs"),
        ("df_title_check", "section 34, the counts it wrote"),
        ("_delta_per_ad", "section 25, the change per advertisement"),
        ("_spearman_within", "section 25, the test of notebook 06 section 14.4"),
        ("_gaps_per_ad", "section 12, the gap per advertisement"),
        ("_jd_quantities", "section 25, the quantities of the advertisement"),
        ("T1_ORDER", "section 25, the order of the occupations"),
        ("_ranks", "section 12, the ranks of Arm 1 and Arm 2"),
        ("_arm3_ranks", "section 12, the ranks of Arm 3"),
        ("LABELS", "section 13, the name of each screener")):
    if _name_w69 not in globals():
        raise AssertionError(
            f"{_name_w69} is missing; it is formed in {_where_w69} and that "
            f"section has to run before this one.")

# ------------------------------------------------- 1. the marked advertisements
_marks_w69 = df_titles_w66[["occ", "jd_idx", "title_lower",
                            "also_matches"]].copy()
_pattern_w69 = pd.Series(False, index=_marks_w69.index)
for _occ_w69, _patterns_w69 in NEIGHBOURS_W66.items():
    _here_w69 = _marks_w69["occ"] == _occ_w69
    for _p_w69 in _patterns_w69:
        _pattern_w69 |= _here_w69 & _marks_w69["title_lower"].str.contains(
            r"\b" + _re_w66.escape(_p_w69) + r"\b", regex=True)
_marks_w69["by_pattern"] = _pattern_w69
_marks_w69["by_second_job"] = _marks_w69["also_matches"] != ""
_marks_w69["marked"] = _marks_w69["by_pattern"] | _marks_w69["by_second_job"]

# Control: the advertisements marked by a pattern are the ones section 34
# counted under "any of the patterns".
_saved_any_w69 = (df_title_check[df_title_check["block"] == "any pattern"]
                  .set_index("occ")["n"])
_worst_w69 = 0
for _occ_w69 in _saved_any_w69.index:
    _here_w69 = _marks_w69[_marks_w69["occ"] == _occ_w69]
    _worst_w69 = max(_worst_w69, abs(int(_here_w69["by_pattern"].sum())
                                     - int(_saved_any_w69[_occ_w69])))
print(f"control the marks against the counts of section 34: largest "
      f"difference {_worst_w69} (expected 0)")
if _worst_w69:
    raise AssertionError(
        f"The advertisements marked here are not the ones section 34 counted; "
        f"they differ by {_worst_w69} in one occupation.")

_keep_w69 = set(zip(_marks_w69.loc[~_marks_w69["marked"], "occ"],
                    _marks_w69.loc[~_marks_w69["marked"], "jd_idx"]))
print()
print(f"Advertisements whose title reaches into a neighbouring occupation")
print(f"  {'occ':<20} {'n_ads':>6} {'a pattern':>10} {'a second job':>13} "
      f"{'either':>8} {'left':>6} {'share left':>11}")
print("  " + "-" * 82)
for occ in KEPT_13 + ["all occupations"]:
    _here_w69 = (_marks_w69 if occ == "all occupations"
                 else _marks_w69[_marks_w69["occ"] == occ])
    print(f"  {occ:<20} {len(_here_w69):>6} "
          f"{int(_here_w69['by_pattern'].sum()):>10} "
          f"{int(_here_w69['by_second_job'].sum()):>13} "
          f"{int(_here_w69['marked'].sum()):>8} "
          f"{int((~_here_w69['marked']).sum()):>6} "
          f"{float((~_here_w69['marked']).mean()):>11.3f}")


def _only_kept_w69(frame):
    """The rows of a table per advertisement that are not marked."""
    _mask = [(_o, int(_j)) in _keep_w69
             for _o, _j in zip(frame["occ"], frame["jd_idx"])]
    return frame[pd.Series(_mask, index=frame.index)]


# ---------------------------------------------- 2. T2, one slope per biography
def _slopes_w69(screener, keep_only):
    """The slope of a biography over the advertisements of its occupation, as
    notebook 06 section 14.14 forms it, and the balance of that biography."""
    _slopes, _values = {}, {}
    for occ in KEPT_13:
        if screener == "arm3":
            _wide = (_arm3_ranks[_arm3_ranks["occ"] == occ]
                     .pivot(index="bio_idx", columns="jd_idx", values="pct")
                     .sort_index())
            _bio_idx = _wide.index.to_numpy(int)
            _columns = _wide.columns.to_numpy(int)
            _matrix = _wide.to_numpy(float)
        else:
            _bio_idx = (df_bio[df_bio["occ"] == occ].sort_values("bio_idx")
                        ["bio_idx"].to_numpy(int))
            _matrix = _ranks[screener][occ]
            _columns = np.arange(_matrix.shape[1])
        _wanted = ([_k for _k, _c in enumerate(_columns)
                    if (occ, int(_c)) in _keep_w69] if keep_only
                   else list(range(len(_columns))))
        if len(_wanted) < 3:
            continue
        _matrix = _matrix[:, _wanted]
        _balance = (df_jd_eda[df_jd_eda["occ"] == occ].set_index("jd_idx")
                    .loc[[int(_columns[_k]) for _k in _wanted], "balance"]
                    .to_numpy(float))
        _centred = _balance - _balance.mean()
        _denominator = float(_centred @ _centred)
        if _denominator < 1e-12:
            continue
        _slope = ((_matrix - _matrix.mean(axis=1, keepdims=True)) @ _centred
                  / _denominator)
        if abs(_slope.mean()) > 1e-7:
            raise AssertionError(
                f"control mean slope {occ}/{screener}: {_slope.mean():.2e} "
                f"(expected below 1e-7)")
        _slopes[occ] = _slope
        _values[occ] = (df_bio[df_bio["occ"] == occ].set_index("bio_idx")
                        .loc[_bio_idx, "bio_balance"].to_numpy(float))
    return _slopes, _values


# ------------------------------------------------------- 3. the three tests
def _run_w69(screener, test, keep_only):
    """One test for one screener, on all advertisements or on the rest."""
    if test == "T2":
        _slopes, _values = _slopes_w69(screener, keep_only)
        _, _pool = study2_core.jd_language_test(_slopes, _values,
                                                n_perm=N_PERM, seed=SEED)
        return _pool
    if test == "T1":
        _parts = _arm3_parts if screener == "arm3" else _jd_parts
        _frame = (_gaps_per_ad(screener, _parts)[["occ", "jd_idx", "c"]]
                  .rename(columns={"c": "y"}))
        _order = T1_ORDER[screener]
    else:
        _frame = _delta_per_ad(screener, "communion")
        _order = KEPT_13
    _frame = _frame.merge(_jd_quantities, on=["occ", "jd_idx"])
    if keep_only:
        _frame = _only_kept_w69(_frame)
    _, _pool = _spearman_within(_frame, "balance", _order)
    return _pool


# The values of the main analysis, read from the tables that carry them.
FIT_TESTS_FILE_W69 = Path(OUT_BASE) / "study2_fit_tests.csv"
if not FIT_TESTS_FILE_W69.exists():
    raise FileNotFoundError(
        f"{FIT_TESTS_FILE_W69} not found; it carries T2 of Arm 1 as notebook "
        f"06 published it.")
_fit_w69 = pd.read_csv(FIT_TESTS_FILE_W69)
_t2_arm1_w69 = _fit_w69[(_fit_w69["test_id"] == "T2_main")
                        & (_fit_w69["status"] == "primary")
                        & (_fit_w69["occ"] == "POOLED")]
_t2_arm2_w69 = df_arm2_fit_tests[(df_arm2_fit_tests["test"] == "T2")
                                 & (df_arm2_fit_tests["variant"] == "balance")]
_scale_w69 = df_scale[df_scale["quantity"] == "communal balance"].set_index(
    ["screener", "row"])
PUBLISHED_W69 = {}
for _screener_w69, _ in SCREENERS:
    for _row_w69, _test_w69 in (("T1", "T1"), ("communion", "H4a")):
        _key_w69 = (LABELS[_screener_w69], _row_w69)
        if _key_w69 in _scale_w69.index:
            PUBLISHED_W69[(_screener_w69, _test_w69)] = (
                float(_scale_w69.loc[_key_w69, "rho"]),
                float(_scale_w69.loc[_key_w69, "p"]))
PUBLISHED_W69[("arm1", "T2")] = (float(_t2_arm1_w69["rho"].iloc[0]),
                                 float(_t2_arm1_w69["p_rho"].iloc[0]))
PUBLISHED_W69[("arm2", "T2")] = (float(_t2_arm2_w69["rho"].iloc[0]),
                                 float(_t2_arm2_w69["p_rho"].iloc[0]))

TESTS_W69 = (("T2", "slope of a bio over the ads"),
             ("T1", "gap women minus men per ad"),
             ("H4a", "change of the communal arm per ad"))
_rows_w69 = []
print()
print(f"T1, T2 and the experimental test of H4a on the communal balance, "
      f"permutations {N_PERM}, seed {SEED}")
print(f"  {'test':<5} {'screener':<16} {'advertisements':<28} "
      f"{'points':>7} {'occ':>4} {'rho':>8} {'p':>8} "
      f"{'rho published':>14} {'p published':>12}")
print("  " + "-" * 110)
for _test_w69, _what_w69 in TESTS_W69:
    for _screener_w69, _ in SCREENERS:
        for _scope_w69, _keep_w69_flag in ((SCOPE_ALL_W69, False),
                                           (SCOPE_KEPT_W69, True)):
            _pool_w69 = _run_w69(_screener_w69, _test_w69, _keep_w69_flag)
            _published_w69 = PUBLISHED_W69.get((_screener_w69, _test_w69))
            _show_w69 = (_published_w69 if _published_w69 is not None
                         and not _keep_w69_flag else None)
            print(f"  {_test_w69:<5} {LABELS[_screener_w69]:<16} "
                  f"{_scope_w69:<28} {int(_pool_w69['n_jds']):>7} "
                  f"{int(_pool_w69['n_occ']):>4} "
                  f"{float(_pool_w69['rho']):>+8.3f} "
                  f"{float(_pool_w69['p_rho']):>8.4f} "
                  + (f"{_show_w69[0]:>+14.3f} {_show_w69[1]:>12.4f}"
                     if _show_w69 is not None else f"{'':>14} {'':>12}"))
            _rows_w69.append({
                "test": _test_w69, "quantity": _what_w69,
                "screener": _screener_w69, "encoder": LABELS[_screener_w69],
                "scope": _scope_w69, "n_points": int(_pool_w69["n_jds"]),
                "n_occupations": int(_pool_w69["n_occ"]),
                "rho": float(_pool_w69["rho"]),
                "p": float(_pool_w69["p_rho"]),
                "rho_published": (_published_w69[0] if _published_w69
                                  and not _keep_w69_flag else float("nan")),
                "p_published": (_published_w69[1] if _published_w69
                                and not _keep_w69_flag else float("nan"))})
    print("  " + "-" * 110)
print("  a point is a biography for T2, whose slope is formed over the "
      "advertisements that")
print("  are left, and an advertisement for T1 and for H4a")

# Control: on all advertisements the rows above have to be the published ones.
# T2 of the language model has no published counterpart: notebook 06 runs T2
# on the two encoders, and the language model rated twenty advertisements per
# occupation, too few for a slope the main analysis would report.
_worst_pub_w69 = 0.0
for _row_w69 in _rows_w69:
    if _row_w69["scope"] != SCOPE_ALL_W69 or pd.isna(_row_w69["rho_published"]):
        continue
    _worst_pub_w69 = max(_worst_pub_w69,
                         abs(_row_w69["rho"] - _row_w69["rho_published"]))
print(f"control the rows on all advertisements against the published values: "
      f"largest deviation {_worst_pub_w69:.3g} (expected below 1e-06)")
if _worst_pub_w69 > 1e-6:
    raise AssertionError(
        f"A test on all advertisements does not reproduce the value the main "
        f"analysis published; it deviates by {_worst_pub_w69:.3g}.")
print(f"T2 of the language model carries no published value: the main "
      f"analysis runs T2 on the two encoders only.")

# ------------------------------- 4. the lead of the biographies of women, SBERT
print()
print(f"The lead of the biographies of women over those of men under SBERT, "
      f"in percentile points")
print(f"  {'advertisements':<28} {'n_ads':>6} {'over the ads':>13} "
      f"{'occupations equal':>18}")
print("  " + "-" * 70)
_lead_rows_w69 = []
_gaps_all_w69 = _gaps_per_ad("arm1", _jd_parts)[["occ", "jd_idx", "c"]]
for _scope_w69, _frame_w69 in ((SCOPE_ALL_W69, _gaps_all_w69),
                               (SCOPE_KEPT_W69, _only_kept_w69(_gaps_all_w69))):
    _over_ads_w69 = float(_frame_w69["c"].mean())
    _equal_w69 = float(_frame_w69.groupby("occ")["c"].mean().mean())
    print(f"  {_scope_w69:<28} {len(_frame_w69):>6} {_over_ads_w69:>+13.3f} "
          f"{_equal_w69:>+18.3f}")
    _lead_rows_w69.append({"test": "lead", "quantity": "gap women minus men",
                           "screener": "arm1", "encoder": LABELS["arm1"],
                           "scope": _scope_w69, "n_points": len(_frame_w69),
                           "n_occupations": int(_frame_w69["occ"].nunique()),
                           "rho": _over_ads_w69, "p": float("nan"),
                           "rho_published": float("nan"),
                           "p_published": float("nan"),
                           "lead_occupations_equal": _equal_w69})

df_neighbour_robustness = pd.DataFrame(_rows_w69 + _lead_rows_w69)
df_neighbour_robustness.to_csv(ROBUST_FILE, index=False, encoding="utf-8")
print()
print(f"written: {ROBUST_FILE} ({len(df_neighbour_robustness)} rows)")
print("  the rows on the rest are a robustness check added after the results; no "
      "direction was fixed")
print("  for them, and the main analysis keeps every advertisement. In the row "
      "named lead, the column")
print("  rho holds the mean over the advertisements and is not a "
      "correlation.")
```
